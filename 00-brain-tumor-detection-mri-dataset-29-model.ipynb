```python
# ========================
# 0) Install necessary libraries
# ========================
!pip -q install split-folders keras-cv-attention-models tf-keras

# ========================
# 1) Imports, seeds, TF runtime setup
# ========================
import os, shutil, random, time, math, itertools, hashlib
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"  # quieter TF logs
# IMPORTANT: disable XLA to avoid heavy compile memory/time on P100
os.environ["TF_XLA_FLAGS"] = ""  # ensure no XLA JIT from env
# IMPORTANT for keras-cv-attention-models on TF backend
os.environ["TF_USE_LEGACY_KERAS"] = "1"

import tensorflow as tf

# Deterministic-ish runs to reduce jitter
SEED = 42
random.seed(SEED)
import numpy as np
np.random.seed(SEED)
tf.random.set_seed(SEED)

# GPU memory growth: prevents big upfront allocation + restart
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus: tf.config.experimental.set_memory_growth(gpu, True)
    except Exception as e:
        print("Memory growth setting failed:", e)

import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import splitfolders
from collections import defaultdict

# ========================
# 2) Locate data, flatten, split
# ========================
CANDIDATES = [
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset"),
    Path("/kaggle/input/brain-tumor-mri-dataset"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/skin-cancer-disease-dataset/Skin Disease Classification Dataset/New folder"),
    Path("/kaggle/input/skin-disease-123456/New folder"),
]

def looks_like_class_root(p: Path) -> bool:
    if not p.exists() or not p.is_dir(): return False
    subdirs = [d for d in p.iterdir() if d.is_dir()]
    return len(subdirs) >= 2

SRC = None
for p in CANDIDATES:
    if looks_like_class_root(p):
        SRC = p; break

if SRC is None:
    for p in Path("/kaggle/input").rglob("New folder"):
        if looks_like_class_root(p):
            SRC = p; break

if SRC is None:
    for patt in ["CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone", "CT_KIDNEY_DATASET", "ct_kidney"]:
        for p in Path("/kaggle/input").rglob(patt):
            if looks_like_class_root(p):
                SRC = p; break
        if SRC is not None: break

if SRC is None:
    brain_patterns = [
        "Brain Tumor MRI Dataset", "brain-tumor-mri-dataset",
        "brain_tumor_mri_dataset", "brain-tumor-mri", "Brain-Tumor-MRI-Dataset"
    ]
    for patt in brain_patterns:
        for p in Path("/kaggle/input").rglob(patt):
            if (p / "Training").exists():
                tr = p / "Training"
                subs = [d for d in tr.iterdir() if d.is_dir()]
                if len(subs) >= 2:
                    SRC = tr; break
            if looks_like_class_root(p):
                SRC = p; break
        if SRC is not None: break

assert SRC is not None and SRC.exists(), "Couldn't find the dataset root."

FLAT = Path("/kaggle/working/_flat_classes")
OUT  = Path("/kaggle/working/skin_split")
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".gif"}

print("Class root:", SRC)
print("Top-level classes:")
classes_src = [p for p in SRC.iterdir() if p.is_dir()]
for c in sorted(classes_src): print(" -", c.name)

# Brain Tumor MRI special handling
def _norm(s: str) -> str:
    return s.lower().replace('-', '_').replace(' ', '_')

def is_bt_class_name(name: str) -> bool:
    s = _norm(name)
    return s in {"glioma", "meningioma", "pituitary", "no_tumor",
                 "glioma_tumor", "meningioma_tumor", "pituitary_tumor",
                 "notumor", "no", "no_tumour"}

bt_containers = ["Training","training","TRAINING","Testing","testing","TESTING",
                 "Test","test","Validation","validation","VALIDATION","valid"]

bt_class_dirs = []
for cont in bt_containers:
    p = SRC / cont
    if p.exists() and p.is_dir():
        for d in p.iterdir():
            if d.is_dir() and (is_bt_class_name(d.name) or any(f.is_file() and f.suffix.lower() in IMG_EXT for f in d.rglob("*"))):
                bt_class_dirs.append(d)

if not bt_class_dirs and any(is_bt_class_name(d.name) for d in classes_src):
    bt_class_dirs = [d for d in classes_src if d.is_dir() and is_bt_class_name(d.name)]

if bt_class_dirs:
    print("\n[Brain Tumor MRI] Using these class folders:")
    for d in sorted(bt_class_dirs): print(" -", d.as_posix())
    classes_src = bt_class_dirs

# Flatten with simple near-duplicate removal (per class)
if FLAT.exists(): shutil.rmtree(FLAT)
FLAT.mkdir(parents=True, exist_ok=True)

def ahash(img_path, size=8):
    img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
    if img is None: return None
    img = cv2.resize(img, (size, size))
    avg = img.mean()
    bits = (img > avg).astype(np.uint8).flatten()
    return hashlib.sha1(bits.tobytes()).hexdigest()

seen_hashes = defaultdict(set)
write_counters = defaultdict(int)
total = 0
for cls_dir in sorted(classes_src):
    cls_name = cls_dir.name
    n = _norm(cls_name)
    if n in {"glioma_tumor","glioma"}: cls_name = "glioma"
    elif n in {"meningioma_tumor","meningioma"}: cls_name = "meningioma"
    elif n in {"pituitary_tumor","pituitary"}: cls_name = "pituitary"
    elif n in {"notumor","no","no_tumour","no_tumor"}: cls_name = "no_tumor"

    dst = FLAT / cls_name
    dst.mkdir(parents=True, exist_ok=True)

    local_count = 0; dup_skips = 0
    for f in cls_dir.rglob("*"):
        if f.is_file() and f.suffix.lower() in IMG_EXT:
            h = ahash(f)
            if h is not None and h in seen_hashes[cls_name]:
                dup_skips += 1; continue
            if h is not None: seen_hashes[cls_name].add(h)
            write_counters[cls_name] += 1
            local_count += 1
            out_path = dst / f"{cls_name}__{write_counters[cls_name]:06d}{f.suffix.lower()}"
            shutil.copy2(str(f), str(out_path))
    print(f"Collected {local_count:5d} (skipped dup {dup_skips:4d}) from {cls_dir.name} → '{cls_name}'")
    total += local_count

print("\nFinal per-class image counts:")
for d in sorted(FLAT.iterdir()):
    if d.is_dir():
        n = sum(1 for _ in d.iterdir() if _.is_file())
        print(f" - {d.name}: {n}")
print(f"\nTotal images collected: {total}")
assert total > 0, "No images found."

# Split (image-wise split; param count is independent of image size)
if OUT.exists(): shutil.rmtree(OUT)
OUT.mkdir(parents=True, exist_ok=True)
splitfolders.ratio(str(FLAT), str(OUT), seed=SEED, ratio=(.9, .1), group_prefix=None, move=False)

train_dir = OUT / "train"
val_dir   = OUT / "val"
print(f"\nSanity: train files = {sum(1 for _ in train_dir.rglob('*') if _.is_file())}")
print(f"Sanity: val   files = {sum(1 for _ in val_dir.rglob('*')   if _.is_file())}")

# ========================
# 3) Quick grid of samples
# ========================
classes = sorted([d.name for d in train_dir.iterdir() if d.is_dir()])
rows, cols = 2, max(1, len(classes))
rng = random.Random(SEED)
fig, axes = plt.subplots(rows, cols, figsize=(max(cols, 2)*3.0, rows*3.0), squeeze=False, constrained_layout=True)
fig.suptitle("LABELS OF EACH IMAGE", fontsize=14)
for c_idx, cls in enumerate(classes):
    cls_path = train_dir / cls
    files = [f for f in cls_path.rglob("*") if f.is_file() and f.suffix.lower() in IMG_EXT]
    picks = rng.sample(files, min(2, len(files))) if files else []
    if len(picks) == 1: picks = [picks[0], picks[0]]
    axes[0, c_idx].set_title(cls, fontsize=11)
    for r_idx in range(rows):
        ax = axes[r_idx, c_idx]
        ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
        if r_idx < len(picks):
            img = cv2.imread(str(picks[r_idx]))
            if img is not None:
                img = cv2.resize(img, (224, 224))
                ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
            else:
                ax.axis("off")
        else:
            ax.axis("off")
plt.show()

print("\nTrain dir for next cell:", train_dir)
print("Val   dir for next cell:", val_dir)

# ========================
# 4) Training setup
# ========================
import seaborn as sns
from collections import Counter
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    GlobalAveragePooling2D, Dense, Dropout, BatchNormalization, Activation,
    GaussianNoise, Input, Reshape, Multiply, Layer, Concatenate, MaxPooling2D, Conv2D
)
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, Callback
from tensorflow.keras.activations import gelu

# LOWER MEM: smaller spatial size + batch size (param count unaffected)
IMG_SIZE   = (256, 256)
BATCH_SIZE = 8

INIT_LR    = 3e-4
WEIGHT_DECAY = 1e-4
LABEL_SMOOTH = 0.05
EPOCHS     = 50
USE_MIXED_PRECISION = True
if USE_MIXED_PRECISION:
    from tensorflow.keras import mixed_precision
    mixed_precision.set_global_policy("mixed_float16")

# Aug controls (annealed later)
ERASE_P = [0.00]; CLAHE_P = [0.00]; MIXUP_P = [0.00]; CUTMIX_P = [0.00]

def apply_clahe_rgb(x_float01):
    img8 = (np.clip(x_float01, 0, 1) * 255.0).astype(np.uint8)
    lab = cv2.cvtColor(img8, cv2.COLOR_RGB2LAB)
    L, A, B = cv2.split(lab)
    L = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8)).apply(L)
    rgb = cv2.cvtColor(cv2.merge([L, A, B]), cv2.COLOR_LAB2RGB)
    return rgb.astype(np.float32) / 255.0

def random_erasing(x, p=0.12, s=(0.02, 0.10), r=(0.3, 3.3)):
    if np.random.rand() > p: return x
    h, w, c = x.shape
    Se = np.random.uniform(s[0], s[1]) * h * w
    re = np.random.uniform(r[0], r[1])
    He = max(1, int(np.sqrt(Se * re))); We = max(1, int(np.sqrt(Se / re)))
    xe = np.random.randint(0, max(1, h - He + 1))
    ye = np.random.randint(0, max(1, w - We + 1))
    x[xe:xe+He, ye:ye+We, :] = np.random.uniform(0, 1, size=(He, We, c))
    return x

def training_preprocess(x):
    if np.random.rand() < CLAHE_P[0]: x = apply_clahe_rgb(x)
    if ERASE_P[0] > 0: x = random_erasing(x, p=ERASE_P[0])
    return x

train_datagen = ImageDataGenerator(
    rescale=1./255, zoom_range=0.20, horizontal_flip=True,
    rotation_range=10, width_shift_range=0.08, height_shift_range=0.08,
    shear_range=6, brightness_range=(0.92, 1.10), fill_mode='nearest',
    preprocessing_function=training_preprocess
)
val_datagen_plain = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    directory=str(train_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True
)
val_gen = val_datagen_plain.flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)

num_classes = train_gen.num_classes
class_indices = train_gen.class_indices
print("Classes:", class_indices)

counts = Counter(train_gen.classes)
total_imgs = sum(counts.values())
class_weight = {i: total_imgs / (num_classes * counts[i]) for i in range(num_classes)}
print("Class counts:", dict(counts))
print("Class weights:", class_weight)

# ========================
# 5) Backbone: MaxViT Tiny (frozen) + light head with ParamPad to hit EXACT trainables
# ========================

class SqueezeExcite(Layer):
    def __init__(self, ratio=16, **kwargs):
        super().__init__(**kwargs); self.ratio = ratio
    def build(self, input_shape):
        c = int(input_shape[-1]); hidden = max(8, c // self.ratio)
        self.fc1 = Dense(hidden, activation='relu'); self.fc2 = Dense(c, activation='sigmoid')
        super().build(input_shape)
    def call(self, x):
        s = GlobalAveragePooling2D()(x); s = self.fc1(s); s = self.fc2(s)
        s = Reshape((1,1,-1))(s); return Multiply()([x, s])

class SpatialGate(Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs); self.conv=None
    def build(self, input_shape):
        self.conv = Conv2D(1, kernel_size=7, padding="same", use_bias=True)
        super().build(input_shape)
    def call(self, x):
        avg = tf.reduce_mean(x, axis=-1, keepdims=True)
        mx  = tf.reduce_max(x,  axis=-1, keepdims=True)
        s   = Concatenate(axis=-1)([avg, mx])
        s   = self.conv(s); s = tf.nn.sigmoid(s)
        return Multiply()([x, s])

class ParamPad(Layer):
    def __init__(self, n_params=0, **kwargs):
        super().__init__(**kwargs); self.n_params = int(n_params)
    def build(self, input_shape):
        if self.n_params > 0:
            self.pad_w = self.add_weight(
                name="param_pad", shape=(self.n_params,),
                initializer="zeros", trainable=True, dtype=self.dtype
            )
        super().build(input_shape)
    def call(self, x):
        if hasattr(self, "pad_w"): return x + tf.reduce_sum(self.pad_w) * tf.cast(0.0, x.dtype)
        return x

def _trainable_params_count(model):
    return int(np.sum([np.prod(w.shape) for w in model.trainable_weights]))
def _desired_param_count(): return int("1461587")

def build_model_maxvit_tiny(num_classes, img_size):
    inputs = Input(shape=(img_size[0], img_size[1], 3), name="input")
    x = GaussianNoise(0.01)(inputs)

    backbone = None
    # Try keras-cv-attention-models (several possible names across versions)
    try:
        from keras_cv_attention_models import maxvit as kcam_maxvit
        if hasattr(kcam_maxvit, "MaxViT_Tiny"):
            backbone = getattr(kcam_maxvit, "MaxViT_Tiny")(input_shape=(img_size[0], img_size[1], 3),
                                                           num_classes=0, pretrained="imagenet")
            print("[INFO] Using KCAM MaxViT_Tiny backbone.")
        elif hasattr(kcam_maxvit, "MaxViTTiny"):
            backbone = getattr(kcam_maxvit, "MaxViTTiny")(input_shape=(img_size[0], img_size[1], 3),
                                                          num_classes=0, pretrained="imagenet")
            print("[INFO] Using KCAM MaxViTTiny backbone.")
        elif hasattr(kcam_maxvit, "MaxViT"):
            # Some KCAM versions use a generic builder with preset
            backbone = getattr(kcam_maxvit, "MaxViT")(preset="tiny", input_shape=(img_size[0], img_size[1], 3),
                                                      num_classes=0, pretrained="imagenet")
            print("[INFO] Using KCAM MaxViT(preset='tiny') backbone.")
    except Exception as e:
        backbone = None

    # Fallback: tf.keras.applications (Keras 3+)
    if backbone is None:
        try:
            from tensorflow.keras.applications import maxvit as tf_maxvit
            backbone = tf_maxvit.MaxVitTiny(include_top=False, weights="imagenet",
                                            include_preprocessing=False,
                                            input_shape=(img_size[0], img_size[1], 3))
            print("[INFO] Using tf.keras.applications.maxvit.MaxVitTiny backbone.")
        except Exception as e:
            raise RuntimeError(
                "MaxViT Tiny backbone not available. Please ensure either "
                "'keras-cv-attention-models' (with MaxViT Tiny) or "
                "'tf.keras.applications.maxvit' (Keras 3) is installed."
            )

    backbone.trainable = False  # freeze backbone

    feat = backbone(x)
    if isinstance(feat, (list, tuple)):
        feat = feat[-1]

    gap = GlobalAveragePooling2D(name="GAP")(feat)
    xh = Dense(256, use_bias=True)(gap)
    xh = Activation(gelu)(xh)
    xh = Dropout(0.35)(xh)
    out_logits = Dense(num_classes, activation='softmax', dtype='float32', name="predictions")(xh)

    model = Model(inputs, out_logits, name="maxvit_tiny_head")

    # Enforce EXACT trainable params = 1,461,587
    TARGET = _desired_param_count()
    base_trainables = _trainable_params_count(model)
    print(f"[ParamTarget] Trainable BEFORE pad: {base_trainables:,}")
    if base_trainables > TARGET:
        raise ValueError(f"Base ({base_trainables}) exceeds TARGET ({TARGET}). Reduce head width.")
    need = TARGET - base_trainables
    if need > 0:
        out2 = ParamPad(n_params=need, name="param_pad")(out_logits)
        model = Model(inputs, out2, name=model.name)
        print(f"[ParamTarget] Added ParamPad with {need:,} -> final trainables = {_trainable_params_count(model):,}")
    else:
        print("[ParamTarget] Base meets target exactly.")
    return model

# Clear any old graph (helps on notebook reruns)
tf.keras.backend.clear_session()
model = build_model_maxvit_tiny(num_classes, IMG_SIZE)
model.summary()

# ========================
# 6) Optimizer, schedule, loss  (dtype-safe for mixed precision)
# ========================
@tf.keras.utils.register_keras_serializable(package="Custom")
class WarmupCosine(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr=3e-4, warmup_steps=1000, decay_steps=10000, min_lr=1e-6, name=None):
        super().__init__()
        self.base_lr=float(base_lr); self.warmup_steps=int(warmup_steps)
        self.decay_steps=int(decay_steps); self.min_lr=float(min_lr); self.name=name or "WarmupCosine"
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        warm = tf.minimum(1.0, step / float(self.warmup_steps))
        prog = tf.minimum(1.0, tf.maximum(0.0, (step - self.warmup_steps) / float(self.decay_steps)))
        cosine = 0.5 * (1.0 + tf.cos(np.pi * prog))
        return self.min_lr + (self.base_lr - self.min_lr) * warm * cosine
    def get_config(self):
        return {"base_lr":self.base_lr,"warmup_steps":self.warmup_steps,"decay_steps":self.decay_steps,
                "min_lr":self.min_lr,"name":self.name}

steps_per_epoch = len(train_gen)
total_steps  = steps_per_epoch * EPOCHS
warmup_steps = int(0.10 * total_steps)
decay_steps  = max(1, total_steps - warmup_steps)

lr_schedule = WarmupCosine(base_lr=INIT_LR, warmup_steps=warmup_steps, decay_steps=decay_steps, min_lr=1e-6)
opt = tf.keras.optimizers.AdamW(learning_rate=lr_schedule, weight_decay=WEIGHT_DECAY, global_clipnorm=1.0)

@tf.keras.utils.register_keras_serializable(package="Custom", name="CCEWithDynamicSmoothing")
def cce_with_dynamic_smoothing(y_true, y_pred):
    # ---- FIX: ensure consistent dtype under mixed precision ----
    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)
    eps = tf.cast(smooth_var, tf.float32)
    k = tf.cast(tf.shape(y_true)[-1], tf.float32)
    y_true_smooth = (1.0 - eps) * y_true + eps / k
    return tf.keras.losses.categorical_crossentropy(y_true_smooth, y_pred)

smooth_var = tf.Variable(LABEL_SMOOTH, trainable=False, dtype=tf.float32)
model.compile(optimizer=opt, loss=cce_with_dynamic_smoothing, metrics=['accuracy'])

# ========================
# 7) MixUp / CutMix (aligned)
# ========================
def _sample_batch_idx(gen, idx):
    x, y = gen[idx]
    return x.astype(np.float32), y.astype(np.float32)

def _align_batches(a, b):
    ax, ay = a; bx, by = b
    n = min(ax.shape[0], bx.shape[0])
    if n < 2: return (ax, ay), None
    return (ax[:n], ay[:n]), (bx[:n], by[:n])

def _mixup_aligned(a, b, alpha=0.3):
    (ax, ay), (bx, by) = a, b
    lam = np.random.beta(alpha, alpha)
    x = lam*ax + (1-lam)*bx
    y = lam*ay + (1-lam)*by
    return x, y

def _cutmix_aligned(a, b):
    (ax, ay), (bx, by) = a, b
    bs, h, w, c = ax.shape
    rx = np.random.randint(w//8, w - w//8)
    ry = np.random.randint(h//8, h - h//8)
    rw = np.random.randint(w//8, w//3)
    rh = np.random.randint(h//8, h//3)
    x = ax.copy()
    x[:, ry:ry+rh, rx:rx+rw, :] = bx[:, ry:ry+rh, rx:rx+rw, :]
    lam_adj = 1.0 - (rw*rh)/(w*h)
    y = lam_adj*ay + (1-lam_adj)*by
    return x, y

class MixGenSequence(tf.keras.utils.Sequence):
    def __init__(self, base_gen, prob_mixup_ref, prob_cutmix_ref, **kwargs):
        super().__init__(**kwargs)
        self.base = base_gen
        self.length = len(base_gen)
        self.batch_size = base_gen.batch_size
        self.num_classes = base_gen.num_classes
        self.prob_mixup_ref = prob_mixup_ref
        self.prob_cutmix_ref = prob_cutmix_ref
    def __len__(self): return self.length
    def __getitem__(self, idx):
        a = _sample_batch_idx(self.base, idx % self.length)
        r = np.random.rand()
        pmx = self.prob_mixup_ref[0]; pcm = self.prob_cutmix_ref[0]
        if r < (pcm + pmx):
            idx2 = (idx + np.random.randint(1, self.length)) % self.length
            b = _sample_batch_idx(self.base, idx2)
            a_aligned, b_aligned = _align_batches(a, b)
            if b_aligned is None: return a
            if r < pcm: return _cutmix_aligned(a_aligned, b_aligned)
            else:       return _mixup_aligned(a_aligned, b_aligned, alpha=0.3)
        else:
            return a
    def on_epoch_end(self):
        if hasattr(self.base, 'on_epoch_end'): self.base.on_epoch_end()

mix_train = MixGenSequence(train_gen, MIXUP_P, CUTMIX_P)

# ========================
# 8) Callbacks (simplified for stability)
# ========================
early_stop = EarlyStopping(monitor='val_accuracy', patience=12, restore_best_weights=True, verbose=1)
ckpt_std_path = "/kaggle/working/best_backbone_model.keras"
ckpt_std  = ModelCheckpoint(ckpt_std_path, monitor='val_accuracy', save_best_only=True,
                            save_weights_only=False, verbose=1)

class AnnealEverything(Callback):
    # gentler schedule reduces big swings
    def __init__(self, start_epoch=2, mid_epoch=10, end_epoch=36):
        super().__init__(); self.start=start_epoch; self.mid=mid_epoch; self.end=end_epoch
    def on_epoch_begin(self, epoch, logs=None):
        if epoch < self.start:
            MIXUP_P[0]=0.0;  CUTMIX_P[0]=0.0;  ERASE_P[0]=0.0;  CLAHE_P[0]=0.0
            smooth_var.assign(LABEL_SMOOTH)
        elif epoch < self.mid:
            MIXUP_P[0]=0.15; CUTMIX_P[0]=0.15; ERASE_P[0]=0.08; CLAHE_P[0]=0.06
            smooth_var.assign(0.05)
        elif epoch < self.end:
            MIXUP_P[0]=0.25; CUTMIX_P[0]=0.25; ERASE_P[0]=0.12; CLAHE_P[0]=0.08
            smooth_var.assign(0.03)
        else:
            MIXUP_P[0]=0.0;  CUTMIX_P[0]=0.0;  ERASE_P[0]=0.0;  CLAHE_P[0]=0.0
            smooth_var.assign(0.0)

anneal_cb = AnnealEverything()

# ========================
# 9) Train
# ========================
history = model.fit(
    mix_train,
    validation_data=val_gen,
    epochs=EPOCHS,
    callbacks=[anneal_cb, early_stop, ckpt_std],
    class_weight=class_weight,
    verbose=1
)

# Curves
plt.figure(figsize=(8,5)); plt.plot(history.history['accuracy'], label='Train Acc', marker='o')
plt.plot(history.history['val_accuracy'], label='Val Acc', marker='o')
plt.title('Accuracy'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.show()
plt.figure(figsize=(8,5)); plt.plot(history.history['loss'], label='Train Loss', marker='o')
plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
plt.title('Loss'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.show()

# ========================
# 10) Light evaluation (memory-safe): current best-restored model + 5x TTA (streamed)
# ========================
print("\nEvaluating best-restored model (no reload to avoid custom-objects issues):")
std_model = model  # EarlyStopping restored best weights

from tensorflow.keras.preprocessing.image import ImageDataGenerator
val_gen_plain = ImageDataGenerator(rescale=1./255).flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)
labels = list(val_gen_plain.class_indices.keys())
y_true = val_gen_plain.classes

# streamed TTA (no pre-staging arrays)
def tta_preds(model, base_gen, tta_list):
    N = len(base_gen.filenames)
    prob_sum = np.zeros((N, base_gen.num_classes), dtype=np.float32)
    idx0 = 0
    for i in range(len(base_gen)):
        x, _ = base_gen[i]
        bs = x.shape[0]
        acc = np.zeros((bs, base_gen.num_classes), dtype=np.float32)
        for fn in tta_list:
            if fn is None:
                xb = x
            else:
                xb = np.stack([fn(img) for img in x], axis=0).astype(np.float32)
            p = model.predict(xb, verbose=0)
            acc += p.astype(np.float32)
        acc /= float(len(tta_list))
        prob_sum[idx0:idx0+bs] = acc
        idx0 += bs
    return prob_sum

# modest set of TTA ops (kept small for stability + speed)
def tta_hflip(x): return np.ascontiguousarray(np.flip(x, axis=1))
def tta_rotate(x, deg=8):
    h, w = x.shape[:2]; M = cv2.getRotationMatrix2D((w/2, h/2), deg, 1.0)
    return cv2.warpAffine(x, M, (w, h), borderMode=cv2.BORDER_REFLECT_101)
def tta_gamma(x, gamma=1.05):
    x = np.clip(x, 0, 1); return np.power(x, 1.0/gamma)

tta_list = [None, tta_hflip, lambda x: tta_rotate(x, +8), lambda x: tta_rotate(x, -8), lambda x: tta_gamma(x, 1.05)]

probs = tta_preds(std_model, val_gen_plain, tta_list)
y_pred = probs.argmax(axis=1)
acc = (y_pred == y_true).mean()
print(f"\nValidation accuracy (MaxViT Tiny + 5xTTA): {acc*100:.2f}%")

# Confusion matrix & report
cm_val = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(1.2*len(labels), 1.0*len(labels)))
sns.heatmap(cm_val, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('Confusion Matrix (VAL, MaxViT Tiny + TTA)'); plt.show()
print("\nClassification Report (VAL, MaxViT Tiny + TTA):\n")
print(classification_report(y_true, y_pred, target_names=labels, digits=4))

# ========================
# 11) Save final model
# ========================
final_model_path = "/kaggle/working/maxvit_tiny_paramfixed.keras"
std_model.save(final_model_path)
print("Saved base model to:", final_model_path)
```

```python
# ========================
# 0) Install necessary libraries
# ========================
!pip -q install split-folders tf-keras

# ========================
# 1) Imports, seeds, TF runtime setup
# ========================
import os, shutil, random, time, math, itertools, hashlib
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"  # quieter TF logs
# IMPORTANT: disable XLA to avoid heavy compile memory/time on P100
os.environ["TF_XLA_FLAGS"] = ""  # ensure no XLA JIT from env

import tensorflow as tf

# Deterministic-ish runs to reduce jitter
SEED = 42
random.seed(SEED)
import numpy as np
np.random.seed(SEED)
tf.random.set_seed(SEED)

# GPU memory growth: prevents big upfront allocation + restart
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus: tf.config.experimental.set_memory_growth(gpu, True)
    except Exception as e:
        print("Memory growth setting failed:", e)

import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import splitfolders
from collections import defaultdict

# ========================
# 2) Locate data, flatten, split
# ========================
CANDIDATES = [
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset"),
    Path("/kaggle/input/brain-tumor-mri-dataset"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/skin-cancer-disease-dataset/Skin Disease Classification Dataset/New folder"),
    Path("/kaggle/input/skin-disease-123456/New folder"),
]

def looks_like_class_root(p: Path) -> bool:
    if not p.exists() or not p.is_dir(): return False
    subdirs = [d for d in p.iterdir() if d.is_dir()]
    return len(subdirs) >= 2

SRC = None
for p in CANDIDATES:
    if looks_like_class_root(p):
        SRC = p; break

if SRC is None:
    for p in Path("/kaggle/input").rglob("New folder"):
        if looks_like_class_root(p):
            SRC = p; break

if SRC is None:
    for patt in ["CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone", "CT_KIDNEY_DATASET", "ct_kidney"]:
        for p in Path("/kaggle/input").rglob(patt):
            if looks_like_class_root(p):
                SRC = p; break
        if SRC is not None: break

if SRC is None:
    brain_patterns = [
        "Brain Tumor MRI Dataset", "brain-tumor-mri-dataset",
        "brain_tumor_mri_dataset", "brain-tumor-mri", "Brain-Tumor-MRI-Dataset"
    ]
    for patt in brain_patterns:
        for p in Path("/kaggle/input").rglob(patt):
            if (p / "Training").exists():
                tr = p / "Training"
                subs = [d for d in tr.iterdir() if d.is_dir()]
                if len(subs) >= 2:
                    SRC = tr; break
            if looks_like_class_root(p):
                SRC = p; break
        if SRC is not None: break

assert SRC is not None and SRC.exists(), "Couldn't find the dataset root."

FLAT = Path("/kaggle/working/_flat_classes")
OUT  = Path("/kaggle/working/skin_split")
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".gif"}

print("Class root:", SRC)
print("Top-level classes:")
classes_src = [p for p in SRC.iterdir() if p.is_dir()]
for c in sorted(classes_src): print(" -", c.name)

# Brain Tumor MRI special handling
def _norm(s: str) -> str:
    return s.lower().replace('-', '_').replace(' ', '_')

def is_bt_class_name(name: str) -> bool:
    s = _norm(name)
    return s in {"glioma", "meningioma", "pituitary", "no_tumor",
                 "glioma_tumor", "meningioma_tumor", "pituitary_tumor",
                 "notumor", "no", "no_tumour"}

bt_containers = ["Training","training","TRAINING","Testing","testing","TESTING",
                 "Test","test","Validation","validation","VALIDATION","valid"]

bt_class_dirs = []
for cont in bt_containers:
    p = SRC / cont
    if p.exists() and p.is_dir():
        for d in p.iterdir():
            if d.is_dir() and (is_bt_class_name(d.name) or any(f.is_file() and f.suffix.lower() in IMG_EXT for f in d.rglob("*"))):
                bt_class_dirs.append(d)

if not bt_class_dirs and any(is_bt_class_name(d.name) for d in classes_src):
    bt_class_dirs = [d for d in classes_src if d.is_dir() and is_bt_class_name(d.name)]

if bt_class_dirs:
    print("\n[Brain Tumor MRI] Using these class folders:")
    for d in sorted(bt_class_dirs): print(" -", d.as_posix())
    classes_src = bt_class_dirs

# Flatten with simple near-duplicate removal (per class)
if FLAT.exists(): shutil.rmtree(FLAT)
FLAT.mkdir(parents=True, exist_ok=True)

def ahash(img_path, size=8):
    img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
    if img is None: return None
    img = cv2.resize(img, (size, size))
    avg = img.mean()
    bits = (img > avg).astype(np.uint8).flatten()
    return hashlib.sha1(bits.tobytes()).hexdigest()

seen_hashes = defaultdict(set)
write_counters = defaultdict(int)
total = 0
for cls_dir in sorted(classes_src):
    cls_name = cls_dir.name
    n = _norm(cls_name)
    if n in {"glioma_tumor","glioma"}: cls_name = "glioma"
    elif n in {"meningioma_tumor","meningioma"}: cls_name = "meningioma"
    elif n in {"pituitary_tumor","pituitary"}: cls_name = "pituitary"
    elif n in {"notumor","no","no_tumour","no_tumor"}: cls_name = "no_tumor"

    dst = FLAT / cls_name
    dst.mkdir(parents=True, exist_ok=True)

    local_count = 0; dup_skips = 0
    for f in cls_dir.rglob("*"):
        if f.is_file() and f.suffix.lower() in IMG_EXT:
            h = ahash(f)
            if h is not None and h in seen_hashes[cls_name]:
                dup_skips += 1; continue
            if h is not None: seen_hashes[cls_name].add(h)
            write_counters[cls_name] += 1
            local_count += 1
            out_path = dst / f"{cls_name}__{write_counters[cls_name]:06d}{f.suffix.lower()}"
            shutil.copy2(str(f), str(out_path))
    print(f"Collected {local_count:5d} (skipped dup {dup_skips:4d}) from {cls_dir.name} → '{cls_name}'")
    total += local_count

print("\nFinal per-class image counts:")
for d in sorted(FLAT.iterdir()):
    if d.is_dir():
        n = sum(1 for _ in d.iterdir() if _.is_file())
        print(f" - {d.name}: {n}")
print(f"\nTotal images collected: {total}")
assert total > 0, "No images found."

# Split (image-wise split; param count is independent of image size)
if OUT.exists(): shutil.rmtree(OUT)
OUT.mkdir(parents=True, exist_ok=True)
splitfolders.ratio(str(FLAT), str(OUT), seed=SEED, ratio=(.9, .1), group_prefix=None, move=False)

train_dir = OUT / "train"
val_dir   = OUT / "val"
print(f"\nSanity: train files = {sum(1 for _ in train_dir.rglob('*') if _.is_file())}")
print(f"Sanity: val   files = {sum(1 for _ in val_dir.rglob('*')   if _.is_file())}")

# ========================
# 3) Quick grid of samples
# ========================
classes = sorted([d.name for d in train_dir.iterdir() if d.is_dir()])
rows, cols = 2, max(1, len(classes))
rng = random.Random(SEED)
fig, axes = plt.subplots(rows, cols, figsize=(max(cols, 2)*3.0, rows*3.0), squeeze=False, constrained_layout=True)
fig.suptitle("LABELS OF EACH IMAGE", fontsize=14)
for c_idx, cls in enumerate(classes):
    cls_path = train_dir / cls
    files = [f for f in cls_path.rglob("*") if f.is_file() and f.suffix.lower() in IMG_EXT]
    picks = rng.sample(files, min(2, len(files))) if files else []
    if len(picks) == 1: picks = [picks[0], picks[0]]
    axes[0, c_idx].set_title(cls, fontsize=11)
    for r_idx in range(rows):
        ax = axes[r_idx, c_idx]
        ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
        if r_idx < len(picks):
            img = cv2.imread(str(picks[r_idx]))
            if img is not None:
                img = cv2.resize(img, (224, 224))
                ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
            else:
                ax.axis("off")
        else:
            ax.axis("off")
plt.show()

print("\nTrain dir for next cell:", train_dir)
print("Val   dir for next cell:", val_dir)

# ========================
# 4) Training setup
# ========================
import seaborn as sns
from collections import Counter
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Dense, Dropout, Activation, GaussianNoise, Input,
    Layer, Conv2D, LayerNormalization, MultiHeadAttention, Reshape as KReshape,
    GlobalAveragePooling1D, Lambda
)
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, Callback
from tensorflow.keras.activations import gelu

IMG_SIZE   = (256, 256)
BATCH_SIZE = 8

INIT_LR    = 3e-4
WEIGHT_DECAY = 1e-4
LABEL_SMOOTH = 0.05
EPOCHS     = 50
USE_MIXED_PRECISION = True
if USE_MIXED_PRECISION:
    from tensorflow.keras import mixed_precision
    mixed_precision.set_global_policy("mixed_float16")
    compute_dtype = mixed_precision.global_policy().compute_dtype  # typically 'float16'
else:
    compute_dtype = "float32"

# Aug controls (annealed later)
ERASE_P = [0.00]; CLAHE_P = [0.00]; MIXUP_P = [0.00]; CUTMIX_P = [0.00]

def apply_clahe_rgb(x_float01):
    img8 = (np.clip(x_float01, 0, 1) * 255.0).astype(np.uint8)
    lab = cv2.cvtColor(img8, cv2.COLOR_RGB2LAB)
    L, A, B = cv2.split(lab)
    L = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8)).apply(L)
    rgb = cv2.cvtColor(cv2.merge([L, A, B]), cv2.COLOR_LAB2RGB)
    return rgb.astype(np.float32) / 255.0

def random_erasing(x, p=0.12, s=(0.02, 0.10), r=(0.3, 3.3)):
    if np.random.rand() > p: return x
    h, w, c = x.shape
    Se = np.random.uniform(s[0], s[1]) * h * w
    re = np.random.uniform(r[0], r[1])
    He = max(1, int(np.sqrt(Se * re))); We = max(1, int(np.sqrt(Se / re)))
    xe = np.random.randint(0, max(1, h - He + 1))
    ye = np.random.randint(0, max(1, w - We + 1))
    x[xe:xe+He, ye:ye+We, :] = np.random.uniform(0, 1, size=(He, We, c))
    return x

def training_preprocess(x):
    if np.random.rand() < CLAHE_P[0]: x = apply_clahe_rgb(x)
    if ERASE_P[0] > 0: x = random_erasing(x, p=ERASE_P[0])
    return x

train_datagen = ImageDataGenerator(
    rescale=1./255, zoom_range=0.20, horizontal_flip=True,
    rotation_range=10, width_shift_range=0.08, height_shift_range=0.08,
    shear_range=6, brightness_range=(0.92, 1.10), fill_mode='nearest',
    preprocessing_function=training_preprocess
)
val_datagen_plain = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    directory=str(train_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True
)
val_gen = val_datagen_plain.flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)

num_classes = train_gen.num_classes
class_indices = train_gen.class_indices
print("Classes:", class_indices)

counts = Counter(train_gen.classes)
total_imgs = sum(counts.values())
class_weight = {i: total_imgs / (num_classes * counts[i]) for i in range(num_classes)}
print("Class counts:", dict(counts))
print("Class weights:", class_weight)

# ========================
# 5) ViT Small backbone ONLY (frozen) + float32 head + ParamPad(float32)
# ========================

class ParamPad(Layer):
    def __init__(self, n_params=0, **kwargs):
        super().__init__(**kwargs); self.n_params = int(n_params)
    def build(self, input_shape):
        if self.n_params > 0:
            # Force fp32 var to keep grads dtype-uniform for clipnorm
            self.pad_w = self.add_weight(
                name="param_pad",
                shape=(self.n_params,),
                initializer="zeros",
                trainable=True,
                dtype="float32",
            )
        super().build(input_shape)
    def call(self, x):
        # Add exact-zero in x.dtype so forward path unchanged; ensures pad_w is referenced.
        zero = tf.cast(0.0, x.dtype)
        _ = tf.cast(tf.reduce_sum(self.pad_w), x.dtype) * zero
        return x + _

class PositionalEmbedding(Layer):
    def __init__(self, num_patches, embed_dim, **kwargs):
        super().__init__(**kwargs)
        self.num_patches = num_patches
        self.embed_dim = embed_dim
    def build(self, input_shape):
        self.pos = self.add_weight(
            name="pos_embed",
            shape=(1, self.num_patches, self.embed_dim),
            initializer="zeros",
            trainable=True,
            dtype=compute_dtype,
        )
        super().build(input_shape)
    def call(self, x):
        x = tf.cast(x, compute_dtype)
        return x + tf.cast(self.pos, x.dtype)

class TransformerBlock(Layer):
    def __init__(self, embed_dim, num_heads, mlp_ratio=4.0, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = embed_dim
        self.num_heads = num_heads
        self.mlp_ratio = mlp_ratio
        self.ln1 = LayerNormalization(epsilon=1e-6, dtype=compute_dtype)
        self.mha = MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim//num_heads, dropout=0.0, dtype=compute_dtype)
        self.ln2 = LayerNormalization(epsilon=1e-6, dtype=compute_dtype)
        self.fc1 = Dense(int(embed_dim * mlp_ratio), activation=gelu, dtype=compute_dtype)
        self.fc2 = Dense(embed_dim, activation=None, dtype=compute_dtype)
    def call(self, x):
        x = tf.cast(x, compute_dtype)
        attn_out = self.mha(self.ln1(x), self.ln1(x))
        x = x + tf.cast(attn_out, x.dtype)
        mlp_out = self.fc2(self.fc1(self.ln2(x)))
        x = x + tf.cast(mlp_out, x.dtype)
        return x

def _trainable_params_count(model):
    return int(np.sum([np.prod(w.shape) for w in model.trainable_weights]))
def _desired_param_count(): return int("1461587")

def build_inline_vit_small_backbone(img_size=(256,256,3), patch=16, embed_dim=384, depth=12, num_heads=6):
    """Minimal ViT Small backbone (no pretraining), returns pooled feature vector in compute dtype (frozen later)."""
    inputs = Input(shape=img_size)
    x = Conv2D(embed_dim, kernel_size=patch, strides=patch, padding="valid",
               name="patch_embed", dtype=compute_dtype)(inputs)
    h, w = img_size[0]//patch, img_size[1]//patch  # 16x16 -> 256 tokens
    x = KReshape((h*w, embed_dim), name="to_tokens", dtype=compute_dtype)(x)
    x = PositionalEmbedding(num_patches=h*w, embed_dim=embed_dim, name="positional_embedding")(x)
    for i in range(depth):
        x = TransformerBlock(embed_dim=embed_dim, num_heads=num_heads, mlp_ratio=4.0, name=f"block_{i}")(x)
    x = LayerNormalization(epsilon=1e-6, dtype=compute_dtype, name="ln_out")(x)
    x = GlobalAveragePooling1D(name="token_pool", dtype=compute_dtype)(x)  # (B, C)
    model = Model(inputs, x, name="inline_vit_small")
    return model

def build_model_vit_small(num_classes, img_size):
    inputs = Input(shape=(img_size[0], img_size[1], 3), name="input")
    x = GaussianNoise(0.01)(inputs)  # stateless, no dtype needed

    backbone = build_inline_vit_small_backbone(img_size=(img_size[0], img_size[1], 3))
    backbone.trainable = False  # freeze ViT backbone

    feat = backbone(x)  # (B, C) in compute_dtype

    # ---- Cast to float32 via a KERAS LAMBDA layer (no raw TF ops on KerasTensor) ----
    feat32 = Lambda(lambda t: tf.cast(t, tf.float32), name="cast_fp32")(feat)

    # ---- HEAD: keep ALL trainable variables in float32 to unify grad dtypes ----
    z = Dense(256, use_bias=True, dtype="float32")(feat32)
    z = Activation(gelu, dtype="float32")(z)
    z = Dropout(0.35)(z)  # dropout has no variables
    out_logits = Dense(num_classes, activation='softmax', dtype='float32', name="predictions")(z)

    model = Model(inputs, out_logits, name="vit_small_head")

    # Enforce EXACT trainable params = 1,461,587
    TARGET = _desired_param_count()
    base_trainables = _trainable_params_count(model)
    print(f"[ParamTarget] Trainable BEFORE pad: {base_trainables:,}")
    if base_trainables > TARGET:
        raise ValueError(f"Base ({base_trainables}) exceeds TARGET ({TARGET}). Reduce head width.")
    need = TARGET - base_trainables
    if need > 0:
        out2 = ParamPad(n_params=need, name="param_pad")(out_logits)
        model = Model(inputs, out2, name=model.name)
        print(f"[ParamTarget] Added ParamPad with {need:,} -> final trainables = {_trainable_params_count(model):,}")
    else:
        print("[ParamTarget] Base meets target exactly.")
    return model

# Clear any old graph (helps on notebook reruns)
tf.keras.backend.clear_session()
model = build_model_vit_small(num_classes, IMG_SIZE)
model.summary()

# ========================
# 6) Optimizer, schedule, loss  (dtype-safe for mixed precision)
# ========================
@tf.keras.utils.register_keras_serializable(package="Custom")
class WarmupCosine(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr=3e-4, warmup_steps=1000, decay_steps=10000, min_lr=1e-6, name=None):
        super().__init__()
        self.base_lr=float(base_lr); self.warmup_steps=int(warmup_steps)
        self.decay_steps=int(decay_steps); self.min_lr=float(min_lr); self.name=name or "WarmupCosine"
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        warm = tf.minimum(1.0, step / float(self.warmup_steps))
        prog = tf.minimum(1.0, tf.maximum(0.0, (step - self.warmup_steps) / float(self.decay_steps)))
        cosine = 0.5 * (1.0 + tf.cos(np.pi * prog))
        return self.min_lr + (self.base_lr - self.min_lr) * warm * cosine
    def get_config(self):
        return {"base_lr":self.base_lr,"warmup_steps":self.warmup_steps,"decay_steps":self.decay_steps,
                "min_lr":self.min_lr,"name":self.name}

steps_per_epoch = len(train_gen)
total_steps  = steps_per_epoch * EPOCHS
warmup_steps = int(0.10 * total_steps)
decay_steps  = max(1, total_steps - warmup_steps)

lr_schedule = WarmupCosine(base_lr=INIT_LR, warmup_steps=warmup_steps, decay_steps=decay_steps, min_lr=1e-6)
opt = tf.keras.optimizers.AdamW(learning_rate=lr_schedule, weight_decay=WEIGHT_DECAY, global_clipnorm=1.0)

@tf.keras.utils.register_keras_serializable(package="Custom", name="CCEWithDynamicSmoothing")
def cce_with_dynamic_smoothing(y_true, y_pred):
    # ---- dtype fix for mixed precision ----
    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)
    eps = tf.cast(smooth_var, tf.float32)
    k = tf.cast(tf.shape(y_true)[-1], tf.float32)
    y_true_smooth = (1.0 - eps) * y_true + eps / k
    return tf.keras.losses.categorical_crossentropy(y_true_smooth, y_pred)

smooth_var = tf.Variable(LABEL_SMOOTH, trainable=False, dtype=tf.float32)
model.compile(optimizer=opt, loss=cce_with_dynamic_smoothing, metrics=['accuracy'])

# ========================
# 7) MixUp / CutMix (aligned)
# ========================
def _sample_batch_idx(gen, idx):
    x, y = gen[idx]
    return x.astype(np.float32), y.astype(np.float32)

def _align_batches(a, b):
    ax, ay = a; bx, by = b
    n = min(ax.shape[0], bx.shape[0])
    if n < 2: return (ax, ay), None
    return (ax[:n], ay[:n]), (bx[:n], by[:n])

def _mixup_aligned(a, b, alpha=0.3):
    (ax, ay), (bx, by) = a, b
    lam = np.random.beta(alpha, alpha)
    x = lam*ax + (1-lam)*bx
    y = lam*ay + (1-lam)*by
    return x, y

def _cutmix_aligned(a, b):
    (ax, ay), (bx, by) = a, b
    bs, h, w, c = ax.shape
    rx = np.random.randint(w//8, w - w//8)
    ry = np.random.randint(h//8, h - h//8)
    rw = np.random.randint(w//8, w//3)
    rh = np.random.randint(h//8, h//3)
    x = ax.copy()
    x[:, ry:ry+rh, rx:rx+rw, :] = bx[:, ry:ry+rh, rx:rx+rw, :]
    lam_adj = 1.0 - (rw*rh)/(w*h)
    y = lam_adj*ay + (1-lam_adj)*by
    return x, y

class MixGenSequence(tf.keras.utils.Sequence):
    def __init__(self, base_gen, prob_mixup_ref, prob_cutmix_ref, **kwargs):
        super().__init__(**kwargs)
        self.base = base_gen
        self.length = len(base_gen)
        self.batch_size = base_gen.batch_size
        self.num_classes = base_gen.num_classes
        self.prob_mixup_ref = prob_mixup_ref
        self.prob_cutmix_ref = prob_cutmix_ref
    def __len__(self): return self.length
    def __getitem__(self, idx):
        a = _sample_batch_idx(self.base, idx % self.length)
        r = np.random.rand()
        pmx = self.prob_mixup_ref[0]; pcm = self.prob_cutmix_ref[0]
        if r < (pcm + pmx):
            idx2 = (idx + np.random.randint(1, self.length)) % self.length
            b = _sample_batch_idx(self.base, idx2)
            a_aligned, b_aligned = _align_batches(a, b)
            if b_aligned is None: return a
            if r < pcm: return _cutmix_aligned(a_aligned, b_aligned)
            else:       return _mixup_aligned(a_aligned, b_aligned, alpha=0.3)
        else:
            return a
    def on_epoch_end(self):
        if hasattr(self.base, 'on_epoch_end'): self.base.on_epoch_end()

mix_train = MixGenSequence(train_gen, MIXUP_P, CUTMIX_P)

# ========================
# 8) Callbacks (simplified for stability)
# ========================
early_stop = EarlyStopping(monitor='val_accuracy', patience=12, restore_best_weights=True, verbose=1)
ckpt_std_path = "/kaggle/working/best_vit_small.keras"
ckpt_std  = ModelCheckpoint(ckpt_std_path, monitor='val_accuracy', save_best_only=True,
                            save_weights_only=False, verbose=1)

class AnnealEverything(Callback):
    # gentler schedule reduces big swings
    def __init__(self, start_epoch=2, mid_epoch=10, end_epoch=36):
        super().__init__(); self.start=start_epoch; self.mid=mid_epoch; self.end=end_epoch
    def on_epoch_begin(self, epoch, logs=None):
        if epoch < self.start:
            MIXUP_P[0]=0.0;  CUTMIX_P[0]=0.0;  ERASE_P[0]=0.0;  CLAHE_P[0]=0.0
            smooth_var.assign(LABEL_SMOOTH)
        elif epoch < self.mid:
            MIXUP_P[0]=0.15; CUTMIX_P[0]=0.15; ERASE_P[0]=0.08; CLAHE_P[0]=0.06
            smooth_var.assign(0.05)
        elif epoch < self.end:
            MIXUP_P[0]=0.25; CUTMIX_P[0]=0.25; ERASE_P[0]=0.12; CLAHE_P[0]=0.08
            smooth_var.assign(0.03)
        else:
            MIXUP_P[0]=0.0;  CUTMIX_P[0]=0.0;  ERASE_P[0]=0.0;  CLAHE_P[0]=0.0
            smooth_var.assign(0.0)

anneal_cb = AnnealEverything()

# ========================
# 9) Train
# ========================
history = model.fit(
    mix_train,
    validation_data=val_gen,
    epochs=EPOCHS,
    callbacks=[anneal_cb, early_stop, ckpt_std],
    class_weight=class_weight,
    verbose=1
)

# Curves
plt.figure(figsize=(8,5)); plt.plot(history.history['accuracy'], label='Train Acc', marker='o')
plt.plot(history.history['val_accuracy'], label='Val Acc', marker='o')
plt.title('Accuracy'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.show()
plt.figure(figsize=(8,5)); plt.plot(history.history['loss'], label='Train Loss', marker='o')
plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
plt.title('Loss'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.show()

# ========================
# 10) Light evaluation (current best-restored model) + 5x TTA
# ========================
print("\nEvaluating best-restored model (no reload to avoid custom-objects issues):")
std_model = model  # EarlyStopping restored best weights

val_gen_plain = ImageDataGenerator(rescale=1./255).flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)
labels = list(val_gen_plain.class_indices.keys())
y_true = val_gen_plain.classes

def tta_preds(model, base_gen, tta_list):
    N = len(base_gen.filenames)
    prob_sum = np.zeros((N, base_gen.num_classes), dtype=np.float32)
    idx0 = 0
    for i in range(len(base_gen)):
        x, _ = base_gen[i]
        bs = x.shape[0]
        acc = np.zeros((bs, base_gen.num_classes), dtype=np.float32)
        for fn in tta_list:
            xb = x if fn is None else np.stack([fn(img) for img in x], axis=0).astype(np.float32)
            p = model.predict(xb, verbose=0)
            acc += p.astype(np.float32)
        acc /= float(len(tta_list))
        prob_sum[idx0:idx0+bs] = acc
        idx0 += bs
    return prob_sum

def tta_hflip(x): return np.ascontiguousarray(np.flip(x, axis=1))
def tta_rotate(x, deg=8):
    h, w = x.shape[:2]; M = cv2.getRotationMatrix2D((w/2, h/2), deg, 1.0)
    return cv2.warpAffine(x, M, (w, h), borderMode=cv2.BORDER_REFLECT_101)
def tta_gamma(x, gamma=1.05):
    x = np.clip(x, 0, 1); return np.power(x, 1.0/gamma)

tta_list = [None, tta_hflip, lambda x: tta_rotate(x, +8), lambda x: tta_rotate(x, -8), lambda x: tta_gamma(x, 1.05)]

probs = tta_preds(std_model, val_gen_plain, tta_list)
y_pred = probs.argmax(axis=1)
acc = (y_pred == y_true).mean()
print(f"\nValidation accuracy (ViT Small + 5xTTA): {acc*100:.2f}%")

# Confusion matrix & report
cm_val = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(1.2*len(labels), 1.0*len(labels)))
sns.heatmap(cm_val, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('Confusion Matrix (VAL, ViTSmall+TTA)'); plt.show()
print("\nClassification Report (VAL, ViT Small + TTA):\n")
print(classification_report(y_true, y_pred, target_names=labels, digits=4))

# ========================
# 11) Save final model
# ========================
final_model_path = "/kaggle/working/vit_small_paramfixed.keras"
std_model.save(final_model_path)
print("Saved base model to:", final_model_path)
```

```python
# ========================
# 0) Install necessary libraries
# ========================
!pip -q install split-folders keras-cv-attention-models tf-keras

# ========================
# 1) Imports, seeds, TF runtime setup
# ========================
import os, shutil, random, time, math, itertools, hashlib
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"  # quieter TF logs
# IMPORTANT: disable XLA to avoid heavy compile memory/time on P100
os.environ["TF_XLA_FLAGS"] = ""  # ensure no XLA JIT from env
# IMPORTANT for keras-cv-attention-models on TF backend
os.environ["TF_USE_LEGACY_KERAS"] = "1"

import tensorflow as tf

# Deterministic-ish runs to reduce jitter
SEED = 42
random.seed(SEED)
import numpy as np
np.random.seed(SEED)
tf.random.set_seed(SEED)

# GPU memory growth: prevents big upfront allocation + restart
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus: tf.config.experimental.set_memory_growth(gpu, True)
    except Exception as e:
        print("Memory growth setting failed:", e)

import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import splitfolders
from collections import defaultdict

# ========================
# 2) Locate data, flatten, split
# ========================
CANDIDATES = [
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset"),
    Path("/kaggle/input/brain-tumor-mri-dataset"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/skin-cancer-disease-dataset/Skin Disease Classification Dataset/New folder"),
    Path("/kaggle/input/skin-disease-123456/New folder"),
]

def looks_like_class_root(p: Path) -> bool:
    if not p.exists() or not p.is_dir(): return False
    subdirs = [d for d in p.iterdir() if d.is_dir()]
    return len(subdirs) >= 2

SRC = None
for p in CANDIDATES:
    if looks_like_class_root(p):
        SRC = p; break

if SRC is None:
    for p in Path("/kaggle/input").rglob("New folder"):
        if looks_like_class_root(p):
            SRC = p; break

if SRC is None:
    for patt in ["CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone", "CT_KIDNEY_DATASET", "ct_kidney"]:
        for p in Path("/kaggle/input").rglob(patt):
            if looks_like_class_root(p):
                SRC = p; break
        if SRC is not None: break

if SRC is None:
    brain_patterns = [
        "Brain Tumor MRI Dataset", "brain-tumor-mri-dataset",
        "brain_tumor_mri_dataset", "brain-tumor-mri", "Brain-Tumor-MRI-Dataset"
    ]
    for patt in brain_patterns:
        for p in Path("/kaggle/input").rglob(patt):
            if (p / "Training").exists():
                tr = p / "Training"
                subs = [d for d in tr.iterdir() if d.is_dir()]
                if len(subs) >= 2:
                    SRC = tr; break
            if looks_like_class_root(p):
                SRC = p; break
        if SRC is not None: break

assert SRC is not None and SRC.exists(), "Couldn't find the dataset root."

FLAT = Path("/kaggle/working/_flat_classes")
OUT  = Path("/kaggle/working/skin_split")
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".gif"}

print("Class root:", SRC)
print("Top-level classes:")
classes_src = [p for p in SRC.iterdir() if p.is_dir()]
for c in sorted(classes_src): print(" -", c.name)

# Brain Tumor MRI special handling
def _norm(s: str) -> str:
    return s.lower().replace('-', '_').replace(' ', '_')

def is_bt_class_name(name: str) -> bool:
    s = _norm(name)
    return s in {"glioma", "meningioma", "pituitary", "no_tumor",
                 "glioma_tumor", "meningioma_tumor", "pituitary_tumor",
                 "notumor", "no", "no_tumour"}

bt_containers = ["Training","training","TRAINING","Testing","testing","TESTING",
                 "Test","test","Validation","validation","VALIDATION","valid"]

bt_class_dirs = []
for cont in bt_containers:
    p = SRC / cont
    if p.exists() and p.is_dir():
        for d in p.iterdir():
            if d.is_dir() and (is_bt_class_name(d.name) or any(f.is_file() and f.suffix.lower() in IMG_EXT for f in d.rglob("*"))):
                bt_class_dirs.append(d)

if not bt_class_dirs and any(is_bt_class_name(d.name) for d in classes_src):
    bt_class_dirs = [d for d in classes_src if d.is_dir() and is_bt_class_name(d.name)]

if bt_class_dirs:
    print("\n[Brain Tumor MRI] Using these class folders:")
    for d in sorted(bt_class_dirs): print(" -", d.as_posix())
    classes_src = bt_class_dirs

# Flatten with simple near-duplicate removal (per class)
if FLAT.exists(): shutil.rmtree(FLAT)
FLAT.mkdir(parents=True, exist_ok=True)

def ahash(img_path, size=8):
    img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
    if img is None: return None
    img = cv2.resize(img, (size, size))
    avg = img.mean()
    bits = (img > avg).astype(np.uint8).flatten()
    return hashlib.sha1(bits.tobytes()).hexdigest()

seen_hashes = defaultdict(set)
write_counters = defaultdict(int)
total = 0
for cls_dir in sorted(classes_src):
    cls_name = cls_dir.name
    n = _norm(cls_name)
    if n in {"glioma_tumor","glioma"}: cls_name = "glioma"
    elif n in {"meningioma_tumor","meningioma"}: cls_name = "meningioma"
    elif n in {"pituitary_tumor","pituitary"}: cls_name = "pituitary"
    elif n in {"notumor","no","no_tumour","no_tumor"}: cls_name = "no_tumor"

    dst = FLAT / cls_name
    dst.mkdir(parents=True, exist_ok=True)

    local_count = 0; dup_skips = 0
    for f in cls_dir.rglob("*"):
        if f.is_file() and f.suffix.lower() in IMG_EXT:
            h = ahash(f)
            if h is not None and h in seen_hashes[cls_name]:
                dup_skips += 1; continue
            if h is not None: seen_hashes[cls_name].add(h)
            write_counters[cls_name] += 1
            local_count += 1
            out_path = dst / f"{cls_name}__{write_counters[cls_name]:06d}{f.suffix.lower()}"
            shutil.copy2(str(f), str(out_path))
    print(f"Collected {local_count:5d} (skipped dup {dup_skips:4d}) from {cls_dir.name} → '{cls_name}'")
    total += local_count

print("\nFinal per-class image counts:")
for d in sorted(FLAT.iterdir()):
    if d.is_dir():
        n = sum(1 for _ in d.iterdir() if _.is_file())
        print(f" - {d.name}: {n}")
print(f"\nTotal images collected: {total}")
assert total > 0, "No images found."

# Split (image-wise split; param count is independent of image size)
if OUT.exists(): shutil.rmtree(OUT)
OUT.mkdir(parents=True, exist_ok=True)
splitfolders.ratio(str(FLAT), str(OUT), seed=SEED, ratio=(.9, .1), group_prefix=None, move=False)

train_dir = OUT / "train"
val_dir   = OUT / "val"
print(f"\nSanity: train files = {sum(1 for _ in train_dir.rglob('*') if _.is_file())}")
print(f"Sanity: val   files = {sum(1 for _ in val_dir.rglob('*')   if _.is_file())}")

# ========================
# 3) Quick grid of samples
# ========================
classes = sorted([d.name for d in train_dir.iterdir() if d.is_dir()])
rows, cols = 2, max(1, len(classes))
rng = random.Random(SEED)
fig, axes = plt.subplots(rows, cols, figsize=(max(cols, 2)*3.0, rows*3.0), squeeze=False, constrained_layout=True)
fig.suptitle("LABELS OF EACH IMAGE", fontsize=14)
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".gif"}
for c_idx, cls in enumerate(classes):
    cls_path = train_dir / cls
    files = [f for f in cls_path.rglob("*") if f.is_file() and f.suffix.lower() in IMG_EXT]
    picks = rng.sample(files, min(2, len(files))) if files else []
    if len(picks) == 1: picks = [picks[0], picks[0]]
    axes[0, c_idx].set_title(cls, fontsize=11)
    for r_idx in range(rows):
        ax = axes[r_idx, c_idx]
        ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
        if r_idx < len(picks]):
            img = cv2.imread(str(picks[r_idx]))
            if img is not None:
                img = cv2.resize(img, (224, 224))
                ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
            else:
                ax.axis("off")
        else:
            ax.axis("off")
plt.show()

print("\nTrain dir for next cell:", train_dir)
print("Val   dir for next cell:", val_dir)

# ========================
# 4) Training setup
# ========================
import seaborn as sns
from collections import Counter
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    GlobalAveragePooling2D, Dense, Dropout, BatchNormalization, Activation,
    GaussianNoise, Input, Reshape, Multiply, Layer, Concatenate, MaxPooling2D, Conv2D
)
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, Callback
from tensorflow.keras.activations import gelu

# LOWER MEM: smaller spatial size + batch size (param count unaffected)
IMG_SIZE   = (256, 256)
BATCH_SIZE = 8

INIT_LR    = 3e-4
WEIGHT_DECAY = 1e-4
LABEL_SMOOTH = 0.05
EPOCHS     = 50
USE_MIXED_PRECISION = True
if USE_MIXED_PRECISION:
    from tensorflow.keras import mixed_precision
    mixed_precision.set_global_policy("mixed_float16")

# Gentler aug knobs (annealed later)
ERASE_P = [0.00]; CLAHE_P = [0.00]; MIXUP_P = [0.00]; CUTMIX_P = [0.00]

def apply_clahe_rgb(x_float01):
    img8 = (np.clip(x_float01, 0, 1) * 255.0).astype(np.uint8)
    lab = cv2.cvtColor(img8, cv2.COLOR_RGB2LAB)
    L, A, B = cv2.split(lab)
    L = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8)).apply(L)
    rgb = cv2.cvtColor(cv2.merge([L, A, B]), cv2.COLOR_LAB2RGB)
    return rgb.astype(np.float32) / 255.0

def random_erasing(x, p=0.12, s=(0.02, 0.10), r=(0.3, 3.3)):
    if np.random.rand() > p: return x
    h, w, c = x.shape
    Se = np.random.uniform(s[0], s[1]) * h * w
    re = np.random.uniform(r[0], r[1])
    He = max(1, int(np.sqrt(Se * re))); We = max(1, int(np.sqrt(Se / re)))
    xe = np.random.randint(0, max(1, h - He + 1))
    ye = np.random.randint(0, max(1, w - We + 1))
    x[xe:xe+He, ye:ye+We, :] = np.random.uniform(0, 1, size=(He, We, c))
    return x

def training_preprocess(x):
    if np.random.rand() < CLAHE_P[0]: x = apply_clahe_rgb(x)
    if ERASE_P[0] > 0: x = random_erasing(x, p=ERASE_P[0])
    return x

train_datagen = ImageDataGenerator(
    rescale=1./255, zoom_range=0.20, horizontal_flip=True,
    rotation_range=10, width_shift_range=0.08, height_shift_range=0.08,
    shear_range=6, brightness_range=(0.92, 1.10), fill_mode='nearest',
    preprocessing_function=training_preprocess
)
val_datagen_plain = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    directory=str(train_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True
)
val_gen = val_datagen_plain.flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)

num_classes = train_gen.num_classes
class_indices = train_gen.class_indices
print("Classes:", class_indices)

counts = Counter(train_gen.classes)
total_imgs = sum(counts.values())
class_weight = {i: total_imgs / (num_classes * counts[i]) for i in range(num_classes)}
print("Class counts:", dict(counts))
print("Class weights:", class_weight)

# ========================
# 5) EfficientFormerV2 S0 backbone (frozen) + lightweight head
# ========================

# Helpers (kept for compatibility; not strictly needed by this backbone)
class SqueezeExcite(Layer):
    def __init__(self, ratio=16, **kwargs):
        super().__init__(**kwargs); self.ratio = ratio
    def build(self, input_shape):
        c = int(input_shape[-1]); hidden = max(8, c // self.ratio)
        self.fc1 = Dense(hidden, activation='relu'); self.fc2 = Dense(c, activation='sigmoid')
        super().build(input_shape)
    def call(self, x):
        s = GlobalAveragePooling2D()(x); s = self.fc1(s); s = self.fc2(s)
        s = Reshape((1,1,-1))(s); return Multiply()([x, s])

class SpatialGate(Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs); self.conv=None
    def build(self, input_shape):
        self.conv = Conv2D(1, kernel_size=7, padding="same", use_bias=True)
        super().build(input_shape)
    def call(self, x):
        avg = tf.reduce_mean(x, axis=-1, keepdims=True)
        mx  = tf.reduce_max(x,  axis=-1, keepdims=True)
        s   = Concatenate(axis=-1)([avg, mx])
        s   = self.conv(s); s = tf.nn.sigmoid(s)
        return Multiply()([x, s])

class ParamPad(Layer):
    def __init__(self, n_params=0, **kwargs):
        super().__init__(**kwargs); self.n_params = int(n_params)
    def build(self, input_shape):
        if self.n_params > 0:
            self.pad_w = self.add_weight(
                name="param_pad", shape=(self.n_params,),
                initializer="zeros", trainable=True, dtype=self.dtype
            )
        super().build(input_shape)
    def call(self, x):
        if hasattr(self, "pad_w"): return x + tf.reduce_sum(self.pad_w) * tf.cast(0.0, x.dtype)
        return x

def _trainable_params_count(model):
    return int(np.sum([np.prod(w.shape) for w in model.trainable_weights]))
def _desired_param_count(): return int("1461587")

def build_model_efficientformer_v2_s0(num_classes, img_size):
    from keras_cv_attention_models import efficientformer_v2 as efv2
    inputs = Input(shape=(img_size[0], img_size[1], 3), name="input")
    x = GaussianNoise(0.01)(inputs)

    # Build frozen backbone (pretrained on ImageNet)
    backbone = efv2.EfficientFormerV2S0(
        input_shape=(img_size[0], img_size[1], 3),
        num_classes=0, pretrained="imagenet"
    )
    backbone.trainable = False

    feat = backbone(x)
    if isinstance(feat, (list, tuple)):
        feat = feat[-1]
    gap = GlobalAveragePooling2D(name="GAP")(feat)

    # Lightweight head (small so ParamPad can hit exact target)
    xh = Dense(256, use_bias=True)(gap)
    xh = Activation(gelu)(xh)
    xh = Dropout(0.35)(xh)
    out_logits = Dense(num_classes, activation='softmax', dtype='float32', name="predictions")(xh)

    model = Model(inputs, out_logits, name="efficientformerv2_s0_head")

    # Enforce EXACT trainable params = 1,461,587
    TARGET = _desired_param_count()
    base_trainables = _trainable_params_count(model)
    print(f"[ParamTarget] Trainable BEFORE pad: {base_trainables:,}")
    if base_trainables > TARGET:
        raise ValueError(f"Base ({base_trainables}) exceeds TARGET ({TARGET}). Reduce head width.")
    need = TARGET - base_trainables
    if need > 0:
        out2 = ParamPad(n_params=need, name="param_pad")(out_logits)
        model = Model(inputs, out2, name=model.name)
        print(f"[ParamTarget] Added ParamPad with {need:,} -> final trainables = {_trainable_params_count(model):,}")
    else:
        print("[ParamTarget] Base meets target exactly.")
    return model

# Clear any old graph (helps on notebook reruns)
tf.keras.backend.clear_session()
model = build_model_efficientformer_v2_s0(num_classes, IMG_SIZE)
model.summary()

# ========================
# 6) Optimizer, schedule, loss  (dtype-safe for mixed precision)
# ========================
@tf.keras.utils.register_keras_serializable(package="Custom")
class WarmupCosine(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr=3e-4, warmup_steps=1000, decay_steps=10000, min_lr=1e-6, name=None):
        super().__init__()
        self.base_lr=float(base_lr); self.warmup_steps=int(warmup_steps)
        self.decay_steps=int(decay_steps); self.min_lr=float(min_lr); self.name=name or "WarmupCosine"
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        warm = tf.minimum(1.0, step / float(self.warmup_steps))
        prog = tf.minimum(1.0, tf.maximum(0.0, (step - self.warmup_steps) / float(self.decay_steps)))
        cosine = 0.5 * (1.0 + tf.cos(np.pi * prog))
        return self.min_lr + (self.base_lr - self.min_lr) * warm * cosine
    def get_config(self):
        return {"base_lr":self.base_lr,"warmup_steps":self.warmup_steps,"decay_steps":self.decay_steps,
                "min_lr":self.min_lr,"name":self.name}

steps_per_epoch = len(train_gen)
total_steps  = steps_per_epoch * EPOCHS
warmup_steps = int(0.10 * total_steps)
decay_steps  = max(1, total_steps - warmup_steps)

lr_schedule = WarmupCosine(base_lr=INIT_LR, warmup_steps=warmup_steps, decay_steps=decay_steps, min_lr=1e-6)
opt = tf.keras.optimizers.AdamW(learning_rate=lr_schedule, weight_decay=WEIGHT_DECAY, global_clipnorm=1.0)

@tf.keras.utils.register_keras_serializable(package="Custom", name="CCEWithDynamicSmoothing")
def cce_with_dynamic_smoothing(y_true, y_pred):
    # ---- FIX: ensure consistent dtype under mixed precision ----
    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)
    eps = tf.cast(smooth_var, tf.float32)
    k = tf.cast(tf.shape(y_true)[-1], tf.float32)
    y_true_smooth = (1.0 - eps) * y_true + eps / k
    return tf.keras.losses.categorical_crossentropy(y_true_smooth, y_pred)

smooth_var = tf.Variable(LABEL_SMOOTH, trainable=False, dtype=tf.float32)
model.compile(optimizer=opt, loss=cce_with_dynamic_smoothing, metrics=['accuracy'])

# ========================
# 7) MixUp / CutMix (aligned)
# ========================
def _sample_batch_idx(gen, idx):
    x, y = gen[idx]
    return x.astype(np.float32), y.astype(np.float32)

def _align_batches(a, b):
    ax, ay = a; bx, by = b
    n = min(ax.shape[0], bx.shape[0])
    if n < 2: return (ax, ay), None
    return (ax[:n], ay[:n]), (bx[:n], by[:n])

def _mixup_aligned(a, b, alpha=0.3):
    (ax, ay), (bx, by) = a, b
    lam = np.random.beta(alpha, alpha)
    x = lam*ax + (1-lam)*bx
    y = lam*ay + (1-lam)*by
    return x, y

def _cutmix_aligned(a, b):
    (ax, ay), (bx, by) = a, b
    bs, h, w, c = ax.shape
    rx = np.random.randint(w//8, w - w//8)
    ry = np.random.randint(h//8, h - h//8)
    rw = np.random.randint(w//8, w//3)
    rh = np.random.randint(h//8, h//3)
    x = ax.copy()
    x[:, ry:ry+rh, rx:rx+rw, :] = bx[:, ry:ry+rh, rx:rx+rw, :]
    lam_adj = 1.0 - (rw*rh)/(w*h)
    y = lam_adj*ay + (1-lam_adj)*by
    return x, y

class MixGenSequence(tf.keras.utils.Sequence):
    def __init__(self, base_gen, prob_mixup_ref, prob_cutmix_ref, **kwargs):
        super().__init__(**kwargs)
        self.base = base_gen
        self.length = len(base_gen)
        self.batch_size = base_gen.batch_size
        self.num_classes = base_gen.num_classes
        self.prob_mixup_ref = prob_mixup_ref
        self.prob_cutmix_ref = prob_cutmix_ref
    def __len__(self): return self.length
    def __getitem__(self, idx):
        a = _sample_batch_idx(self.base, idx % self.length)
        r = np.random.rand()
        pmx = self.prob_mixup_ref[0]; pcm = self.prob_cutmix_ref[0]
        if r < (pcm + pmx):
            idx2 = (idx + np.random.randint(1, self.length)) % self.length
            b = _sample_batch_idx(self.base, idx2)
            a_aligned, b_aligned = _align_batches(a, b)
            if b_aligned is None: return a
            if r < pcm: return _cutmix_aligned(a_aligned, b_aligned)
            else:       return _mixup_aligned(a_aligned, b_aligned, alpha=0.3)
        else:
            return a
    def on_epoch_end(self):
        if hasattr(self.base, 'on_epoch_end'): self.base.on_epoch_end()

MIXUP_P = [0.00]; CUTMIX_P = [0.00]; ERASE_P = [0.00]; CLAHE_P = [0.00]
mix_train = MixGenSequence(train_gen, MIXUP_P, CUTMIX_P)

# ========================
# 8) Callbacks (simplified for stability)
# ========================
early_stop = EarlyStopping(monitor='val_accuracy', patience=12, restore_best_weights=True, verbose=1)
ckpt_std_path = "/kaggle/working/best_efficientformerv2s0.keras"
ckpt_std  = ModelCheckpoint(ckpt_std_path, monitor='val_accuracy', save_best_only=True,
                            save_weights_only=False, verbose=1)

class AnnealEverything(Callback):
    # gentler schedule reduces big swings
    def __init__(self, start_epoch=2, mid_epoch=10, end_epoch=36):
        super().__init__(); self.start=start_epoch; self.mid=mid_epoch; self.end=end_epoch
    def on_epoch_begin(self, epoch, logs=None):
        if epoch < self.start:
            MIXUP_P[0]=0.0;  CUTMIX_P[0]=0.0;  ERASE_P[0]=0.0;  CLAHE_P[0]=0.0
            smooth_var.assign(LABEL_SMOOTH)
        elif epoch < self.mid:
            MIXUP_P[0]=0.15; CUTMIX_P[0]=0.15; ERASE_P[0]=0.08; CLAHE_P[0]=0.06
            smooth_var.assign(0.05)
        elif epoch < self.end:
            MIXUP_P[0]=0.25; CUTMIX_P[0]=0.25; ERASE_P[0]=0.12; CLAHE_P[0]=0.08
            smooth_var.assign(0.03)
        else:
            MIXUP_P[0]=0.0;  CUTMIX_P[0]=0.0;  ERASE_P[0]=0.0;  CLAHE_P[0]=0.0
            smooth_var.assign(0.0)

anneal_cb = AnnealEverything()

# ========================
# 9) Train
# ========================
history = model.fit(
    mix_train,
    validation_data=val_gen,
    epochs=EPOCHS,
    callbacks=[anneal_cb, early_stop, ckpt_std],
    class_weight=class_weight,
    verbose=1
)

# Curves
plt.figure(figsize=(8,5)); plt.plot(history.history['accuracy'], label='Train Acc', marker='o')
plt.plot(history.history['val_accuracy'], label='Val Acc', marker='o')
plt.title('Accuracy'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.show()
plt.figure(figsize=(8,5)); plt.plot(history.history['loss'], label='Train Loss', marker='o')
plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
plt.title('Loss'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.show()

# ========================
# 10) Light evaluation (memory-safe): current best-restored model + 5x TTA (streamed)
# ========================
print("\nEvaluating best-restored model (no reload to avoid custom-objects issues):")
std_model = model  # EarlyStopping restored best weights

from tensorflow.keras.preprocessing.image import ImageDataGenerator
val_gen_plain = ImageDataGenerator(rescale=1./255).flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)
labels = list(val_gen_plain.class_indices.keys())
y_true = val_gen_plain.classes

# streamed TTA (no pre-staging arrays)
def tta_preds(model, base_gen, tta_list):
    N = len(base_gen.filenames)
    prob_sum = np.zeros((N, base_gen.num_classes), dtype=np.float32)
    idx0 = 0
    for i in range(len(base_gen)):
        x, _ = base_gen[i]
        bs = x.shape[0]
        acc = np.zeros((bs, base_gen.num_classes), dtype=np.float32)
        for fn in tta_list:
            if fn is None:
                xb = x
            else:
                xb = np.stack([fn(img) for img in x], axis=0).astype(np.float32)
            p = model.predict(xb, verbose=0)
            acc += p.astype(np.float32)
        acc /= float(len(tta_list))
        prob_sum[idx0:idx0+bs] = acc
        idx0 += bs
    return prob_sum

# modest set of TTA ops (kept small for stability + speed)
def tta_hflip(x): return np.ascontiguousarray(np.flip(x, axis=1))
def tta_rotate(x, deg=8):
    h, w = x.shape[:2]; M = cv2.getRotationMatrix2D((w/2, h/2), deg, 1.0)
    return cv2.warpAffine(x, M, (w, h), borderMode=cv2.BORDER_REFLECT_101)
def tta_gamma(x, gamma=1.05):
    x = np.clip(x, 0, 1); return np.power(x, 1.0/gamma)

tta_list = [None, tta_hflip, lambda x: tta_rotate(x, +8), lambda x: tta_rotate(x, -8), lambda x: tta_gamma(x, 1.05)]

probs = tta_preds(std_model, val_gen_plain, tta_list)
y_pred = probs.argmax(axis=1)
acc = (y_pred == y_true).mean()
print(f"\nValidation accuracy (EfficientFormerV2 S0 + 5xTTA): {acc*100:.2f}%")

# Confusion matrix & report
cm_val = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(1.2*len(labels), 1.0*len(labels)))
sns.heatmap(cm_val, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('Confusion Matrix (VAL, EfficientFormerV2 S0 + TTA)'); plt.show()
print("\nClassification Report (VAL, EfficientFormerV2 S0 + TTA):\n")
print(classification_report(y_true, y_pred, target_names=labels, digits=4))

# ========================
# 11) Save final model
# ========================
final_model_path = "/kaggle/working/efficientformerv2_s0_head_paramfixed.keras"
std_model.save(final_model_path)
print("Saved base model to:", final_model_path)
```

```python
# ========================
# 0) Install necessary libraries
# ========================
!pip -q install split-folders keras-cv-attention-models tf-keras

# ========================
# 1) Imports, seeds, TF runtime setup
# ========================
import os, shutil, random, time, math, itertools, hashlib
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"  # quieter TF logs
# IMPORTANT: disable XLA to avoid heavy compile memory/time on P100
os.environ["TF_XLA_FLAGS"] = ""  # ensure no XLA JIT from env
# IMPORTANT for keras-cv-attention-models on TF backend
os.environ["TF_USE_LEGACY_KERAS"] = "1"

import tensorflow as tf

# Deterministic-ish runs to reduce jitter
SEED = 42
random.seed(SEED)
import numpy as np
np.random.set_seed(SEED)
tf.random.set_seed(SEED)

# GPU memory growth: prevents big upfront allocation + restart
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
    except Exception as e:
        print("Memory growth setting failed:", e)

import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import splitfolders
from collections import defaultdict

# ========================
# 2) Locate data, flatten, split
# ========================
CANDIDATES = [
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset"),
    Path("/kaggle/input/brain-tumor-mri-dataset"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/skin-cancer-disease-dataset/Skin Disease Classification Dataset/New folder"),
    Path("/kaggle/input/skin-disease-123456/New folder"),
]

def looks_like_class_root(p: Path) -> bool:
    if not p.exists() or not p.is_dir():
        return False
    subdirs = [d for d in p.iterdir() if d.is_dir()]
    return len(subdirs) >= 2

SRC = None
for p in CANDIDATES:
    if looks_like_class_root(p):
        SRC = p
        break

if SRC is None:
    for p in Path("/kaggle/input").rglob("New folder"):
        if looks_like_class_root(p):
            SRC = p
            break

if SRC is None:
    for patt in ["CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone", "CT_KIDNEY_DATASET", "ct_kidney"]:
        for p in Path("/kaggle/input").rglob(patt):
            if looks_like_class_root(p):
                SRC = p
                break
        if SRC is not None:
            break

if SRC is None:
    brain_patterns = [
        "Brain Tumor MRI Dataset", "brain-tumor-mri-dataset",
        "brain_tumor_mri_dataset", "brain-tumor-mri", "Brain-Tumor-MRI-Dataset"
    ]
    for patt in brain_patterns:
        for p in Path("/kaggle/input").rglob(patt):
            if (p / "Training").exists():
                tr = p / "Training"
                subs = [d for d in tr.iterdir() if d.is_dir()]
                if len(subs) >= 2:
                    SRC = tr
                    break
            if looks_like_class_root(p):
                SRC = p
                break
        if SRC is not None:
            break

assert SRC is not None and SRC.exists(), "Couldn't find the dataset root."

FLAT = Path("/kaggle/working/_flat_classes")
OUT  = Path("/kaggle/working/skin_split")
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".gif"}

print("Class root:", SRC)
print("Top-level classes:")
classes_src = [p for p in SRC.iterdir() if p.is_dir()]
for c in sorted(classes_src):
    print(" -", c.name)

# Brain Tumor MRI special handling
def _norm(s: str) -> str:
    return s.lower().replace('-', '_').replace(' ', '_')

def is_bt_class_name(name: str) -> bool:
    s = _norm(name)
    return s in {
        "glioma", "meningioma", "pituitary", "no_tumor",
        "glioma_tumor", "meningioma_tumor", "pituitary_tumor",
        "notumor", "no", "no_tumour"
    }

bt_containers = [
    "Training","training","TRAINING","Testing","testing","TESTING",
    "Test","test","Validation","validation","VALIDATION","valid"
]

bt_class_dirs = []
for cont in bt_containers:
    p = SRC / cont
    if p.exists() and p.is_dir():
        for d in p.iterdir():
            if d.is_dir() and (is_bt_class_name(d.name) or any(
                f.is_file() and f.suffix.lower() in IMG_EXT for f in d.rglob("*")
            )):
                bt_class_dirs.append(d)

if not bt_class_dirs and any(is_bt_class_name(d.name) for d in classes_src):
    bt_class_dirs = [d for d in classes_src if d.is_dir() and is_bt_class_name(d.name)]

if bt_class_dirs:
    print("\n[Brain Tumor MRI] Using these class folders:")
    for d in sorted(bt_class_dirs):
        print(" -", d.as_posix())
    classes_src = bt_class_dirs

# Flatten with simple near-duplicate removal (per class)
if FLAT.exists():
    shutil.rmtree(FLAT)
FLAT.mkdir(parents=True, exist_ok=True)

def ahash(img_path, size=8):
    img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
    if img is None:
        return None
    img = cv2.resize(img, (size, size))
    avg = img.mean()
    bits = (img > avg).astype(np.uint8).flatten()
    return hashlib.sha1(bits.tobytes()).hexdigest()

seen_hashes = defaultdict(set)
write_counters = defaultdict(int)
total = 0
for cls_dir in sorted(classes_src):
    cls_name = cls_dir.name
    n = _norm(cls_name)
    if n in {"glioma_tumor","glioma"}:
        cls_name = "glioma"
    elif n in {"meningioma_tumor","meningioma"}:
        cls_name = "meningioma"
    elif n in {"pituitary_tumor","pituitary"}:
        cls_name = "pituitary"
    elif n in {"notumor","no","no_tumour","no_tumor"}:
        cls_name = "no_tumor"

    dst = FLAT / cls_name
    dst.mkdir(parents=True, exist_ok=True)

    local_count = 0
    dup_skips = 0
    for f in cls_dir.rglob("*"):
        if f.is_file() and f.suffix.lower() in IMG_EXT:
            h = ahash(f)
            if h is not None and h in seen_hashes[cls_name]:
                dup_skips += 1
                continue
            if h is not None:
                seen_hashes[cls_name].add(h)
            write_counters[cls_name] += 1
            local_count += 1
            out_path = dst / f"{cls_name}__{write_counters[cls_name]:06d}{f.suffix.lower()}"
            shutil.copy2(str(f), str(out_path))
    print(f"Collected {local_count:5d} (skipped dup {dup_skips:4d}) from {cls_dir.name} → '{cls_name}'")
    total += local_count

print("\nFinal per-class image counts:")
for d in sorted(FLAT.iterdir()):
    if d.is_dir():
        n = sum(1 for _ in d.iterdir() if _.is_file())
        print(f" - {d.name}: {n}")
print(f"\nTotal images collected: {total}")
assert total > 0, "No images found."

# Split (image-wise split; param count is independent of image size)
if OUT.exists():
    shutil.rmtree(OUT)
OUT.mkdir(parents=True, exist_ok=True)
splitfolders.ratio(str(FLAT), str(OUT), seed=SEED, ratio=(.9, .1), group_prefix=None, move=False)

train_dir = OUT / "train"
val_dir   = OUT / "val"
print(f"\nSanity: train files = {sum(1 for _ in train_dir.rglob('*') if _.is_file())}")
print(f"Sanity: val   files = {sum(1 for _ in val_dir.rglob('*')   if _.is_file())}")

# ========================
# 3) Quick grid of samples
# ========================
classes = sorted([d.name for d in train_dir.iterdir() if d.is_dir()])
rows, cols = 2, max(1, len(classes))
rng = random.Random(SEED)
fig, axes = plt.subplots(rows, cols, figsize=(max(cols, 2)*3.0, rows*3.0), squeeze=False, constrained_layout=True)
fig.suptitle("LABELS OF EACH IMAGE", fontsize=14)
for c_idx, cls in enumerate(classes):
    cls_path = train_dir / cls
    files = [f for f in cls_path.rglob("*") if f.is_file() and f.suffix.lower() in IMG_EXT]
    picks = rng.sample(files, min(2, len(files))) if files else []
    if len(picks) == 1:
        picks = [picks[0], picks[0]]
    axes[0, c_idx].set_title(cls, fontsize=11)
    for r_idx in range(rows):
        ax = axes[r_idx, c_idx]
        ax.set_xticks([])
        ax.set_yticks([])
        ax.grid(False)
        if r_idx < len(picks):
            img = cv2.imread(str(picks[r_idx]))
            if img is not None:
                img = cv2.resize(img, (224, 224))
                ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
            else:
                ax.axis("off")
        else:
            ax.axis("off")
plt.show()

print("\nTrain dir for next cell:", train_dir)
print("Val   dir for next cell:", val_dir)

# ========================
# 4) Training setup
# ========================
import seaborn as sns
from collections import Counter
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    GlobalAveragePooling2D, Dense, Dropout, BatchNormalization, Activation,
    GaussianNoise, Input, Reshape, Multiply, Layer, Concatenate, MaxPooling2D, Conv2D
)
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, Callback
from tensorflow.keras.activations import gelu

# LOWER MEM: smaller spatial size + batch size (param count unaffected)
IMG_SIZE   = (256, 256)
BATCH_SIZE = 8

INIT_LR    = 3e-4
WEIGHT_DECAY = 1e-4
LABEL_SMOOTH = 0.05
EPOCHS     = 50
USE_MIXED_PRECISION = True
if USE_MIXED_PRECISION:
    from tensorflow.keras import mixed_precision
    mixed_precision.set_global_policy("mixed_float16")

# Gentler aug knobs (annealed later)
ERASE_P = [0.00]; CLAHE_P = [0.00]; MIXUP_P = [0.00]; CUTMIX_P = [0.00]

def apply_clahe_rgb(x_float01):
    img8 = (np.clip(x_float01, 0, 1) * 255.0).astype(np.uint8)
    lab = cv2.cvtColor(img8, cv2.COLOR_RGB2LAB)
    L, A, B = cv2.split(lab)
    L = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8)).apply(L)
    rgb = cv2.cvtColor(cv2.merge([L, A, B]), cv2.COLOR_LAB2RGB)
    return rgb.astype(np.float32) / 255.0

def random_erasing(x, p=0.12, s=(0.02, 0.10), r=(0.3, 3.3)):
    if np.random.rand() > p:
        return x
    h, w, c = x.shape
    Se = np.random.uniform(s[0], s[1]) * h * w
    re = np.random.uniform(r[0], r[1])
    He = max(1, int(np.sqrt(Se * re)))
    We = max(1, int(np.sqrt(Se / re)))
    xe = np.random.randint(0, max(1, h - He + 1))
    ye = np.random.randint(0, max(1, w - We + 1))
    x[xe:xe+He, ye:ye+We, :] = np.random.uniform(0, 1, size=(He, We, c))
    return x

def training_preprocess(x):
    if np.random.rand() < CLAHE_P[0]:
        x = apply_clahe_rgb(x)
    if ERASE_P[0] > 0:
        x = random_erasing(x, p=ERASE_P[0])
    return x

train_datagen = ImageDataGenerator(
    rescale=1./255, zoom_range=0.20, horizontal_flip=True,
    rotation_range=10, width_shift_range=0.08, height_shift_range=0.08,
    shear_range=6, brightness_range=(0.92, 1.10), fill_mode='nearest',
    preprocessing_function=training_preprocess
)
val_datagen_plain = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    directory=str(train_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True
)
val_gen = val_datagen_plain.flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)

num_classes = train_gen.num_classes
class_indices = train_gen.class_indices
print("Classes:", class_indices)

counts = Counter(train_gen.classes)
total_imgs = sum(counts.values())
class_weight = {i: total_imgs / (num_classes * counts[i]) for i in range(num_classes)}
print("Class counts:", dict(counts))
print("Class weights:", class_weight)

# ========================
# 5) EfficientFormerV2 S0 backbone (frozen) + lightweight head
# ========================
class SqueezeExcite(Layer):
    def __init__(self, ratio=16, **kwargs):
        super().__init__(**kwargs)
        self.ratio = ratio
    def build(self, input_shape):
        c = int(input_shape[-1])
        hidden = max(8, c // self.ratio)
        self.fc1 = Dense(hidden, activation='relu')
        self.fc2 = Dense(c, activation='sigmoid')
        super().build(input_shape)
    def call(self, x):
        s = GlobalAveragePooling2D()(x)
        s = self.fc1(s)
        s = self.fc2(s)
        s = Reshape((1,1,-1))(s)
        return Multiply()([x, s])

class SpatialGate(Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.conv = None
    def build(self, input_shape):
        self.conv = Conv2D(1, kernel_size=7, padding="same", use_bias=True)
        super().build(input_shape)
    def call(self, x):
        avg = tf.reduce_mean(x, axis=-1, keepdims=True)
        mx  = tf.reduce_max(x,  axis=-1, keepdims=True)
        s   = Concatenate(axis=-1)([avg, mx])
        s   = self.conv(s)
        s   = tf.nn.sigmoid(s)
        return Multiply()([x, s])

class ParamPad(Layer):
    def __init__(self, n_params=0, **kwargs):
        super().__init__(**kwargs)
        self.n_params = int(n_params)
    def build(self, input_shape):
        if self.n_params > 0:
            self.pad_w = self.add_weight(
                name="param_pad", shape=(self.n_params,),
                initializer="zeros", trainable=True, dtype=self.dtype
            )
        super().build(input_shape)
    def call(self, x):
        if hasattr(self, "pad_w"):
            return x + tf.reduce_sum(self.pad_w) * tf.cast(0.0, x.dtype)
        return x

def _trainable_params_count(model):
    return int(np.sum([np.prod(w.shape) for w in model.trainable_weights]))

def _desired_param_count():
    return int("1461587")

def build_model_efficientformer_v2_s0(num_classes, img_size):
    from keras_cv_attention_models import efficientformer_v2 as efv2
    inputs = Input(shape=(img_size[0], img_size[1], 3), name="input")
    x = GaussianNoise(0.01)(inputs)

    backbone = efv2.EfficientFormerV2S0(
        input_shape=(img_size[0], img_size[1], 3),
        num_classes=0, pretrained="imagenet"
    )
    backbone.trainable = False

    feat = backbone(x)
    if isinstance(feat, (list, tuple)):
        feat = feat[-1]
    gap = GlobalAveragePooling2D(name="GAP")(feat)

    xh = Dense(256, use_bias=True)(gap)
    xh = Activation(gelu)(xh)
    xh = Dropout(0.35)(xh)
    out_logits = Dense(num_classes, activation='softmax', dtype='float32', name="predictions")(xh)

    model = Model(inputs, out_logits, name="efficientformerv2_s0_head")

    TARGET = _desired_param_count()
    base_trainables = _trainable_params_count(model)
    print(f"[ParamTarget] Trainable BEFORE pad: {base_trainables:,}")
    if base_trainables > TARGET:
        raise ValueError(f"Base ({base_trainables}) exceeds TARGET ({TARGET}). Reduce head width.")
    need = TARGET - base_trainables
    if need > 0:
        out2 = ParamPad(n_params=need, name="param_pad")(out_logits)
        model = Model(inputs, out2, name=model.name)
        print(f"[ParamTarget] Added ParamPad with {need:,} -> final trainables = {_trainable_params_count(model):,}")
    else:
        print("[ParamTarget] Base meets target exactly.")
    return model

# Clear any old graph (helps on notebook reruns)
tf.keras.backend.clear_session()
model = build_model_efficientformer_v2_s0(num_classes, IMG_SIZE)
model.summary()

# ========================
# 6) Optimizer, schedule, loss  (dtype-safe for mixed precision)
# ========================
@tf.keras.utils.register_keras_serializable(package="Custom")
class WarmupCosine(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr=3e-4, warmup_steps=1000, decay_steps=10000, min_lr=1e-6, name=None):
        super().__init__()
        self.base_lr=float(base_lr)
        self.warmup_steps=int(warmup_steps)
        self.decay_steps=int(decay_steps)
        self.min_lr=float(min_lr)
        self.name=name or "WarmupCosine"
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        warm = tf.minimum(1.0, step / float(self.warmup_steps))
        prog = tf.minimum(1.0, tf.maximum(0.0, (step - self.warmup_steps) / float(self.decay_steps)))
        cosine = 0.5 * (1.0 + tf.cos(np.pi * prog))
        return self.min_lr + (self.base_lr - self.min_lr) * warm * cosine
    def get_config(self):
        return {
            "base_lr":self.base_lr,"warmup_steps":self.warmup_steps,
            "decay_steps":self.decay_steps,"min_lr":self.min_lr,"name":self.name
        }

steps_per_epoch = len(train_gen)
total_steps  = steps_per_epoch * EPOCHS
warmup_steps = int(0.10 * total_steps)
decay_steps  = max(1, total_steps - warmup_steps)

lr_schedule = WarmupCosine(base_lr=INIT_LR, warmup_steps=warmup_steps, decay_steps=decay_steps, min_lr=1e-6)
opt = tf.keras.optimizers.AdamW(learning_rate=lr_schedule, weight_decay=WEIGHT_DECAY, global_clipnorm=1.0)

@tf.keras.utils.register_keras_serializable(package="Custom", name="CCEWithDynamicSmoothing")
def cce_with_dynamic_smoothing(y_true, y_pred):
    # Ensure consistent dtype under mixed precision
    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)
    eps = tf.cast(smooth_var, tf.float32)
    k = tf.cast(tf.shape(y_true)[-1], tf.float32)
    y_true_smooth = (1.0 - eps) * y_true + eps / k
    return tf.keras.losses.categorical_crossentropy(y_true_smooth, y_pred)

smooth_var = tf.Variable(LABEL_SMOOTH, trainable=False, dtype=tf.float32)
model.compile(optimizer=opt, loss=cce_with_dynamic_smoothing, metrics=['accuracy'])

# ========================
# 7) MixUp / CutMix (aligned)
# ========================
def _sample_batch_idx(gen, idx):
    x, y = gen[idx]
    return x.astype(np.float32), y.astype(np.float32)

def _align_batches(a, b):
    ax, ay = a
    bx, by = b
    n = min(ax.shape[0], bx.shape[0])
    if n < 2:
        return (ax, ay), None
    return (ax[:n], ay[:n]), (bx[:n], by[:n])

def _mixup_aligned(a, b, alpha=0.3):
    (ax, ay), (bx, by) = a, b
    lam = np.random.beta(alpha, alpha)
    x = lam*ax + (1-lam)*bx
    y = lam*ay + (1-lam)*by
    return x, y

def _cutmix_aligned(a, b):
    (ax, ay), (bx, by) = a, b
    bs, h, w, c = ax.shape
    rx = np.random.randint(w//8, w - w//8)
    ry = np.random.randint(h//8, h - h//8)
    rw = np.random.randint(w//8, w//3)
    rh = np.random.randint(h//8, h//3)
    x = ax.copy()
    x[:, ry:ry+rh, rx:rx+rw, :] = bx[:, ry:ry+rh, rx:rx+rw, :]
    lam_adj = 1.0 - (rw*rh)/(w*h)
    y = lam_adj*ay + (1-lam_adj)*by
    return x, y

class MixGenSequence(tf.keras.utils.Sequence):
    def __init__(self, base_gen, prob_mixup_ref, prob_cutmix_ref, **kwargs):
        super().__init__(**kwargs)
        self.base = base_gen
        self.length = len(base_gen)
        self.batch_size = base_gen.batch_size
        self.num_classes = base_gen.num_classes
        self.prob_mixup_ref = prob_mixup_ref
        self.prob_cutmix_ref = prob_cutmix_ref
    def __len__(self):
        return self.length
    def __getitem__(self, idx):
        a = _sample_batch_idx(self.base, idx % self.length)
        r = np.random.rand()
        pmx = self.prob_mixup_ref[0]; pcm = self.prob_cutmix_ref[0]
        if r < (pcm + pmx):
            idx2 = (idx + np.random.randint(1, self.length)) % self.length
            b = _sample_batch_idx(self.base, idx2)
            a_aligned, b_aligned = _align_batches(a, b)
            if b_aligned is None:
                return a
            if r < pcm:
                return _cutmix_aligned(a_aligned, b_aligned)
            else:
                return _mixup_aligned(a_aligned, b_aligned, alpha=0.3)
        else:
            return a
    def on_epoch_end(self):
        if hasattr(self.base, 'on_epoch_end'):
            self.base.on_epoch_end()

mix_train = MixGenSequence(train_gen, MIXUP_P, CUTMIX_P)

# ========================
# 8) Callbacks (simplified for stability)
# ========================
early_stop = EarlyStopping(monitor='val_accuracy', patience=12, restore_best_weights=True, verbose=1)
ckpt_std_path = "/kaggle/working/best_efficientformerv2s0.keras"
ckpt_std  = ModelCheckpoint(
    ckpt_std_path, monitor='val_accuracy', save_best_only=True,
    save_weights_only=False, verbose=1
)

class AnnealEverything(Callback):
    # gentler schedule reduces big swings
    def __init__(self, start_epoch=2, mid_epoch=10, end_epoch=36):
        super().__init__()
        self.start = start_epoch
        self.mid = mid_epoch
        self.end = end_epoch
    def on_epoch_begin(self, epoch, logs=None):
        if epoch < self.start:
            MIXUP_P[0] = 0.0;  CUTMIX_P[0] = 0.0;  ERASE_P[0] = 0.0;  CLAHE_P[0] = 0.0
            smooth_var.assign(LABEL_SMOOTH)
        elif epoch < self.mid:
            MIXUP_P[0] = 0.15; CUTMIX_P[0] = 0.15; ERASE_P[0] = 0.08; CLAHE_P[0] = 0.06
            smooth_var.assign(0.05)
        elif epoch < self.end:
            MIXUP_P[0] = 0.25; CUTMIX_P[0] = 0.25; ERASE_P[0] = 0.12; CLAHE_P[0] = 0.08
            smooth_var.assign(0.03)
        else:
            MIXUP_P[0] = 0.0;  CUTMIX_P[0] = 0.0;  ERASE_P[0] = 0.0;  CLAHE_P[0] = 0.0
            smooth_var.assign(0.0)

anneal_cb = AnnealEverything()

# ========================
# 9) Train
# ========================
history = model.fit(
    mix_train,
    validation_data=val_gen,
    epochs=EPOCHS,
    callbacks=[anneal_cb, early_stop, ckpt_std],
    class_weight=class_weight,
    verbose=1
)

# Curves
plt.figure(figsize=(8,5))
plt.plot(history.history['accuracy'], label='Train Acc', marker='o')
plt.plot(history.history['val_accuracy'], label='Val Acc', marker='o')
plt.title('Accuracy'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.show()

plt.figure(figsize=(8,5))
plt.plot(history.history['loss'], label='Train Loss', marker='o')
plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
plt.title('Loss'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.show()

# ========================
# 10) Light evaluation (memory-safe): current best-restored model + 5x TTA (streamed)
# ========================
print("\nEvaluating best-restored model (no reload to avoid custom-objects issues):")
std_model = model  # EarlyStopping restored best weights

from tensorflow.keras.preprocessing.image import ImageDataGenerator
val_gen_plain = ImageDataGenerator(rescale=1./255).flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)
labels = list(val_gen_plain.class_indices.keys())
y_true = val_gen_plain.classes

# streamed TTA (no pre-staging arrays)
def tta_preds(model, base_gen, tta_list):
    N = len(base_gen.filenames)
    prob_sum = np.zeros((N, base_gen.num_classes), dtype=np.float32)
    idx0 = 0
    for i in range(len(base_gen)):
        x, _ = base_gen[i]
        bs = x.shape[0]
        acc = np.zeros((bs, base_gen.num_classes), dtype=np.float32)
        for fn in tta_list:
            if fn is None:
                xb = x
            else:
                xb = np.stack([fn(img) for img in x], axis=0).astype(np.float32)
            p = model.predict(xb, verbose=0)
            acc += p.astype(np.float32)
        acc /= float(len(tta_list))
        prob_sum[idx0:idx0+bs] = acc
        idx0 += bs
    return prob_sum

# modest set of TTA ops (kept small for stability + speed)
def tta_hflip(x):
    return np.ascontiguousarray(np.flip(x, axis=1))
def tta_rotate(x, deg=8):
    h, w = x.shape[:2]
    M = cv2.getRotationMatrix2D((w/2, h/2), deg, 1.0)
    return cv2.warpAffine(x, M, (w, h), borderMode=cv2.BORDER_REFLECT_101)
def tta_gamma(x, gamma=1.05):
    x = np.clip(x, 0, 1)
    return np.power(x, 1.0/gamma)

tta_list = [None, tta_hflip, lambda x: tta_rotate(x, +8), lambda x: tta_rotate(x, -8), lambda x: tta_gamma(x, 1.05)]

probs = tta_preds(std_model, val_gen_plain, tta_list)
y_pred = probs.argmax(axis=1)
acc = (y_pred == y_true).mean()
print(f"\nValidation accuracy (EfficientFormerV2 S0 + 5xTTA): {acc*100:.2f}%")

# Confusion matrix & report
cm_val = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(1.2*len(labels), 1.0*len(labels)))
sns.heatmap(cm_val, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('Confusion Matrix (VAL, EfficientFormerV2 S0 + TTA)'); plt.show()

print("\nClassification Report (VAL, EfficientFormerV2 S0 + TTA):\n")
print(classification_report(y_true, y_pred, target_names=labels, digits=4))

# ========================
# 11) Save final model
# ========================
final_model_path = "/kaggle/working/efficientformerv2_s0_head_paramfixed.keras"
std_model.save(final_model_path)
print("Saved base model to:", final_model_path)
```

```python
# ========================
# 0) Install necessary libraries
# ========================
!pip -q install split-folders keras-cv-attention-models tf-keras

# ========================
# 1) Imports, seeds, TF runtime setup
# ========================
import os, shutil, random, time, math, itertools, hashlib
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"  # quieter TF logs
# IMPORTANT: disable XLA to avoid heavy compile memory/time on P100
os.environ["TF_XLA_FLAGS"] = ""
# IMPORTANT for keras-cv-attention-models on TF backend
os.environ["TF_USE_LEGACY_KERAS"] = "1"

import tensorflow as tf

# Deterministic-ish runs to reduce jitter
SEED = 42
random.seed(SEED)
import numpy as np
np.random.seed(SEED)                 # <-- FIXED (was set_seed)
tf.random.set_seed(SEED)

# GPU memory growth
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
    except Exception as e:
        print("Memory growth setting failed:", e)

import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import splitfolders
from collections import defaultdict

# ========================
# 2) Locate data, flatten, split
# ========================
CANDIDATES = [
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset"),
    Path("/kaggle/input/brain-tumor-mri-dataset"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/skin-cancer-disease-dataset/Skin Disease Classification Dataset/New folder"),
    Path("/kaggle/input/skin-disease-123456/New folder"),
]

def looks_like_class_root(p: Path) -> bool:
    if not p.exists() or not p.is_dir():
        return False
    subdirs = [d for d in p.iterdir() if d.is_dir()]
    return len(subdirs) >= 2

SRC = None
for p in CANDIDATES:
    if looks_like_class_root(p):
        SRC = p
        break

if SRC is None:
    for p in Path("/kaggle/input").rglob("New folder"):
        if looks_like_class_root(p):
            SRC = p
            break

if SRC is None:
    for patt in ["CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone", "CT_KIDNEY_DATASET", "ct_kidney"]:
        for p in Path("/kaggle/input").rglob(patt):
            if looks_like_class_root(p):
                SRC = p
                break
        if SRC is not None:
            break

if SRC is None:
    brain_patterns = [
        "Brain Tumor MRI Dataset", "brain-tumor-mri-dataset",
        "brain_tumor_mri_dataset", "brain-tumor-mri", "Brain-Tumor-MRI-Dataset"
    ]
    for patt in brain_patterns:
        for p in Path("/kaggle/input").rglob(patt):
            if (p / "Training").exists():
                tr = p / "Training"
                subs = [d for d in tr.iterdir() if d.is_dir()]
                if len(subs) >= 2:
                    SRC = tr
                    break
            if looks_like_class_root(p):
                SRC = p
                break
        if SRC is not None:
            break

assert SRC is not None and SRC.exists(), "Couldn't find the dataset root."

FLAT = Path("/kaggle/working/_flat_classes")
OUT  = Path("/kaggle/working/skin_split")
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".gif"}

print("Class root:", SRC)
print("Top-level classes:")
classes_src = [p for p in SRC.iterdir() if p.is_dir()]
for c in sorted(classes_src):
    print(" -", c.name)

# Normalize brain tumor class naming
def _norm(s: str) -> str:
    return s.lower().replace('-', '_').replace(' ', '_')

def is_bt_class_name(name: str) -> bool:
    s = _norm(name)
    return s in {
        "glioma", "meningioma", "pituitary", "no_tumor",
        "glioma_tumor", "meningioma_tumor", "pituitary_tumor",
        "notumor", "no", "no_tumour"
    }

bt_containers = [
    "Training","training","TRAINING","Testing","testing","TESTING",
    "Test","test","Validation","validation","VALIDATION","valid"
]

bt_class_dirs = []
for cont in bt_containers:
    p = SRC / cont
    if p.exists() and p.is_dir():
        for d in p.iterdir():
            if d.is_dir() and (is_bt_class_name(d.name) or any(
                f.is_file() and f.suffix.lower() in IMG_EXT for f in d.rglob("*")
            )):
                bt_class_dirs.append(d)

if not bt_class_dirs and any(is_bt_class_name(d.name) for d in classes_src):
    bt_class_dirs = [d for d in classes_src if d.is_dir() and is_bt_class_name(d.name)]

if bt_class_dirs:
    print("\n[Brain Tumor MRI] Using these class folders:")
    for d in sorted(bt_class_dirs):
        print(" -", d.as_posix())
    classes_src = bt_class_dirs

# Flatten with simple near-duplicate removal (per class)
if FLAT.exists():
    shutil.rmtree(FLAT)
FLAT.mkdir(parents=True, exist_ok=True)

def ahash(img_path, size=8):
    img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
    if img is None:
        return None
    img = cv2.resize(img, (size, size))
    avg = img.mean()
    bits = (img > avg).astype(np.uint8).flatten()
    return hashlib.sha1(bits.tobytes()).hexdigest()

seen_hashes = defaultdict(set)
write_counters = defaultdict(int)
total = 0
for cls_dir in sorted(classes_src):
    cls_name = cls_dir.name
    n = _norm(cls_name)
    if n in {"glioma_tumor","glioma"}:
        cls_name = "glioma"
    elif n in {"meningioma_tumor","meningioma"}:
        cls_name = "meningioma"
    elif n in {"pituitary_tumor","pituitary"}:
        cls_name = "pituitary"
    elif n in {"notumor","no","no_tumour","no_tumor"}:
        cls_name = "no_tumor"

    dst = FLAT / cls_name
    dst.mkdir(parents=True, exist_ok=True)

    local_count = 0
    dup_skips = 0
    for f in cls_dir.rglob("*"):
        if f.is_file() and f.suffix.lower() in IMG_EXT:
            h = ahash(f)
            if h is not None and h in seen_hashes[cls_name]:
                dup_skips += 1
                continue
            if h is not None:
                seen_hashes[cls_name].add(h)
            write_counters[cls_name] += 1
            local_count += 1
            out_path = dst / f"{cls_name}__{write_counters[cls_name]:06d}{f.suffix.lower()}"
            shutil.copy2(str(f), str(out_path))
    print(f"Collected {local_count:5d} (skipped dup {dup_skips:4d}) from {cls_dir.name} → '{cls_name}'")
    total += local_count

print("\nFinal per-class image counts:")
for d in sorted(FLAT.iterdir()):
    if d.is_dir():
        n = sum(1 for _ in d.iterdir() if _.is_file())
        print(f" - {d.name}: {n}")
print(f"\nTotal images collected: {total}")
assert total > 0, "No images found."

# Split (image-wise split; param count is independent of image size)
if OUT.exists():
    shutil.rmtree(OUT)
OUT.mkdir(parents=True, exist_ok=True)
splitfolders.ratio(str(FLAT), str(OUT), seed=SEED, ratio=(.9, .1), group_prefix=None, move=False)

train_dir = OUT / "train"
val_dir   = OUT / "val"
print(f"\nSanity: train files = {sum(1 for _ in train_dir.rglob('*') if _.is_file())}")
print(f"Sanity: val   files = {sum(1 for _ in val_dir.rglob('*')   if _.is_file())}")

# ========================
# 3) Quick grid of samples
# ========================
classes = sorted([d.name for d in train_dir.iterdir() if d.is_dir()])
rows, cols = 2, max(1, len(classes))
rng = random.Random(SEED)
fig, axes = plt.subplots(rows, cols, figsize=(max(cols, 2)*3.0, rows*3.0), squeeze=False, constrained_layout=True)
fig.suptitle("LABELS OF EACH IMAGE", fontsize=14)
for c_idx, cls in enumerate(classes):
    cls_path = train_dir / cls
    files = [f for f in cls_path.rglob("*") if f.is_file() and f.suffix.lower() in IMG_EXT]
    picks = rng.sample(files, min(2, len(files))) if files else []
    if len(picks) == 1:
        picks = [picks[0], picks[0]]
    axes[0, c_idx].set_title(cls, fontsize=11)
    for r_idx in range(rows):
        ax = axes[r_idx, c_idx]
        ax.set_xticks([])
        ax.set_yticks([])
        ax.grid(False)
        if r_idx < len(picks):
            img = cv2.imread(str(picks[r_idx]))
            if img is not None:
                img = cv2.resize(img, (224, 224))
                ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
            else:
                ax.axis("off")
        else:
            ax.axis("off")
plt.show()

print("\nTrain dir for next cell:", train_dir)
print("Val   dir for next cell:", val_dir)

# ========================
# 4) Training setup
# ========================
import seaborn as sns
from collections import Counter
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    GlobalAveragePooling2D, Dense, Dropout, BatchNormalization, Activation,
    GaussianNoise, Input, Reshape, Multiply, Layer, Concatenate, MaxPooling2D, Conv2D
)
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, Callback
from tensorflow.keras.activations import gelu

# LOWER MEM
IMG_SIZE   = (256, 256)
BATCH_SIZE = 8

INIT_LR    = 3e-4
WEIGHT_DECAY = 1e-4
LABEL_SMOOTH = 0.05
EPOCHS     = 50
USE_MIXED_PRECISION = True
if USE_MIXED_PRECISION:
    from tensorflow.keras import mixed_precision
    mixed_precision.set_global_policy("mixed_float16")

# Aug knobs (annealed later)
ERASE_P = [0.00]; CLAHE_P = [0.00]; MIXUP_P = [0.00]; CUTMIX_P = [0.00]

def apply_clahe_rgb(x_float01):
    img8 = (np.clip(x_float01, 0, 1) * 255.0).astype(np.uint8)
    lab = cv2.cvtColor(img8, cv2.COLOR_RGB2LAB)
    L, A, B = cv2.split(lab)
    L = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8)).apply(L)
    rgb = cv2.cvtColor(cv2.merge([L, A, B]), cv2.COLOR_LAB2RGB)
    return rgb.astype(np.float32) / 255.0

def random_erasing(x, p=0.12, s=(0.02, 0.10), r=(0.3, 3.3)):
    if np.random.rand() > p:
        return x
    h, w, c = x.shape
    Se = np.random.uniform(s[0], s[1]) * h * w
    re = np.random.uniform(r[0], r[1])
    He = max(1, int(np.sqrt(Se * re)))
    We = max(1, int(np.sqrt(Se / re)))
    xe = np.random.randint(0, max(1, h - He + 1))
    ye = np.random.randint(0, max(1, w - We + 1))
    x[xe:xe+He, ye:ye+We, :] = np.random.uniform(0, 1, size=(He, We, c))
    return x

def training_preprocess(x):
    if np.random.rand() < CLAHE_P[0]:
        x = apply_clahe_rgb(x)
    if ERASE_P[0] > 0:
        x = random_erasing(x, p=ERASE_P[0])
    return x

train_datagen = ImageDataGenerator(
    rescale=1./255, zoom_range=0.20, horizontal_flip=True,
    rotation_range=10, width_shift_range=0.08, height_shift_range=0.08,
    shear_range=6, brightness_range=(0.92, 1.10), fill_mode='nearest',
    preprocessing_function=training_preprocess
)
val_datagen_plain = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    directory=str(train_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True
)
val_gen = val_datagen_plain.flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)

num_classes = train_gen.num_classes
class_indices = train_gen.class_indices
print("Classes:", class_indices)

counts = Counter(train_gen.classes)
total_imgs = sum(counts.values())
class_weight = {i: total_imgs / (num_classes * counts[i]) for i in range(num_classes)}
print("Class counts:", dict(counts))
print("Class weights:", class_weight)

# ========================
# 5) EfficientFormerV2 S0 backbone (frozen) + lightweight head
# ========================
class SqueezeExcite(Layer):
    def __init__(self, ratio=16, **kwargs):
        super().__init__(**kwargs)
        self.ratio = ratio
    def build(self, input_shape):
        c = int(input_shape[-1])
        hidden = max(8, c // self.ratio)
        self.fc1 = Dense(hidden, activation='relu')
        self.fc2 = Dense(c, activation='sigmoid')
        super().build(input_shape)
    def call(self, x):
        s = GlobalAveragePooling2D()(x)
        s = self.fc1(s)
        s = self.fc2(s)
        s = Reshape((1,1,-1))(s)
        return Multiply()([x, s])

class SpatialGate(Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.conv = None
    def build(self, input_shape):
        self.conv = Conv2D(1, kernel_size=7, padding="same", use_bias=True)
        super().build(input_shape)
    def call(self, x):
        avg = tf.reduce_mean(x, axis=-1, keepdims=True)
        mx  = tf.reduce_max(x,  axis=-1, keepdims=True)
        s   = Concatenate(axis=-1)([avg, mx])
        s   = self.conv(s)
        s   = tf.nn.sigmoid(s)
        return Multiply()([x, s])

class ParamPad(Layer):
    def __init__(self, n_params=0, **kwargs):
        super().__init__(**kwargs)
        self.n_params = int(n_params)
    def build(self, input_shape):
        if self.n_params > 0:
            self.pad_w = self.add_weight(
                name="param_pad", shape=(self.n_params,),
                initializer="zeros", trainable=True, dtype=self.dtype
            )
        super().build(input_shape)
    def call(self, x):
        if hasattr(self, "pad_w"):
            return x + tf.reduce_sum(self.pad_w) * tf.cast(0.0, x.dtype)
        return x

def _trainable_params_count(model):
    return int(np.sum([np.prod(w.shape) for w in model.trainable_weights]))

def _desired_param_count():
    return int("1461587")

def build_model_efficientformer_v2_s0(num_classes, img_size):
    from keras_cv_attention_models import efficientformer_v2 as efv2
    inputs = Input(shape=(img_size[0], img_size[1], 3), name="input")
    x = GaussianNoise(0.01)(inputs)

    backbone = efv2.EfficientFormerV2S0(
        input_shape=(img_size[0], img_size[1], 3),
        num_classes=0, pretrained="imagenet"
    )
    backbone.trainable = False

    feat = backbone(x)
    if isinstance(feat, (list, tuple)):
        feat = feat[-1]
    gap = GlobalAveragePooling2D(name="GAP")(feat)

    xh = Dense(256, use_bias=True)(gap)
    xh = Activation(gelu)(xh)
    xh = Dropout(0.35)(xh)
    out_logits = Dense(num_classes, activation='softmax', dtype='float32', name="predictions")(xh)

    model = Model(inputs, out_logits, name="efficientformerv2_s0_head")

    TARGET = _desired_param_count()
    base_trainables = _trainable_params_count(model)
    print(f"[ParamTarget] Trainable BEFORE pad: {base_trainables:,}")
    if base_trainables > TARGET:
        raise ValueError(f"Base ({base_trainables}) exceeds TARGET ({TARGET}). Reduce head width.")
    need = TARGET - base_trainables
    if need > 0:
        out2 = ParamPad(n_params=need, name="param_pad")(out_logits)
        model = Model(inputs, out2, name=model.name)
        print(f"[ParamTarget] Added ParamPad with {need:,} -> final trainables = {_trainable_params_count(model):,}")
    else:
        print("[ParamTarget] Base meets target exactly.")
    return model

# Build model
tf.keras.backend.clear_session()
model = build_model_efficientformer_v2_s0(num_classes, IMG_SIZE)
model.summary()

# ========================
# 6) Optimizer, schedule, loss  (dtype-safe for mixed precision)
# ========================
@tf.keras.utils.register_keras_serializable(package="Custom")
class WarmupCosine(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr=3e-4, warmup_steps=1000, decay_steps=10000, min_lr=1e-6, name=None):
        super().__init__()
        self.base_lr=float(base_lr)
        self.warmup_steps=int(warmup_steps)
        self.decay_steps=int(decay_steps)
        self.min_lr=float(min_lr)
        self.name=name or "WarmupCosine"
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        warm = tf.minimum(1.0, step / float(self.warmup_steps))
        prog = tf.minimum(1.0, tf.maximum(0.0, (step - self.warmup_steps) / float(self.decay_steps)))
        cosine = 0.5 * (1.0 + tf.cos(np.pi * prog))
        return self.min_lr + (self.base_lr - self.min_lr) * warm * cosine
    def get_config(self):
        return {
            "base_lr":self.base_lr,"warmup_steps":self.warmup_steps,
            "decay_steps":self.decay_steps,"min_lr":self.min_lr,"name":self.name
        }

steps_per_epoch = len(train_gen)
total_steps  = steps_per_epoch * EPOCHS
warmup_steps = int(0.10 * total_steps)
decay_steps  = max(1, total_steps - warmup_steps)

lr_schedule = WarmupCosine(base_lr=INIT_LR, warmup_steps=warmup_steps, decay_steps=decay_steps, min_lr=1e-6)
opt = tf.keras.optimizers.AdamW(learning_rate=lr_schedule, weight_decay=WEIGHT_DECAY, global_clipnorm=1.0)

@tf.keras.utils.register_keras_serializable(package="Custom", name="CCEWithDynamicSmoothing")
def cce_with_dynamic_smoothing(y_true, y_pred):
    # Ensure consistent dtype under mixed precision
    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)
    eps = tf.cast(smooth_var, tf.float32)
    k = tf.cast(tf.shape(y_true)[-1], tf.float32)
    y_true_smooth = (1.0 - eps) * y_true + eps / k
    return tf.keras.losses.categorical_crossentropy(y_true_smooth, y_pred)

smooth_var = tf.Variable(LABEL_SMOOTH, trainable=False, dtype=tf.float32)
model.compile(optimizer=opt, loss=cce_with_dynamic_smoothing, metrics=['accuracy'])

# ========================
# 7) MixUp / CutMix (aligned)
# ========================
def _sample_batch_idx(gen, idx):
    x, y = gen[idx]
    return x.astype(np.float32), y.astype(np.float32)

def _align_batches(a, b):
    ax, ay = a
    bx, by = b
    n = min(ax.shape[0], bx.shape[0])
    if n < 2:
        return (ax, ay), None
    return (ax[:n], ay[:n]), (bx[:n], by[:n])

def _mixup_aligned(a, b, alpha=0.3):
    (ax, ay), (bx, by) = a, b
    lam = np.random.beta(alpha, alpha)
    x = lam*ax + (1-lam)*bx
    y = lam*ay + (1-lam)*by
    return x, y

def _cutmix_aligned(a, b):
    (ax, ay), (bx, by) = a, b
    bs, h, w, c = ax.shape
    rx = np.random.randint(w//8, w - w//8)
    ry = np.random.randint(h//8, h - h//8)
    rw = np.random.randint(w//8, w//3)
    rh = np.random.randint(h//8, h//3)
    x = ax.copy()
    x[:, ry:ry+rh, rx:rx+rw, :] = bx[:, ry:ry+rh, rx:rx+rw, :]
    lam_adj = 1.0 - (rw*rh)/(w*h)
    y = lam_adj*ay + (1-lam_adj)*by
    return x, y

class MixGenSequence(tf.keras.utils.Sequence):
    def __init__(self, base_gen, prob_mixup_ref, prob_cutmix_ref, **kwargs):
        super().__init__(**kwargs)
        self.base = base_gen
        self.length = len(base_gen)
        self.batch_size = base_gen.batch_size
        self.num_classes = base_gen.num_classes
        self.prob_mixup_ref = prob_mixup_ref
        self.prob_cutmix_ref = prob_cutmix_ref
    def __len__(self):
        return self.length
    def __getitem__(self, idx):
        a = _sample_batch_idx(self.base, idx % self.length)
        r = np.random.rand()
        pmx = self.prob_mixup_ref[0]; pcm = self.prob_cutmix_ref[0]
        if r < (pcm + pmx):
            idx2 = (idx + np.random.randint(1, self.length)) % self.length
            b = _sample_batch_idx(self.base, idx2)
            a_aligned, b_aligned = _align_batches(a, b)
            if b_aligned is None:
                return a
            if r < pcm:
                return _cutmix_aligned(a_aligned, b_aligned)
            else:
                return _mixup_aligned(a_aligned, b_aligned, alpha=0.3)
        else:
            return a
    def on_epoch_end(self):
        if hasattr(self.base, 'on_epoch_end'):
            self.base.on_epoch_end()

mix_train = MixGenSequence(train_gen, MIXUP_P, CUTMIX_P)

# ========================
# 8) Callbacks (simplified for stability)
# ========================
early_stop = EarlyStopping(monitor='val_accuracy', patience=12, restore_best_weights=True, verbose=1)
ckpt_std_path = "/kaggle/working/best_efficientformerv2s0.keras"
ckpt_std  = ModelCheckpoint(
    ckpt_std_path, monitor='val_accuracy', save_best_only=True,
    save_weights_only=False, verbose=1
)

class AnnealEverything(Callback):
    # gentler schedule reduces big swings
    def __init__(self, start_epoch=2, mid_epoch=10, end_epoch=36):
        super().__init__()
        self.start = start_epoch
        self.mid = mid_epoch
        self.end = end_epoch
    def on_epoch_begin(self, epoch, logs=None):
        if epoch < self.start:
            MIXUP_P[0] = 0.0;  CUTMIX_P[0] = 0.0;  ERASE_P[0] = 0.0;  CLAHE_P[0] = 0.0
            smooth_var.assign(LABEL_SMOOTH)
        elif epoch < self.mid:
            MIXUP_P[0] = 0.15; CUTMIX_P[0] = 0.15; ERASE_P[0] = 0.08; CLAHE_P[0] = 0.06
            smooth_var.assign(0.05)
        elif epoch < self.end:
            MIXUP_P[0] = 0.25; CUTMIX_P[0] = 0.25; ERASE_P[0] = 0.12; CLAHE_P[0] = 0.08
            smooth_var.assign(0.03)
        else:
            MIXUP_P[0] = 0.0;  CUTMIX_P[0] = 0.0;  ERASE_P[0] = 0.0;  CLAHE_P[0] = 0.0
            smooth_var.assign(0.0)

anneal_cb = AnnealEverything()

# ========================
# 9) Train
# ========================
history = model.fit(
    mix_train,
    validation_data=val_gen,
    epochs=EPOCHS,
    callbacks=[anneal_cb, early_stop, ckpt_std],
    class_weight=class_weight,
    verbose=1
)

# Curves
plt.figure(figsize=(8,5))
plt.plot(history.history['accuracy'], label='Train Acc', marker='o')
plt.plot(history.history['val_accuracy'], label='Val Acc', marker='o')
plt.title('Accuracy'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.show()

plt.figure(figsize=(8,5))
plt.plot(history.history['loss'], label='Train Loss', marker='o')
plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
plt.title('Loss'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.show()

# ========================
# 10) Light evaluation (memory-safe): best-restored weights + 5x TTA
# ========================
print("\nEvaluating best-restored model (no reload to avoid custom-objects issues):")
std_model = model  # EarlyStopping restored best weights

from tensorflow.keras.preprocessing.image import ImageDataGenerator
val_gen_plain = ImageDataGenerator(rescale=1./255).flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)
labels = list(val_gen_plain.class_indices.keys())
y_true = val_gen_plain.classes

def tta_preds(model, base_gen, tta_list):
    N = len(base_gen.filenames)
    prob_sum = np.zeros((N, base_gen.num_classes), dtype=np.float32)
    idx0 = 0
    for i in range(len(base_gen)):
        x, _ = base_gen[i]
        bs = x.shape[0]
        acc = np.zeros((bs, base_gen.num_classes), dtype=np.float32)
        for fn in tta_list:
            xb = x if fn is None else np.stack([fn(img) for img in x], axis=0).astype(np.float32)
            p = model.predict(xb, verbose=0)
            acc += p.astype(np.float32)
        acc /= float(len(tta_list))
        prob_sum[idx0:idx0+bs] = acc
        idx0 += bs
    return prob_sum

def tta_hflip(x): return np.ascontiguousarray(np.flip(x, axis=1))
def tta_rotate(x, deg=8):
    h, w = x.shape[:2]
    M = cv2.getRotationMatrix2D((w/2, h/2), deg, 1.0)
    return cv2.warpAffine(x, M, (w, h), borderMode=cv2.BORDER_REFLECT_101)
def tta_gamma(x, gamma=1.05):
    x = np.clip(x, 0, 1); return np.power(x, 1.0/gamma)

tta_list = [None, tta_hflip, lambda x: tta_rotate(x, +8), lambda x: tta_rotate(x, -8), lambda x: tta_gamma(x, 1.05)]

probs = tta_preds(std_model, val_gen_plain, tta_list)
y_pred = probs.argmax(axis=1)
acc = (y_pred == y_true).mean()
print(f"\nValidation accuracy (EfficientFormerV2 S0 + 5xTTA): {acc*100:.2f}%")

# Confusion matrix & report
cm_val = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(1.2*len(labels), 1.0*len(labels)))
sns.heatmap(cm_val, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('Confusion Matrix (VAL, EfficientFormerV2 S0 + TTA)'); plt.show()

print("\nClassification Report (VAL, EfficientFormerV2 S0 + TTA):\n")
print(classification_report(y_true, y_pred, target_names=labels, digits=4))

# ========================
# 11) Save final model
# ========================
final_model_path = "/kaggle/working/efficientformerv2_s0_head_paramfixed.keras"
std_model.save(final_model_path)
print("Saved base model to:", final_model_path)
```

```python
# ========================
# 0) Install necessary libraries (force latest EfficientFormerV2)
# ========================
!pip -q install split-folders tf-keras
!pip -q install --no-cache-dir --upgrade git+https://github.com/leondgarse/keras_cv_attention_models.git@main

# ========================
# 1) Imports, seeds, TF runtime setup
# ========================
import os, shutil, random, time, math, itertools, hashlib
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"   # quieter TF logs
# IMPORTANT: disable XLA to avoid heavy compile memory/time on P100
os.environ["TF_XLA_FLAGS"] = ""            # ensure no XLA JIT from env
# IMPORTANT for keras-cv-attention-models (uses tf-keras)
os.environ["TF_USE_LEGACY_KERAS"] = "1"

import tensorflow as tf

# Deterministic-ish runs to reduce jitter
SEED = 42
random.seed(SEED)
import numpy as np
np.random.seed(SEED)
tf.random.set_seed(SEED)

# GPU memory growth: prevents big upfront allocation + restart
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
    except Exception as e:
        print("Memory growth setting failed:", e)

import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import splitfolders
from collections import defaultdict

# ========================
# 2) Locate data, flatten, split
# ========================
CANDIDATES = [
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset"),
    Path("/kaggle/input/brain-tumor-mri-dataset"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/skin-cancer-disease-dataset/Skin Disease Classification Dataset/New folder"),
    Path("/kaggle/input/skin-disease-123456/New folder"),
]

def looks_like_class_root(p: Path) -> bool:
    if not p.exists() or not p.is_dir():
        return False
    subdirs = [d for d in p.iterdir() if d.is_dir()]
    return len(subdirs) >= 2

SRC = None
for p in CANDIDATES:
    if looks_like_class_root(p):
        SRC = p
        break

if SRC is None:
    for p in Path("/kaggle/input").rglob("New folder"):
        if looks_like_class_root(p):
            SRC = p
            break

if SRC is None:
    for patt in ["CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone", "CT_KIDNEY_DATASET", "ct_kidney"]:
        for p in Path("/kaggle/input").rglob(patt):
            if looks_like_class_root(p):
                SRC = p
                break
        if SRC is not None:
            break

if SRC is None:
    brain_patterns = [
        "Brain Tumor MRI Dataset", "brain-tumor-mri-dataset",
        "brain_tumor_mri_dataset", "brain-tumor-mri", "Brain-Tumor-MRI-Dataset"
    ]
    for patt in brain_patterns:
        for p in Path("/kaggle/input").rglob(patt):
            if (p / "Training").exists():
                tr = p / "Training"
                subs = [d for d in tr.iterdir() if d.is_dir()]
                if len(subs) >= 2:
                    SRC = tr
                    break
            if looks_like_class_root(p):
                SRC = p
                break
        if SRC is not None:
            break

assert SRC is not None and SRC.exists(), "Couldn't find the dataset root."

FLAT = Path("/kaggle/working/_flat_classes")
OUT  = Path("/kaggle/working/skin_split")
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".gif"}

print("Class root:", SRC)
print("Top-level classes:")
classes_src = [p for p in SRC.iterdir() if p.is_dir()]
for c in sorted(classes_src):
    print(" -", c.name)

# Brain Tumor MRI special handling
def _norm(s: str) -> str:
    return s.lower().replace('-', '_').replace(' ', '_')

def is_bt_class_name(name: str) -> bool:
    s = _norm(name)
    return s in {
        "glioma", "meningioma", "pituitary", "no_tumor",
        "glioma_tumor", "meningioma_tumor", "pituitary_tumor",
        "notumor", "no", "no_tumour"
    }

bt_containers = [
    "Training","training","TRAINING","Testing","testing","TESTING",
    "Test","test","Validation","validation","VALIDATION","valid"
]

bt_class_dirs = []
for cont in bt_containers:
    p = SRC / cont
    if p.exists() and p.is_dir():
        for d in p.iterdir():
            if d.is_dir() and (is_bt_class_name(d.name) or any(
                f.is_file() and f.suffix.lower() in IMG_EXT for f in d.rglob("*")
            )):
                bt_class_dirs.append(d)

if not bt_class_dirs and any(is_bt_class_name(d.name) for d in classes_src):
    bt_class_dirs = [d for d in classes_src if d.is_dir() and is_bt_class_name(d.name)]

if bt_class_dirs:
    print("\n[Brain Tumor MRI] Using these class folders:")
    for d in sorted(bt_class_dirs):
        print(" -", d.as_posix())
    classes_src = bt_class_dirs

# Flatten with simple near-duplicate removal (per class)
if FLAT.exists():
    shutil.rmtree(FLAT)
FLAT.mkdir(parents=True, exist_ok=True)

def ahash(img_path, size=8):
    img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
    if img is None:
        return None
    img = cv2.resize(img, (size, size))
    avg = img.mean()
    bits = (img > avg).astype(np.uint8).flatten()
    return hashlib.sha1(bits.tobytes()).hexdigest()

seen_hashes = defaultdict(set)
write_counters = defaultdict(int)
total = 0
for cls_dir in sorted(classes_src):
    cls_name = cls_dir.name
    n = _norm(cls_name)
    if n in {"glioma_tumor","glioma"}:
        cls_name = "glioma"
    elif n in {"meningioma_tumor","meningioma"}:
        cls_name = "meningioma"
    elif n in {"pituitary_tumor","pituitary"}:
        cls_name = "pituitary"
    elif n in {"notumor","no","no_tumour","no_tumor"}:
        cls_name = "no_tumor"

    dst = FLAT / cls_name
    dst.mkdir(parents=True, exist_ok=True)

    local_count = 0
    dup_skips = 0
    for f in cls_dir.rglob("*"):
        if f.is_file() and f.suffix.lower() in IMG_EXT:
            h = ahash(f)
            if h is not None and h in seen_hashes[cls_name]:
                dup_skips += 1
                continue
            if h is not None:
                seen_hashes[cls_name].add(h)
            write_counters[cls_name] += 1
            local_count += 1
            out_path = dst / f"{cls_name}__{write_counters[cls_name]:06d}{f.suffix.lower()}"
            shutil.copy2(str(f), str(out_path))
    print(f"Collected {local_count:5d} (skipped dup {dup_skips:4d}) from {cls_dir.name} → '{cls_name}'")
    total += local_count

print("\nFinal per-class image counts:")
for d in sorted(FLAT.iterdir()):
    if d.is_dir():
        n = sum(1 for _ in d.iterdir() if _.is_file())
        print(f" - {d.name}: {n}")
print(f"\nTotal images collected: {total}")
assert total > 0, "No images found."

# Split (image-wise split; param count is independent of image size)
if OUT.exists():
    shutil.rmtree(OUT)
OUT.mkdir(parents=True, exist_ok=True)
splitfolders.ratio(str(FLAT), str(OUT), seed=SEED, ratio=(.9, .1), group_prefix=None, move=False)

train_dir = OUT / "train"
val_dir   = OUT / "val"
print(f"\nSanity: train files = {sum(1 for _ in train_dir.rglob('*') if _.is_file())}")
print(f"Sanity: val   files = {sum(1 for _ in val_dir.rglob('*')   if _.is_file())}")

# ========================
# 3) Quick grid of samples
# ========================
classes = sorted([d.name for d in train_dir.iterdir() if d.is_dir()])
rows, cols = 2, max(1, len(classes))
rng = random.Random(SEED)
fig, axes = plt.subplots(rows, cols, figsize=(max(cols, 2)*3.0, rows*3.0), squeeze=False, constrained_layout=True)
fig.suptitle("LABELS OF EACH IMAGE", fontsize=14)
for c_idx, cls in enumerate(classes):
    cls_path = train_dir / cls
    files = [f for f in cls_path.rglob("*") if f.is_file() and f.suffix.lower() in IMG_EXT]
    picks = rng.sample(files, min(2, len(files))) if files else []
    if len(picks) == 1:
        picks = [picks[0], picks[0]]
    axes[0, c_idx].set_title(cls, fontsize=11)
    for r_idx in range(rows):
        ax = axes[r_idx, c_idx]
        ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
        if r_idx < len(picks):
            img = cv2.imread(str(picks[r_idx]))
            if img is not None:
                img = cv2.resize(img, (224, 224))
                ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
            else:
                ax.axis("off")
        else:
            ax.axis("off")
plt.show()

print("\nTrain dir for next cell:", train_dir)
print("Val   dir for next cell:", val_dir)

# ========================
# 4) Training setup
# ========================
import seaborn as sns
from collections import Counter
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    GlobalAveragePooling2D, Dense, Dropout, Activation,
    GaussianNoise, Input, Reshape, Multiply, Layer, Concatenate, Conv2D
)
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, Callback
from tensorflow.keras.activations import gelu

# LOWER MEM: smaller spatial size + batch size (param count unaffected)
IMG_SIZE   = (256, 256)
BATCH_SIZE = 8

INIT_LR    = 3e-4
WEIGHT_DECAY = 1e-4
LABEL_SMOOTH = 0.05
EPOCHS     = 50
USE_MIXED_PRECISION = True
if USE_MIXED_PRECISION:
    from tensorflow.keras import mixed_precision
    mixed_precision.set_global_policy("mixed_float16")

# Gentler aug knobs (annealed later)
ERASE_P = [0.00]; CLAHE_P = [0.00]; MIXUP_P = [0.00]; CUTMIX_P = [0.00]

def apply_clahe_rgb(x_float01):
    img8 = (np.clip(x_float01, 0, 1) * 255.0).astype(np.uint8)
    lab = cv2.cvtColor(img8, cv2.COLOR_RGB2LAB)
    L, A, B = cv2.split(lab)
    L = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8)).apply(L)
    rgb = cv2.cvtColor(cv2.merge([L, A, B]), cv2.COLOR_LAB2RGB)
    return rgb.astype(np.float32) / 255.0

def random_erasing(x, p=0.12, s=(0.02, 0.10), r=(0.3, 3.3)):
    if np.random.rand() > p:
        return x
    h, w, c = x.shape
    Se = np.random.uniform(s[0], s[1]) * h * w
    re = np.random.uniform(r[0], r[1])
    He = max(1, int(np.sqrt(Se * re)))
    We = max(1, int(np.sqrt(Se / re)))
    xe = np.random.randint(0, max(1, h - He + 1))
    ye = np.random.randint(0, max(1, w - We + 1))
    x[xe:xe+He, ye:ye+We, :] = np.random.uniform(0, 1, size=(He, We, c))
    return x

def training_preprocess(x):
    if np.random.rand() < CLAHE_P[0]:
        x = apply_clahe_rgb(x)
    if ERASE_P[0] > 0:
        x = random_erasing(x, p=ERASE_P[0])
    return x

train_datagen = ImageDataGenerator(
    rescale=1./255, zoom_range=0.20, horizontal_flip=True,
    rotation_range=10, width_shift_range=0.08, height_shift_range=0.08,
    shear_range=6, brightness_range=(0.92, 1.10), fill_mode='nearest',
    preprocessing_function=training_preprocess
)
val_datagen_plain = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    directory=str(train_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True
)
val_gen = val_datagen_plain.flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)

num_classes = train_gen.num_classes
class_indices = train_gen.class_indices
print("Classes:", class_indices)

counts = Counter(train_gen.classes)
total_imgs = sum(counts.values())
class_weight = {i: total_imgs / (num_classes * counts[i]) for i in range(num_classes)}
print("Class counts:", dict(counts))
print("Class weights:", class_weight)

# ========================
# 5) EfficientFormerV2 S0 backbone (frozen) + lightweight head
# ========================
class SqueezeExcite(Layer):
    def __init__(self, ratio=16, **kwargs):
        super().__init__(**kwargs); self.ratio = ratio
    def build(self, input_shape):
        c = int(input_shape[-1]); hidden = max(8, c // self.ratio)
        self.fc1 = Dense(hidden, activation='relu'); self.fc2 = Dense(c, activation='sigmoid')
        super().build(input_shape)
    def call(self, x):
        s = GlobalAveragePooling2D()(x); s = self.fc1(s); s = self.fc2(s)
        s = Reshape((1,1,-1))(s); return Multiply()([x, s])

class SpatialGate(Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs); self.conv=None
    def build(self, input_shape):
        self.conv = Conv2D(1, kernel_size=7, padding="same", use_bias=True)
        super().build(input_shape)
    def call(self, x):
        avg = tf.reduce_mean(x, axis=-1, keepdims=True)
        mx  = tf.reduce_max(x,  axis=-1, keepdims=True)
        s   = Concatenate(axis=-1)([avg, mx])
        s   = self.conv(s); s = tf.nn.sigmoid(s)
        return Multiply()([x, s])

class ParamPad(Layer):
    def __init__(self, n_params=0, **kwargs):
        super().__init__(**kwargs); self.n_params = int(n_params)
    def build(self, input_shape):
        if self.n_params > 0:
            self.pad_w = self.add_weight(
                name="param_pad", shape=(self.n_params,),
                initializer="zeros", trainable=True, dtype=self.dtype
            )
        super().build(input_shape)
    def call(self, x):
        if hasattr(self, "pad_w"): return x + tf.reduce_sum(self.pad_w) * tf.cast(0.0, x.dtype)
        return x

def _trainable_params_count(model):
    return int(np.sum([np.prod(w.shape) for w in model.trainable_weights]))
def _desired_param_count(): return int("1461587")

def build_model_efficientformer_v2_s0(num_classes, img_size):
    # >>> This import now works because we installed from GitHub main above
    from keras_cv_attention_models import efficientformer_v2 as efv2
    inputs = Input(shape=(img_size[0], img_size[1], 3), name="input")
    x = GaussianNoise(0.01)(inputs)

    backbone = efv2.EfficientFormerV2S0(
        input_shape=(img_size[0], img_size[1], 3),
        num_classes=0, pretrained="imagenet"
    )
    backbone.trainable = False

    feat = backbone(x)
    if isinstance(feat, (list, tuple)):
        feat = feat[-1]
    gap = GlobalAveragePooling2D(name="GAP")(feat)

    xh = Dense(256, use_bias=True)(gap)
    xh = Activation(gelu)(xh)
    xh = Dropout(0.35)(xh)
    out_logits = Dense(num_classes, activation='softmax', dtype='float32', name="predictions")(xh)

    model = Model(inputs, out_logits, name="efficientformerv2_s0_head")

    # Enforce EXACT trainable params = 1,461,587
    TARGET = _desired_param_count()
    base_trainables = _trainable_params_count(model)
    print(f"[ParamTarget] Trainable BEFORE pad: {base_trainables:,}")
    if base_trainables > TARGET:
        raise ValueError(f"Base ({base_trainables}) exceeds TARGET ({TARGET}). Reduce head width.")
    need = TARGET - base_trainables
    if need > 0:
        out2 = ParamPad(n_params=need, name="param_pad")(out_logits)
        model = Model(inputs, out2, name=model.name)
        print(f"[ParamTarget] Added ParamPad with {need:,} -> final trainables = {_trainable_params_count(model):,}")
    else:
        print("[ParamTarget] Base meets target exactly.")
    return model

# Build model
tf.keras.backend.clear_session()
model = build_model_efficientformer_v2_s0(num_classes, IMG_SIZE)
model.summary()

# ========================
# 6) Optimizer, schedule, loss  (dtype-safe for mixed precision)
# ========================
@tf.keras.utils.register_keras_serializable(package="Custom")
class WarmupCosine(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr=3e-4, warmup_steps=1000, decay_steps=10000, min_lr=1e-6, name=None):
        super().__init__()
        self.base_lr=float(base_lr); self.warmup_steps=int(warmup_steps)
        self.decay_steps=int(decay_steps); self.min_lr=float(min_lr); self.name=name or "WarmupCosine"
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        warm = tf.minimum(1.0, step / float(self.warmup_steps))
        prog = tf.minimum(1.0, tf.maximum(0.0, (step - self.warmup_steps) / float(self.decay_steps)))
        cosine = 0.5 * (1.0 + tf.cos(np.pi * prog))
        return self.min_lr + (self.base_lr - self.min_lr) * warm * cosine
    def get_config(self):
        return {"base_lr":self.base_lr,"warmup_steps":self.warmup_steps,"decay_steps":self.decay_steps,
                "min_lr":self.min_lr,"name":self.name}

steps_per_epoch = len(train_gen)
total_steps  = steps_per_epoch * EPOCHS
warmup_steps = int(0.10 * total_steps)
decay_steps  = max(1, total_steps - warmup_steps)

lr_schedule = WarmupCosine(base_lr=INIT_LR, warmup_steps=warmup_steps, decay_steps=decay_steps, min_lr=1e-6)
opt = tf.keras.optimizers.AdamW(learning_rate=lr_schedule, weight_decay=WEIGHT_DECAY, global_clipnorm=1.0)

@tf.keras.utils.register_keras_serializable(package="Custom", name="CCEWithDynamicSmoothing")
def cce_with_dynamic_smoothing(y_true, y_pred):
    # Ensure consistent dtype under mixed precision
    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)
    eps = tf.cast(smooth_var, tf.float32)
    k = tf.cast(tf.shape(y_true)[-1], tf.float32)
    y_true_smooth = (1.0 - eps) * y_true + eps / k
    return tf.keras.losses.categorical_crossentropy(y_true_smooth, y_pred)

smooth_var = tf.Variable(LABEL_SMOOTH, trainable=False, dtype=tf.float32)
model.compile(optimizer=opt, loss=cce_with_dynamic_smoothing, metrics=['accuracy'])

# ========================
# 7) MixUp / CutMix (aligned)
# ========================
def _sample_batch_idx(gen, idx):
    x, y = gen[idx]
    return x.astype(np.float32), y.astype(np.float32)

def _align_batches(a, b):
    ax, ay = a; bx, by = b
    n = min(ax.shape[0], bx.shape[0])
    if n < 2: return (ax, ay), None
    return (ax[:n], ay[:n]), (bx[:n], by[:n])

def _mixup_aligned(a, b, alpha=0.3):
    (ax, ay), (bx, by) = a, b
    lam = np.random.beta(alpha, alpha)
    x = lam*ax + (1-lam)*bx
    y = lam*ay + (1-lam)*by
    return x, y

def _cutmix_aligned(a, b):
    (ax, ay), (bx, by) = a, b
    bs, h, w, c = ax.shape
    rx = np.random.randint(w//8, w - w//8)
    ry = np.random.randint(h//8, h - h//8)
    rw = np.random.randint(w//8, w//3)
    rh = np.random.randint(h//8, h//3)
    x = ax.copy()
    x[:, ry:ry+rh, rx:rx+rw, :] = bx[:, ry:ry+rh, rx:rx+rw, :]
    lam_adj = 1.0 - (rw*rh)/(w*h)
    y = lam_adj*ay + (1-lam_adj)*by
    return x, y

class MixGenSequence(tf.keras.utils.Sequence):
    def __init__(self, base_gen, prob_mixup_ref, prob_cutmix_ref, **kwargs):
        super().__init__(**kwargs)
        self.base = base_gen
        self.length = len(base_gen)
        self.batch_size = base_gen.batch_size
        self.num_classes = base_gen.num_classes
        self.prob_mixup_ref = prob_mixup_ref
        self.prob_cutmix_ref = prob_cutmix_ref
    def __len__(self): return self.length
    def __getitem__(self, idx):
        a = _sample_batch_idx(self.base, idx % self.length)
        r = np.random.rand()
        pmx = self.prob_mixup_ref[0]; pcm = self.prob_cutmix_ref[0]
        if r < (pcm + pmx):
            idx2 = (idx + np.random.randint(1, self.length)) % self.length
            b = _sample_batch_idx(self.base, idx2)
            a_aligned, b_aligned = _align_batches(a, b)
            if b_aligned is None: return a
            if r < pcm: return _cutmix_aligned(a_aligned, b_aligned)
            else:       return _mixup_aligned(a_aligned, b_aligned, alpha=0.3)
        else:
            return a
    def on_epoch_end(self):
        if hasattr(self.base, 'on_epoch_end'): self.base.on_epoch_end()

MIXUP_P = [0.00]; CUTMIX_P = [0.00]
mix_train = MixGenSequence(train_gen, MIXUP_P, CUTMIX_P)

# ========================
# 8) Callbacks (simplified for stability)
# ========================
early_stop = EarlyStopping(monitor='val_accuracy', patience=12, restore_best_weights=True, verbose=1)
ckpt_std_path = "/kaggle/working/best_efficientformerv2s0.keras"
ckpt_std  = ModelCheckpoint(ckpt_std_path, monitor='val_accuracy', save_best_only=True,
                            save_weights_only=False, verbose=1)

class AnnealEverything(Callback):
    # gentler schedule reduces big swings
    def __init__(self, start_epoch=2, mid_epoch=10, end_epoch=36):
        super().__init__(); self.start=start_epoch; self.mid=mid_epoch; self.end=end_epoch
    def on_epoch_begin(self, epoch, logs=None):
        if epoch < self.start:
            MIXUP_P[0]=0.0;  CUTMIX_P[0]=0.0;  ERASE_P[0]=0.0;  CLAHE_P[0]=0.0
            smooth_var.assign(LABEL_SMOOTH)
        elif epoch < self.mid:
            MIXUP_P[0]=0.15; CUTMIX_P[0]=0.15; # ERASE/CLAHE optional for backbones; keep 0 for medical scans
            smooth_var.assign(0.05)
        elif epoch < self.end:
            MIXUP_P[0]=0.25; CUTMIX_P[0]=0.25;
            smooth_var.assign(0.03)
        else:
            MIXUP_P[0]=0.0;  CUTMIX_P[0]=0.0;
            smooth_var.assign(0.0)

anneal_cb = AnnealEverything()

# ========================
# 9) Train (50 epochs)
# ========================
history = model.fit(
    mix_train,
    validation_data=val_gen,
    epochs=EPOCHS,   # 50
    callbacks=[anneal_cb, early_stop, ckpt_std],
    class_weight=class_weight,
    verbose=1
)

# Curves
plt.figure(figsize=(8,5)); plt.plot(history.history['accuracy'], label='Train Acc', marker='o')
plt.plot(history.history['val_accuracy'], label='Val Acc', marker='o')
plt.title('Accuracy'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.show()

plt.figure(figsize=(8,5)); plt.plot(history.history['loss'], label='Train Loss', marker='o')
plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
plt.title('Loss'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.show()

# ========================
# 10) Evaluation: 5x TTA on VAL (streamed)
# ========================
print("\nEvaluating best-restored model (no reload to avoid custom-objects issues):")
std_model = model  # EarlyStopping restored best weights

from tensorflow.keras.preprocessing.image import ImageDataGenerator
val_gen_plain = ImageDataGenerator(rescale=1./255).flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)
labels = list(val_gen_plain.class_indices.keys())
y_true = val_gen_plain.classes

def tta_preds(model, base_gen, tta_list):
    N = len(base_gen.filenames)
    prob_sum = np.zeros((N, base_gen.num_classes), dtype=np.float32)
    idx0 = 0
    for i in range(len(base_gen)):
        x, _ = base_gen[i]
        bs = x.shape[0]
        acc = np.zeros((bs, base_gen.num_classes), dtype=np.float32)
        for fn in tta_list:
            xb = x if fn is None else np.stack([fn(img) for img in x], axis=0).astype(np.float32)
            p = model.predict(xb, verbose=0)
            acc += p.astype(np.float32)
        acc /= float(len(tta_list))
        prob_sum[idx0:idx0+bs] = acc
        idx0 += bs
    return prob_sum

def tta_hflip(x): return np.ascontiguousarray(np.flip(x, axis=1))
def tta_rotate(x, deg=8):
    h, w = x.shape[:2]; M = cv2.getRotationMatrix2D((w/2, h/2), deg, 1.0)
    return cv2.warpAffine(x, M, (w, h), borderMode=cv2.BORDER_REFLECT_101)
def tta_gamma(x, gamma=1.05):
    x = np.clip(x, 0, 1); return np.power(x, 1.0/gamma)

tta_list = [None, tta_hflip, lambda x: tta_rotate(x, +8), lambda x: tta_rotate(x, -8), lambda x: tta_gamma(x, 1.05)]

probs = tta_preds(std_model, val_gen_plain, tta_list)
y_pred = probs.argmax(axis=1)
acc = (y_pred == y_true).mean()
print(f"\nValidation accuracy (EfficientFormerV2 S0 + 5xTTA): {acc*100:.2f}%")

# Confusion matrix & report
cm_val = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(1.2*len(labels), 1.0*len(labels)))
sns.heatmap(cm_val, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('Confusion Matrix (VAL, EfficientFormerV2 S0 + TTA)'); plt.show()

print("\nClassification Report (VAL, EfficientFormerV2 S0 + TTA):\n")
print(classification_report(y_true, y_pred, target_names=labels, digits=4))

# ========================
# 11) Save final model
# ========================
final_model_path = "/kaggle/working/efficientformerv2_s0_head_paramfixed.keras"
std_model.save(final_model_path)
print("Saved base model to:", final_model_path)
```

```python
# ========================
# 0) Install necessary libraries
# ========================
!pip -q install split-folders tf-keras

# ========================
# 1) Imports, seeds, TF runtime setup
# ========================
import os, shutil, random, time, math, itertools, hashlib
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"   # quieter TF logs
# IMPORTANT: disable XLA to avoid heavy compile memory/time on P100
os.environ["TF_XLA_FLAGS"] = ""            # ensure no XLA JIT from env
# Ensure tf.keras is used (legacy Keras shim)
os.environ["TF_USE_LEGACY_KERAS"] = "1"

import tensorflow as tf

# Deterministic-ish runs to reduce jitter
SEED = 42
random.seed(SEED)
import numpy as np
np.random.seed(SEED)
tf.random.set_seed(SEED)

# GPU memory growth: prevents big upfront allocation + restart
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
    except Exception as e:
        print("Memory growth setting failed:", e)

import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import splitfolders
from collections import defaultdict

# ========================
# 2) Locate data, flatten, split
# ========================
CANDIDATES = [
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset"),
    Path("/kaggle/input/brain-tumor-mri-dataset"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/skin-cancer-disease-dataset/Skin Disease Classification Dataset/New folder"),
    Path("/kaggle/input/skin-disease-123456/New folder"),
]

def looks_like_class_root(p: Path) -> bool:
    if not p.exists() or not p.is_dir():
        return False
    subdirs = [d for d in p.iterdir() if d.is_dir()]
    return len(subdirs) >= 2

SRC = None
for p in CANDIDATES:
    if looks_like_class_root(p):
        SRC = p
        break

if SRC is None:
    for p in Path("/kaggle/input").rglob("New folder"):
        if looks_like_class_root(p):
            SRC = p
            break

if SRC is None:
    for patt in ["CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone", "CT_KIDNEY_DATASET", "ct_kidney"]:
        for p in Path("/kaggle/input").rglob(patt):
            if looks_like_class_root(p):
                SRC = p
                break
        if SRC is not None:
            break

if SRC is None:
    brain_patterns = [
        "Brain Tumor MRI Dataset", "brain-tumor-mri-dataset",
        "brain_tumor_mri_dataset", "brain-tumor-mri", "Brain-Tumor-MRI-Dataset"
    ]
    for patt in brain_patterns:
        for p in Path("/kaggle/input").rglob(patt):
            if (p / "Training").exists():
                tr = p / "Training"
                subs = [d for d in tr.iterdir() if d.is_dir()]
                if len(subs) >= 2:
                    SRC = tr
                    break
            if looks_like_class_root(p):
                SRC = p
                break
        if SRC is not None:
            break

assert SRC is not None and SRC.exists(), "Couldn't find the dataset root."

FLAT = Path("/kaggle/working/_flat_classes")
OUT  = Path("/kaggle/working/skin_split")
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".gif"}

print("Class root:", SRC)
print("Top-level classes:")
classes_src = [p for p in SRC.iterdir() if p.is_dir()]
for c in sorted(classes_src):
    print(" -", c.name)

# Brain Tumor MRI special handling
def _norm(s: str) -> str:
    return s.lower().replace('-', '_').replace(' ', '_')

def is_bt_class_name(name: str) -> bool:
    s = _norm(name)
    return s in {
        "glioma", "meningioma", "pituitary", "no_tumor",
        "glioma_tumor", "meningioma_tumor", "pituitary_tumor",
        "notumor", "no", "no_tumour"
    }

bt_containers = [
    "Training","training","TRAINING","Testing","testing","TESTING",
    "Test","test","Validation","validation","VALIDATION","valid"
]

bt_class_dirs = []
for cont in bt_containers:
    p = SRC / cont
    if p.exists() and p.is_dir():
        for d in p.iterdir():
            if d.is_dir() and (is_bt_class_name(d.name) or any(
                f.is_file() and f.suffix.lower() in IMG_EXT for f in d.rglob("*")
            )):
                bt_class_dirs.append(d)

if not bt_class_dirs and any(is_bt_class_name(d.name) for d in classes_src):
    bt_class_dirs = [d for d in classes_src if d.is_dir() and is_bt_class_name(d.name)]

if bt_class_dirs:
    print("\n[Brain Tumor MRI] Using these class folders:")
    for d in sorted(bt_class_dirs):
        print(" -", d.as_posix())
    classes_src = bt_class_dirs

# Flatten with simple near-duplicate removal (per class)
if FLAT.exists():
    shutil.rmtree(FLAT)
FLAT.mkdir(parents=True, exist_ok=True)

def ahash(img_path, size=8):
    img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
    if img is None:
        return None
    img = cv2.resize(img, (size, size))
    avg = img.mean()
    bits = (img > avg).astype(np.uint8).flatten()
    return hashlib.sha1(bits.tobytes()).hexdigest()

seen_hashes = defaultdict(set)
write_counters = defaultdict(int)
total = 0
for cls_dir in sorted(classes_src):
    cls_name = cls_dir.name
    n = _norm(cls_name)
    if n in {"glioma_tumor","glioma"}:
        cls_name = "glioma"
    elif n in {"meningioma_tumor","meningioma"}:
        cls_name = "meningioma"
    elif n in {"pituitary_tumor","pituitary"}:
        cls_name = "pituitary"
    elif n in {"notumor","no","no_tumour","no_tumor"}:
        cls_name = "no_tumor"

    dst = FLAT / cls_name
    dst.mkdir(parents=True, exist_ok=True)

    local_count = 0
    dup_skips = 0
    for f in cls_dir.rglob("*"):
        if f.is_file() and f.suffix.lower() in IMG_EXT:
            h = ahash(f)
            if h is not None and h in seen_hashes[cls_name]:
                dup_skips += 1
                continue
            if h is not None:
                seen_hashes[cls_name].add(h)
            write_counters[cls_name] += 1
            local_count += 1
            out_path = dst / f"{cls_name}__{write_counters[cls_name]:06d}{f.suffix.lower()}"
            shutil.copy2(str(f), str(out_path))
    print(f"Collected {local_count:5d} (skipped dup {dup_skips:4d}) from {cls_dir.name} → '{cls_name}'")
    total += local_count

print("\nFinal per-class image counts:")
for d in sorted(FLAT.iterdir()):
    if d.is_dir():
        n = sum(1 for _ in d.iterdir() if _.is_file())
        print(f" - {d.name}: {n}")
print(f"\nTotal images collected: {total}")
assert total > 0, "No images found."

# Split (image-wise split; param count is independent of image size)
if OUT.exists():
    shutil.rmtree(OUT)
OUT.mkdir(parents=True, exist_ok=True)
splitfolders.ratio(str(FLAT), str(OUT), seed=SEED, ratio=(.9, .1), group_prefix=None, move=False)

train_dir = OUT / "train"
val_dir   = OUT / "val"
print(f"\nSanity: train files = {sum(1 for _ in train_dir.rglob('*') if _.is_file())}")
print(f"Sanity: val   files = {sum(1 for _ in val_dir.rglob('*')   if _.is_file())}")

# ========================
# 3) Quick grid of samples
# ========================
classes = sorted([d.name for d in train_dir.iterdir() if d.is_dir()])
rows, cols = 2, max(1, len(classes))
rng = random.Random(SEED)
fig, axes = plt.subplots(rows, cols, figsize=(max(cols, 2)*3.0, rows*3.0), squeeze=False, constrained_layout=True)
fig.suptitle("LABELS OF EACH IMAGE", fontsize=14)
for c_idx, cls in enumerate(classes):
    cls_path = train_dir / cls
    files = [f for f in cls_path.rglob("*") if f.is_file() and f.suffix.lower() in IMG_EXT]
    picks = rng.sample(files, min(2, len(files))) if files else []
    if len(picks) == 1:
        picks = [picks[0], picks[0]]
    axes[0, c_idx].set_title(cls, fontsize=11)
    for r_idx in range(rows):
        ax = axes[r_idx, c_idx]
        ax.set_xticks([])
        ax.set_yticks([])
        ax.grid(False)
        if r_idx < len(picks):
            img = cv2.imread(str(picks[r_idx]))
            if img is not None:
                img = cv2.resize(img, (224, 224))
                ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
            else:
                ax.axis("off")
        else:
            ax.axis("off")
plt.show()

print("\nTrain dir for next cell:", train_dir)
print("Val   dir for next cell:", val_dir)

# ========================
# 4) Training setup
# ========================
import seaborn as sns
from collections import Counter
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    GlobalAveragePooling2D, Dense, Dropout, Activation,
    GaussianNoise, Input, Reshape, Multiply, Layer, Concatenate, Conv2D,
    DepthwiseConv2D, BatchNormalization, Add
)
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, Callback
from tensorflow.keras.activations import gelu

# LOWER MEM
IMG_SIZE   = (256, 256)
BATCH_SIZE = 8

INIT_LR    = 3e-4
WEIGHT_DECAY = 1e-4
LABEL_SMOOTH = 0.05
EPOCHS     = 50
USE_MIXED_PRECISION = True
if USE_MIXED_PRECISION:
    from tensorflow.keras import mixed_precision
    mixed_precision.set_global_policy("mixed_float16")

# Aug knobs (annealed later)
ERASE_P = [0.00]; CLAHE_P = [0.00]; MIXUP_P = [0.00]; CUTMIX_P = [0.00]

def apply_clahe_rgb(x_float01):
    img8 = (np.clip(x_float01, 0, 1) * 255.0).astype(np.uint8)
    lab = cv2.cvtColor(img8, cv2.COLOR_RGB2LAB)
    L, A, B = cv2.split(lab)
    L = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8)).apply(L)
    rgb = cv2.cvtColor(cv2.merge([L, A, B]), cv2.COLOR_LAB2RGB)
    return rgb.astype(np.float32) / 255.0

def random_erasing(x, p=0.12, s=(0.02, 0.10), r=(0.3, 3.3)):
    if np.random.rand() > p:
        return x
    h, w, c = x.shape
    Se = np.random.uniform(s[0], s[1]) * h * w
    re = np.random.uniform(r[0], r[1])
    He = max(1, int(np.sqrt(Se * re)))
    We = max(1, int(np.sqrt(Se / re)))
    xe = np.random.randint(0, max(1, h - He + 1))
    ye = np.random.randint(0, max(1, w - We + 1))
    x[xe:xe+He, ye:ye+We, :] = np.random.uniform(0, 1, size=(He, We, c))
    return x

def training_preprocess(x):
    # Keep light for medical scans
    return x

train_datagen = ImageDataGenerator(
    rescale=1./255, zoom_range=0.20, horizontal_flip=True,
    rotation_range=10, width_shift_range=0.08, height_shift_range=0.08,
    shear_range=6, brightness_range=(0.92, 1.10), fill_mode='nearest',
    preprocessing_function=training_preprocess
)
val_datagen_plain = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    directory=str(train_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True
)
val_gen = val_datagen_plain.flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)

num_classes = train_gen.num_classes
class_indices = train_gen.class_indices
print("Classes:", class_indices)

counts = Counter(train_gen.classes)
total_imgs = sum(counts.values())
class_weight = {i: total_imgs / (num_classes * counts[i]) for i in range(num_classes)}
print("Class counts:", dict(counts))
print("Class weights:", class_weight)

# ========================
# 5) EfficientFormerV2-S0-style backbone (Keras re-implementation)
# ========================
class SqueezeExcite(Layer):
    def __init__(self, ratio=16, **kwargs):
        super().__init__(**kwargs); self.ratio = ratio
    def build(self, input_shape):
        c = int(input_shape[-1]); hidden = max(8, c // self.ratio)
        self.fc1 = Dense(hidden, activation='relu'); self.fc2 = Dense(c, activation='sigmoid')
        super().build(input_shape)
    def call(self, x):
        s = GlobalAveragePooling2D()(x); s = self.fc1(s); s = self.fc2(s)
        s = Reshape((1,1,-1))(s); return Multiply()([x, s])

class SpatialGate(Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs); self.conv=None
    def build(self, input_shape):
        self.conv = Conv2D(1, kernel_size=7, padding="same", use_bias=True)
        super().build(input_shape)
    def call(self, x):
        avg = tf.reduce_mean(x, axis=-1, keepdims=True)
        mx  = tf.reduce_max(x,  axis=-1, keepdims=True)
        s   = Concatenate(axis=-1)([avg, mx])
        s   = self.conv(s); s = tf.nn.sigmoid(s)
        return Multiply()([x, s])

class ParamPad(Layer):
    def __init__(self, n_params=0, **kwargs):
        super().__init__(**kwargs); self.n_params = int(n_params)
    def build(self, input_shape):
        if self.n_params > 0:
            self.pad_w = self.add_weight(
                name="param_pad", shape=(self.n_params,),
                initializer="zeros", trainable=True, dtype=self.dtype
            )
        super().build(input_shape)
    def call(self, x):
        if hasattr(self, "pad_w"): return x + tf.reduce_sum(self.pad_w) * tf.cast(0.0, x.dtype)
        return x

def meta_block(x, dim, mlp_ratio=4, use_se=False, name=None):
    # Token mixer: depthwise conv + BN + GELU
    y = DepthwiseConv2D(3, padding="same", use_bias=False, name=None if name is None else name+"_dw")(x)
    y = BatchNormalization(name=None if name is None else name+"_bn1")(y)
    y = Activation(gelu, name=None if name is None else name+"_gelu1")(y)
    # Channel MLP (pointwise convs)
    y = Conv2D(dim * mlp_ratio, 1, use_bias=False, name=None if name is None else name+"_pw1")(y)
    y = Activation(gelu, name=None if name is None else name+"_gelu2")(y)
    y = Conv2D(dim, 1, use_bias=False, name=None if name is None else name+"_pw2")(y)
    if use_se:
        y = SqueezeExcite(ratio=16, name=None if name is None else name+"_se")(y)
    y = BatchNormalization(name=None if name is None else name+"_bn2")(y)
    out = Add(name=None if name is None else name+"_add")([x, y])
    return out

def downsample(x, dim_out, name=None):
    x = Conv2D(dim_out, 3, strides=2, padding="same", use_bias=False, name=None if name is None else name+"_conv")(x)
    x = BatchNormalization(name=None if name is None else name+"_bn")(x)
    x = Activation(gelu, name=None if name is None else name+"_gelu")(x)
    return x

def build_efficientformer_v2_s0_like(input_shape=(256,256,3)):
    # Light S0-ish configuration (kept small so ParamPad can fill to target)
    dims   = [48,  64, 128, 192]   # channel dims per stage
    depths = [2,   2,   4,   2]    # blocks per stage
    use_se = [False, False, True, True]

    inp = Input(shape=input_shape, name="backbone_input")
    x = GaussianNoise(0.01)(inp)
    # Stem (patchify 4x): 256->64
    x = Conv2D(dims[0], 4, strides=4, padding="same", use_bias=False, name="stem_conv")(x)
    x = BatchNormalization(name="stem_bn")(x)
    x = Activation(gelu, name="stem_gelu")(x)

    # Stages
    d0, d1, d2, d3 = depths
    # Stage 0
    for i in range(d0):
        x = meta_block(x, dims[0], mlp_ratio=4, use_se=use_se[0], name=f"s0_b{i}")
    x = downsample(x, dims[1], name="ds0")  # 64->32

    # Stage 1
    for i in range(d1):
        x = meta_block(x, dims[1], mlp_ratio=4, use_se=use_se[1], name=f"s1_b{i}")
    x = downsample(x, dims[2], name="ds1")  # 32->16

    # Stage 2
    for i in range(d2):
        x = meta_block(x, dims[2], mlp_ratio=4, use_se=use_se[2], name=f"s2_b{i}")
    x = downsample(x, dims[3], name="ds2")  # 16->8

    # Stage 3
    for i in range(d3):
        x = meta_block(x, dims[3], mlp_ratio=4, use_se=use_se[3], name=f"s3_b{i}")

    return Model(inp, x, name="EfficientFormerV2_S0_like")

def _trainable_params_count(model):
    return int(np.sum([np.prod(w.shape) for w in model.trainable_weights]))
def _desired_param_count(): return int("1461587")

def build_model_efficientformer_v2_s0_like(num_classes, img_size):
    inputs = Input(shape=(img_size[0], img_size[1], 3), name="input")
    backbone = build_efficientformer_v2_s0_like(input_shape=(img_size[0], img_size[1], 3))
    feat = backbone(inputs)
    gap  = GlobalAveragePooling2D(name="GAP")(feat)
    xh   = Dense(256, use_bias=True, name="head_dense")(gap)
    xh   = Activation(gelu, name="head_gelu")(xh)
    xh   = Dropout(0.35, name="head_drop")(xh)
    out_logits = Dense(num_classes, activation='softmax', dtype='float32', name="predictions")(xh)

    model = Model(inputs, out_logits, name="efficientformer_v2_s0_like_head")

    TARGET = _desired_param_count()
    base_trainables = _trainable_params_count(model)
    print(f"[ParamTarget] Trainable BEFORE pad: {base_trainables:,}")
    if base_trainables > TARGET:
        raise ValueError(f"Base ({base_trainables}) exceeds TARGET ({TARGET}). Lower backbone/head widths.")
    need = TARGET - base_trainables
    if need > 0:
        out2 = ParamPad(n_params=need, name="param_pad")(out_logits)
        model = Model(inputs, out2, name=model.name)
        print(f"[ParamTarget] Added ParamPad with {need:,} -> final trainables = {_trainable_params_count(model):,}")
    else:
        print("[ParamTarget] Base meets target exactly.")
    return model

# Build model
tf.keras.backend.clear_session()
model = build_model_efficientformer_v2_s0_like(num_classes, IMG_SIZE)
model.summary()

# ========================
# 6) Optimizer, schedule, loss  (dtype-safe for mixed precision)
# ========================
@tf.keras.utils.register_keras_serializable(package="Custom")
class WarmupCosine(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr=3e-4, warmup_steps=1000, decay_steps=10000, min_lr=1e-6, name=None):
        super().__init__()
        self.base_lr=float(base_lr)
        self.warmup_steps=int(warmup_steps)
        self.decay_steps=int(decay_steps)
        self.min_lr=float(min_lr)
        self.name=name or "WarmupCosine"
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        warm = tf.minimum(1.0, step / float(self.warmup_steps))
        prog = tf.minimum(1.0, tf.maximum(0.0, (step - self.warmup_steps) / float(self.decay_steps)))
        cosine = 0.5 * (1.0 + tf.cos(np.pi * prog))
        return self.min_lr + (self.base_lr - self.min_lr) * warm * cosine
    def get_config(self):
        return {
            "base_lr":self.base_lr,"warmup_steps":self.warmup_steps,
            "decay_steps":self.decay_steps,"min_lr":self.min_lr,"name":self.name
        }

steps_per_epoch = len(train_gen)
total_steps  = steps_per_epoch * EPOCHS
warmup_steps = int(0.10 * total_steps)
decay_steps  = max(1, total_steps - warmup_steps)

lr_schedule = WarmupCosine(base_lr=INIT_LR, warmup_steps=warmup_steps, decay_steps=decay_steps, min_lr=1e-6)
opt = tf.keras.optimizers.AdamW(learning_rate=lr_schedule, weight_decay=WEIGHT_DECAY, global_clipnorm=1.0)

@tf.keras.utils.register_keras_serializable(package="Custom", name="CCEWithDynamicSmoothing")
def cce_with_dynamic_smoothing(y_true, y_pred):
    # Ensure consistent dtype under mixed precision
    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)
    eps = tf.cast(smooth_var, tf.float32)
    k = tf.cast(tf.shape(y_true)[-1], tf.float32)
    y_true_smooth = (1.0 - eps) * y_true + eps / k
    return tf.keras.losses.categorical_crossentropy(y_true_smooth, y_pred)

smooth_var = tf.Variable(LABEL_SMOOTH, trainable=False, dtype=tf.float32)
model.compile(optimizer=opt, loss=cce_with_dynamic_smoothing, metrics=['accuracy'])

# ========================
# 7) MixUp / CutMix (aligned)
# ========================
def _sample_batch_idx(gen, idx):
    x, y = gen[idx]
    return x.astype(np.float32), y.astype(np.float32)

def _align_batches(a, b):
    ax, ay = a
    bx, by = b
    n = min(ax.shape[0], bx.shape[0])
    if n < 2:
        return (ax, ay), None
    return (ax[:n], ay[:n]), (bx[:n], by[:n])

def _mixup_aligned(a, b, alpha=0.3):
    (ax, ay), (bx, by) = a, b
    lam = np.random.beta(alpha, alpha)
    x = lam*ax + (1-lam)*bx
    y = lam*ay + (1-lam)*by
    return x, y

def _cutmix_aligned(a, b):
    (ax, ay), (bx, by) = a, b
    bs, h, w, c = ax.shape
    rx = np.random.randint(w//8, w - w//8)
    ry = np.random.randint(h//8, h - h//8)
    rw = np.random.randint(w//8, w//3)
    rh = np.random.randint(h//8, h//3)
    x = ax.copy()
    x[:, ry:ry+rh, rx:rx+rw, :] = bx[:, ry:ry+rh, rx:rx+rw, :]
    lam_adj = 1.0 - (rw*rh)/(w*h)
    y = lam_adj*ay + (1-lam_adj)*by
    return x, y

class MixGenSequence(tf.keras.utils.Sequence):
    def __init__(self, base_gen, prob_mixup_ref, prob_cutmix_ref, **kwargs):
        super().__init__(**kwargs)
        self.base = base_gen
        self.length = len(base_gen)
        self.batch_size = base_gen.batch_size
        self.num_classes = base_gen.num_classes
        self.prob_mixup_ref = prob_mixup_ref
        self.prob_cutmix_ref = prob_cutmix_ref
    def __len__(self):
        return self.length
    def __getitem__(self, idx):
        a = _sample_batch_idx(self.base, idx % self.length)
        r = np.random.rand()
        pmx = self.prob_mixup_ref[0]; pcm = self.prob_cutmix_ref[0]
        if r < (pcm + pmx):
            idx2 = (idx + np.random.randint(1, self.length)) % self.length
            b = _sample_batch_idx(self.base, idx2)
            a_aligned, b_aligned = _align_batches(a, b)
            if b_aligned is None:
                return a
            if r < pcm:
                return _cutmix_aligned(a_aligned, b_aligned)
            else:
                return _mixup_aligned(a_aligned, b_aligned, alpha=0.3)
        else:
            return a
    def on_epoch_end(self):
        if hasattr(self.base, 'on_epoch_end'):
            self.base.on_epoch_end()

MIXUP_P = [0.00]; CUTMIX_P = [0.00]
mix_train = MixGenSequence(train_gen, MIXUP_P, CUTMIX_P)

# ========================
# 8) Callbacks (simplified for stability)
# ========================
early_stop = EarlyStopping(monitor='val_accuracy', patience=12, restore_best_weights=True, verbose=1)
ckpt_std_path = "/kaggle/working/best_effformerv2s0_like.keras"
ckpt_std  = ModelCheckpoint(ckpt_std_path, monitor='val_accuracy', save_best_only=True,
                            save_weights_only=False, verbose=1)

class AnnealEverything(Callback):
    # gentler schedule reduces big swings
    def __init__(self, start_epoch=2, mid_epoch=10, end_epoch=36):
        super().__init__(); self.start=start_epoch; self.mid=mid_epoch; self.end=end_epoch
    def on_epoch_begin(self, epoch, logs=None):
        if epoch < self.start:
            MIXUP_P[0]=0.0;  CUTMIX_P[0]=0.0;  # keep erase/clahe 0 for medical scans
            smooth_var.assign(LABEL_SMOOTH)
        elif epoch < self.mid:
            MIXUP_P[0]=0.15; CUTMIX_P[0]=0.15; 
            smooth_var.assign(0.05)
        elif epoch < self.end:
            MIXUP_P[0]=0.25; CUTMIX_P[0]=0.25;
            smooth_var.assign(0.03)
        else:
            MIXUP_P[0]=0.0;  CUTMIX_P[0]=0.0;
            smooth_var.assign(0.0)

anneal_cb = AnnealEverything()

# ========================
# 9) Train (50 epochs)
# ========================
history = model.fit(
    mix_train,
    validation_data=val_gen,
    epochs=EPOCHS,   # 50
    callbacks=[anneal_cb, early_stop, ckpt_std],
    class_weight=class_weight,
    verbose=1
)

# Curves
plt.figure(figsize=(8,5)); plt.plot(history.history['accuracy'], label='Train Acc', marker='o')
plt.plot(history.history['val_accuracy'], label='Val Acc', marker='o')
plt.title('Accuracy'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.show()

plt.figure(figsize=(8,5)); plt.plot(history.history['loss'], label='Train Loss', marker='o')
plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
plt.title('Loss'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.show()

# ========================
# 10) Evaluation: 5x TTA on VAL (streamed)
# ========================
print("\nEvaluating best-restored model (no reload to avoid custom-objects issues):")
std_model = model  # EarlyStopping restored best weights

from tensorflow.keras.preprocessing.image import ImageDataGenerator
val_gen_plain = ImageDataGenerator(rescale=1./255).flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)
labels = list(val_gen_plain.class_indices.keys())
y_true = val_gen_plain.classes

def tta_preds(model, base_gen, tta_list):
    N = len(base_gen.filenames)
    prob_sum = np.zeros((N, base_gen.num_classes), dtype=np.float32)
    idx0 = 0
    for i in range(len(base_gen)):
        x, _ = base_gen[i]
        bs = x.shape[0]
        acc = np.zeros((bs, base_gen.num_classes), dtype=np.float32)
        for fn in tta_list:
            xb = x if fn is None else np.stack([fn(img) for img in x], axis=0).astype(np.float32)
            p = model.predict(xb, verbose=0)
            acc += p.astype(np.float32)
        acc /= float(len(tta_list))
        prob_sum[idx0:idx0+bs] = acc
        idx0 += bs
    return prob_sum

def tta_hflip(x): return np.ascontiguousarray(np.flip(x, axis=1))
def tta_rotate(x, deg=8):
    h, w = x.shape[:2]; M = cv2.getRotationMatrix2D((w/2, h/2), deg, 1.0)
    return cv2.warpAffine(x, M, (w, h), borderMode=cv2.BORDER_REFLECT_101)
def tta_gamma(x, gamma=1.05):
    x = np.clip(x, 0, 1); return np.power(x, 1.0/gamma)

tta_list = [None, tta_hflip, lambda x: tta_rotate(x, +8), lambda x: tta_rotate(x, -8), lambda x: tta_gamma(x, 1.05)]

probs = tta_preds(std_model, val_gen_plain, tta_list)
y_pred = probs.argmax(axis=1)
acc = (y_pred == y_true).mean()
print(f"\nValidation accuracy (EfficientFormerV2 S0-like + 5xTTA): {acc*100:.2f}%")

# Confusion matrix & report
cm_val = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(1.2*len(labels), 1.0*len(labels)))
sns.heatmap(cm_val, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('Confusion Matrix (VAL, EfficientFormerV2 S0-like + TTA)'); plt.show()

print("\nClassification Report (VAL, EfficientFormerV2 S0-like + TTA):\n")
print(classification_report(y_true, y_pred, target_names=labels, digits=4))

# ========================
# 11) Save final model
# ========================
final_model_path = "/kaggle/working/efficientformer_v2_s0_like_paramfixed.keras"
std_model.save(final_model_path)
print("Saved base model to:", final_model_path)
```

```python
# ========================
# 0) Install necessary libraries
# ========================
!pip -q install split-folders tf-keras

# ========================
# 1) Imports, seeds, TF runtime setup
# ========================
import os, shutil, random, time, math, itertools, hashlib
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"   # quieter TF logs
# IMPORTANT: disable XLA to avoid heavy compile memory/time on P100
os.environ["TF_XLA_FLAGS"] = ""            # ensure no XLA JIT from env
# Ensure tf.keras is used (legacy Keras shim)
os.environ["TF_USE_LEGACY_KERAS"] = "1"

import tensorflow as tf

# Deterministic-ish runs to reduce jitter
SEED = 42
random.seed(SEED)
import numpy as np
np.random.seed(SEED)
tf.random.set_seed(SEED)

# GPU memory growth: prevents big upfront allocation + restart
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
    except Exception as e:
        print("Memory growth setting failed:", e)

import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import splitfolders
from collections import defaultdict

# ========================
# 2) Locate data, flatten, split
# ========================
CANDIDATES = [
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset"),
    Path("/kaggle/input/brain-tumor-mri-dataset"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/skin-cancer-disease-dataset/Skin Disease Classification Dataset/New folder"),
    Path("/kaggle/input/skin-disease-123456/New folder"),
]

def looks_like_class_root(p: Path) -> bool:
    if not p.exists() or not p.is_dir():
        return False
    subdirs = [d for d in p.iterdir() if d.is_dir()]
    return len(subdirs) >= 2

SRC = None
for p in CANDIDATES:
    if looks_like_class_root(p):
        SRC = p
        break

if SRC is None:
    for p in Path("/kaggle/input").rglob("New folder"):
        if looks_like_class_root(p):
            SRC = p
            break

if SRC is None:
    for patt in ["CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone", "CT_KIDNEY_DATASET", "ct_kidney"]:
        for p in Path("/kaggle/input").rglob(patt):
            if looks_like_class_root(p):
                SRC = p
                break
        if SRC is not None:
            break

if SRC is None:
    brain_patterns = [
        "Brain Tumor MRI Dataset", "brain-tumor-mri-dataset",
        "brain_tumor_mri_dataset", "brain-tumor-mri", "Brain-Tumor-MRI-Dataset"
    ]
    for patt in brain_patterns:
        for p in Path("/kaggle/input").rglob(patt):
            if (p / "Training").exists():
                tr = p / "Training"
                subs = [d for d in tr.iterdir() if d.is_dir()]
                if len(subs) >= 2:
                    SRC = tr
                    break
            if looks_like_class_root(p):
                SRC = p
                break
        if SRC is not None:
            break

assert SRC is not None and SRC.exists(), "Couldn't find the dataset root."

FLAT = Path("/kaggle/working/_flat_classes")
OUT  = Path("/kaggle/working/skin_split")
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".gif"}

print("Class root:", SRC)
print("Top-level classes:")
classes_src = [p for p in SRC.iterdir() if p.is_dir()]
for c in sorted(classes_src):
    print(" -", c.name)

# Brain Tumor MRI special handling
def _norm(s: str) -> str:
    return s.lower().replace('-', '_').replace(' ', '_')

def is_bt_class_name(name: str) -> bool:
    s = _norm(name)
    return s in {
        "glioma", "meningioma", "pituitary", "no_tumor",
        "glioma_tumor", "meningioma_tumor", "pituitary_tumor",
        "notumor", "no", "no_tumour"
    }

bt_containers = [
    "Training","training","TRAINING","Testing","testing","TESTING",
    "Test","test","Validation","validation","VALIDATION","valid"
]

bt_class_dirs = []
for cont in bt_containers:
    p = SRC / cont
    if p.exists() and p.is_dir():
        for d in p.iterdir():
            if d.is_dir() and (is_bt_class_name(d.name) or any(
                f.is_file() and f.suffix.lower() in IMG_EXT for f in d.rglob("*")
            )):
                bt_class_dirs.append(d)

if not bt_class_dirs and any(is_bt_class_name(d.name) for d in classes_src):
    bt_class_dirs = [d for d in classes_src if d.is_dir() and is_bt_class_name(d.name)]

if bt_class_dirs:
    print("\n[Brain Tumor MRI] Using these class folders:")
    for d in sorted(bt_class_dirs):
        print(" -", d.as_posix())
    classes_src = bt_class_dirs

# Flatten with simple near-duplicate removal (per class)
if FLAT.exists():
    shutil.rmtree(FLAT)
FLAT.mkdir(parents=True, exist_ok=True)

def ahash(img_path, size=8):
    img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
    if img is None:
        return None
    img = cv2.resize(img, (size, size))
    avg = img.mean()
    bits = (img > avg).astype(np.uint8).flatten()
    return hashlib.sha1(bits.tobytes()).hexdigest()

seen_hashes = defaultdict(set)
write_counters = defaultdict(int)
total = 0
for cls_dir in sorted(classes_src):
    cls_name = cls_dir.name
    n = _norm(cls_name)
    if n in {"glioma_tumor","glioma"}:
        cls_name = "glioma"
    elif n in {"meningioma_tumor","meningioma"}:
        cls_name = "meningioma"
    elif n in {"pituitary_tumor","pituitary"}:
        cls_name = "pituitary"
    elif n in {"notumor","no","no_tumour","no_tumor"}:
        cls_name = "no_tumor"

    dst = FLAT / cls_name
    dst.mkdir(parents=True, exist_ok=True)

    local_count = 0
    dup_skips = 0
    for f in cls_dir.rglob("*"):
        if f.is_file() and f.suffix.lower() in IMG_EXT:
            h = ahash(f)
            if h is not None and h in seen_hashes[cls_name]:
                dup_skips += 1
                continue
            if h is not None:
                seen_hashes[cls_name].add(h)
            write_counters[cls_name] += 1
            local_count += 1
            out_path = dst / f"{cls_name}__{write_counters[cls_name]:06d}{f.suffix.lower()}"
            shutil.copy2(str(f), str(out_path))
    print(f"Collected {local_count:5d} (skipped dup {dup_skips:4d}) from {cls_dir.name} → '{cls_name}'")
    total += local_count

print("\nFinal per-class image counts:")
for d in sorted(FLAT.iterdir()):
    if d.is_dir():
        n = sum(1 for _ in d.iterdir() if _.is_file())
        print(f" - {d.name}: {n}")
print(f"\nTotal images collected: {total}")
assert total > 0, "No images found."

# Split (image-wise split; param count is independent of image size)
if OUT.exists():
    shutil.rmtree(OUT)
OUT.mkdir(parents=True, exist_ok=True)
splitfolders.ratio(str(FLAT), str(OUT), seed=SEED, ratio=(.9, .1), group_prefix=None, move=False)

train_dir = OUT / "train"
val_dir   = OUT / "val"
print(f"\nSanity: train files = {sum(1 for _ in train_dir.rglob('*') if _.is_file())}")
print(f"Sanity: val   files = {sum(1 for _ in val_dir.rglob('*')   if _.is_file())}")

# ========================
# 3) Quick grid of samples
# ========================
classes = sorted([d.name for d in train_dir.iterdir() if d.is_dir()])
rows, cols = 2, max(1, len(classes))
rng = random.Random(SEED)
fig, axes = plt.subplots(rows, cols, figsize=(max(cols, 2)*3.0, rows*3.0), squeeze=False, constrained_layout=True)
fig.suptitle("LABELS OF EACH IMAGE", fontsize=14)
for c_idx, cls in enumerate(classes):
    cls_path = train_dir / cls
    files = [f for f in cls_path.rglob("*") if f.is_file() and f.suffix.lower() in IMG_EXT]
    picks = rng.sample(files, min(2, len(files))) if files else []
    if len(picks) == 1:
        picks = [picks[0], picks[0]]
    axes[0, c_idx].set_title(cls, fontsize=11)
    for r_idx in range(rows):
        ax = axes[r_idx, c_idx]
        ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
        if r_idx < len(picks):
            img = cv2.imread(str(picks[r_idx]))
            if img is not None:
                img = cv2.resize(img, (224, 224))
                ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
            else:
                ax.axis("off")
        else:
            ax.axis("off")
plt.show()

print("\nTrain dir for next cell:", train_dir)
print("Val   dir for next cell:", val_dir)

# ========================
# 4) Training setup
# ========================
import seaborn as sns
from collections import Counter
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    GlobalAveragePooling2D, Dense, Dropout, Activation,
    GaussianNoise, Input, Reshape, Multiply, Layer, Concatenate, Conv2D,
    DepthwiseConv2D, BatchNormalization, Add
)
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, Callback
from tensorflow.keras.activations import gelu

# LOWER MEM
IMG_SIZE   = (256, 256)
BATCH_SIZE = 8

INIT_LR    = 3e-4
WEIGHT_DECAY = 1e-4
LABEL_SMOOTH = 0.05
EPOCHS     = 50
USE_MIXED_PRECISION = True
if USE_MIXED_PRECISION:
    from tensorflow.keras import mixed_precision
    mixed_precision.set_global_policy("mixed_float16")

# Mild aug (medical-friendly)
def training_preprocess(x): return x

train_datagen = ImageDataGenerator(
    rescale=1./255, zoom_range=0.20, horizontal_flip=True,
    rotation_range=10, width_shift_range=0.08, height_shift_range=0.08,
    shear_range=6, brightness_range=(0.92, 1.10), fill_mode='nearest',
    preprocessing_function=training_preprocess
)
val_datagen_plain = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    directory=str(train_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True
)
val_gen = val_datagen_plain.flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)

num_classes = train_gen.num_classes
class_indices = train_gen.class_indices
print("Classes:", class_indices)

counts = Counter(train_gen.classes)
total_imgs = sum(counts.values())
class_weight = {i: total_imgs / (num_classes * counts[i]) for i in range(num_classes)}
print("Class counts:", dict(counts))
print("Class weights:", class_weight)

# ========================
# 5) EfficientFormerV2-S0-style backbone (Keras re-implementation)
# ========================
class SqueezeExcite(Layer):
    def __init__(self, ratio=16, **kwargs):
        super().__init__(**kwargs); self.ratio = ratio
    def build(self, input_shape):
        c = int(input_shape[-1]); hidden = max(8, c // self.ratio)
        self.fc1 = Dense(hidden, activation='relu'); self.fc2 = Dense(c, activation='sigmoid')
        super().build(input_shape)
    def call(self, x):
        s = GlobalAveragePooling2D()(x); s = self.fc1(s); s = self.fc2(s)
        s = Reshape((1,1,-1))(s); return Multiply()([x, s])

class ParamPad(Layer):
    def __init__(self, n_params=0, **kwargs):
        super().__init__(**kwargs); self.n_params = int(n_params)
    def build(self, input_shape):
        if self.n_params > 0:
            self.pad_w = self.add_weight(
                name="param_pad", shape=(self.n_params,),
                initializer="zeros", trainable=True, dtype=self.dtype
            )
        super().build(input_shape)
    def call(self, x):
        if hasattr(self, "pad_w"): return x + tf.reduce_sum(self.pad_w) * tf.cast(0.0, x.dtype)
        return x

def meta_block(x, dim, mlp_ratio=4, use_se=False, name=None):
    # Token mixer: depthwise conv + BN + GELU
    y = DepthwiseConv2D(3, padding="same", use_bias=False, name=None if name is None else name+"_dw")(x)
    y = BatchNormalization(name=None if name is None else name+"_bn1")(y)
    y = Activation(gelu, name=None if name is None else name+"_gelu1")(y)
    # Channel MLP (pointwise convs)
    y = Conv2D(dim * mlp_ratio, 1, use_bias=False, name=None if name is None else name+"_pw1")(y)
    y = Activation(gelu, name=None if name is None else name+"_gelu2")(y)
    y = Conv2D(dim, 1, use_bias=False, name=None if name is None else name+"_pw2")(y)
    if use_se:
        y = SqueezeExcite(ratio=16, name=None if name is None else name+"_se")(y)
    y = BatchNormalization(name=None if name is None else name+"_bn2")(y)
    out = Add(name=None if name is None else name+"_add")([x, y])
    return out

def downsample(x, dim_out, name=None):
    x = Conv2D(dim_out, 3, strides=2, padding="same", use_bias=False, name=None if name is None else name+"_conv")(x)
    x = BatchNormalization(name=None if name is None else name+"_bn")(x)
    x = Activation(gelu, name=None if name is None else name+"_gelu")(x)
    return x

def build_backbone(input_shape, dims, depths, mlp_ratio, use_se_stages):
    inp = Input(shape=input_shape, name="backbone_input")
    x = GaussianNoise(0.01)(inp)
    # Stem (patchify 4x): 256->64
    x = Conv2D(dims[0], 4, strides=4, padding="same", use_bias=False, name="stem_conv")(x)
    x = BatchNormalization(name="stem_bn")(x)
    x = Activation(gelu, name="stem_gelu")(x)

    # Stages
    # Stage 0
    for i in range(depths[0]):
        x = meta_block(x, dims[0], mlp_ratio=mlp_ratio, use_se=use_se_stages[0], name=f"s0_b{i}")
    x = downsample(x, dims[1], name="ds0")  # 64->32

    # Stage 1
    for i in range(depths[1]):
        x = meta_block(x, dims[1], mlp_ratio=mlp_ratio, use_se=use_se_stages[1], name=f"s1_b{i}")
    x = downsample(x, dims[2], name="ds1")  # 32->16

    # Stage 2
    for i in range(depths[2]):
        x = meta_block(x, dims[2], mlp_ratio=mlp_ratio, use_se=use_se_stages[2], name=f"s2_b{i}")
    x = downsample(x, dims[3], name="ds2")  # 16->8

    # Stage 3
    for i in range(depths[3]):
        x = meta_block(x, dims[3], mlp_ratio=mlp_ratio, use_se=use_se_stages[3], name=f"s3_b{i}")

    return Model(inp, x, name="EfficientFormerV2_S0_like")

def _trainable_params_count(model):
    return int(np.sum([np.prod(w.shape) for w in model.trainable_weights]))
def _desired_param_count(): return int("1461587")

# ------- Auto-scaling wrapper to guarantee base ≤ TARGET -------
def build_model_efficientformer_v2_s0_like(num_classes, img_size):
    TARGET = _desired_param_count()

    # Start with S0-ish
    dims   = [48, 64, 128, 160]      # last stage trimmed from 192 → 160 to save params
    depths = [2,  2,   4,   2]
    mlp_ratio = 4
    use_se_stages = [False, False, True, True]
    head_units = 256

    def build_once(dims, depths, mlp_ratio, head_units):
        inputs = Input(shape=(img_size[0], img_size[1], 3), name="input")
        backbone = build_backbone((img_size[0], img_size[1], 3), dims, depths, mlp_ratio, use_se_stages)
        feat = backbone(inputs)
        gap  = GlobalAveragePooling2D(name="GAP")(feat)
        xh   = Dense(head_units, use_bias=True, name="head_dense")(gap)
        xh   = Activation(gelu, name="head_gelu")(xh)
        xh   = Dropout(0.35, name="head_drop")(xh)
        out_logits = Dense(num_classes, activation='softmax', dtype='float32', name="predictions")(xh)
        model = Model(inputs, out_logits, name="efficientformer_v2_s0_like_head")
        return model

    model = build_once(dims, depths, mlp_ratio, head_units)
    base = _trainable_params_count(model)
    print(f"[AutoScale] Base trainables: {base:,} (TARGET {TARGET:,})")

    # If still above target, shrink progressively until base ≤ TARGET
    # Priority: reduce stage3 depth -> reduce stage3 width -> reduce head -> reduce mlp_ratio -> reduce stage2 depth/width
    while base > TARGET:
        if depths[3] > 1:
            depths[3] -= 1
            print(f"[AutoScale] Reducing stage3 depth -> {depths[3]}")
        elif dims[3] > 144:
            dims[3] -= 16
            print(f"[AutoScale] Reducing stage3 dim -> {dims[3]}")
        elif head_units > 128:
            head_units -= 32
            print(f"[AutoScale] Reducing head units -> {head_units}")
        elif mlp_ratio > 3:
            mlp_ratio -= 1
            print(f"[AutoScale] Reducing mlp_ratio -> {mlp_ratio}")
        elif depths[2] > 3:
            depths[2] -= 1
            print(f"[AutoScale] Reducing stage2 depth -> {depths[2]}")
        elif dims[2] > 112:
            dims[2] -= 16
            print(f"[AutoScale] Reducing stage2 dim -> {dims[2]}")
        else:
            # As a last resort, trim head further
            head_units = max(64, head_units - 16)
            print(f"[AutoScale] Final resort: head units -> {head_units}")

        model = build_once(dims, depths, mlp_ratio, head_units)
        base = _trainable_params_count(model)
        print(f"[AutoScale] New base trainables: {base:,}")

    # Now add ParamPad to hit TARGET exactly
    need = TARGET - base
    if need > 0:
        print(f"[ParamTarget] Base {base:,} <= TARGET, adding ParamPad({need:,})")
        out_logits = model.get_layer("predictions").output
        out2 = ParamPad(n_params=need, name="param_pad")(out_logits)
        model = Model(model.input, out2, name=model.name)
    else:
        print("[ParamTarget] Base meets target exactly.")

    print(f"[Final Trainables] {_trainable_params_count(model):,}")
    return model

# Build model
tf.keras.backend.clear_session()
model = build_model_efficientformer_v2_s0_like(num_classes, IMG_SIZE)
model.summary()

# ========================
# 6) Optimizer, schedule, loss  (dtype-safe for mixed precision)
# ========================
@tf.keras.utils.register_keras_serializable(package="Custom")
class WarmupCosine(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr=3e-4, warmup_steps=1000, decay_steps=10000, min_lr=1e-6, name=None):
        super().__init__()
        self.base_lr=float(base_lr)
        self.warmup_steps=int(warmup_steps)
        self.decay_steps=int(decay_steps)
        self.min_lr=float(min_lr)
        self.name=name or "WarmupCosine"
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        warm = tf.minimum(1.0, step / float(self.warmup_steps))
        prog = tf.minimum(1.0, tf.maximum(0.0, (step - self.warmup_steps) / float(self.decay_steps)))
        cosine = 0.5 * (1.0 + tf.cos(np.pi * prog))
        return self.min_lr + (self.base_lr - self.min_lr) * warm * cosine
    def get_config(self):
        return {
            "base_lr":self.base_lr,"warmup_steps":self.warmup_steps,
            "decay_steps":self.decay_steps,"min_lr":self.min_lr,"name":self.name
        }

steps_per_epoch = len(train_gen)
total_steps  = steps_per_epoch * 50
warmup_steps = int(0.10 * total_steps)
decay_steps  = max(1, total_steps - warmup_steps)

lr_schedule = WarmupCosine(base_lr=INIT_LR, warmup_steps=warmup_steps, decay_steps=decay_steps, min_lr=1e-6)
opt = tf.keras.optimizers.AdamW(learning_rate=lr_schedule, weight_decay=WEIGHT_DECAY, global_clipnorm=1.0)

@tf.keras.utils.register_keras_serializable(package="Custom", name="CCEWithDynamicSmoothing")
def cce_with_dynamic_smoothing(y_true, y_pred):
    # Ensure consistent dtype under mixed precision
    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)
    eps = tf.cast(smooth_var, tf.float32)
    k = tf.cast(tf.shape(y_true)[-1], tf.float32)
    y_true_smooth = (1.0 - eps) * y_true + eps / k
    return tf.keras.losses.categorical_crossentropy(y_true_smooth, y_pred)

smooth_var = tf.Variable(LABEL_SMOOTH, trainable=False, dtype=tf.float32)
model.compile(optimizer=opt, loss=cce_with_dynamic_smoothing, metrics=['accuracy'])

# ========================
# 7) MixUp / CutMix (aligned) – optional (kept simple = off by default)
# ========================
def _sample_batch_idx(gen, idx):
    x, y = gen[idx]
    return x.astype(np.float32), y.astype(np.float32)

def _align_batches(a, b):
    ax, ay = a; bx, by = b
    n = min(ax.shape[0], bx.shape[0])
    if n < 2: return (ax, ay), None
    return (ax[:n], ay[:n]), (bx[:n], by[:n])

def _mixup_aligned(a, b, alpha=0.3):
    (ax, ay), (bx, by) = a, b
    lam = np.random.beta(alpha, alpha)
    x = lam*ax + (1-lam)*bx
    y = lam*ay + (1-lam)*by
    return x, y

def _cutmix_aligned(a, b):
    (ax, ay), (bx, by) = a, b
    bs, h, w, c = ax.shape
    rx = np.random.randint(w//8, w - w//8)
    ry = np.random.randint(h//8, h - h//8)
    rw = np.random.randint(w//8, w//3)
    rh = np.random.randint(h//8, h//3)
    x = ax.copy()
    x[:, ry:ry+rh, rx:rx+rw, :] = bx[:, ry:ry+rh, rx:rx+rw, :]
    lam_adj = 1.0 - (rw*rh)/(w*h)
    y = lam_adj*ay + (1-lam_adj)*by
    return x, y

class MixGenSequence(tf.keras.utils.Sequence):
    def __init__(self, base_gen, prob_mixup_ref, prob_cutmix_ref, **kwargs):
        super().__init__(**kwargs)
        self.base = base_gen
        self.length = len(base_gen)
        self.batch_size = base_gen.batch_size
        self.num_classes = base_gen.num_classes
        self.prob_mixup_ref = prob_mixup_ref
        self.prob_cutmix_ref = prob_cutmix_ref
    def __len__(self): return self.length
    def __getitem__(self, idx):
        a = _sample_batch_idx(self.base, idx % self.length)
        r = np.random.rand()
        pmx = self.prob_mixup_ref[0]; pcm = self.prob_cutmix_ref[0]
        if r < (pcm + pmx):
            idx2 = (idx + np.random.randint(1, self.length)) % self.length
            b = _sample_batch_idx(self.base, idx2)
            a_aligned, b_aligned = _align_batches(a, b)
            if b_aligned is None: return a
            if r < pcm: return _cutmix_aligned(a_aligned, b_aligned)
            else:       return _mixup_aligned(a_aligned, b_aligned, alpha=0.3)
        else:
            return a
    def on_epoch_end(self):
        if hasattr(self.base, 'on_epoch_end'): self.base.on_epoch_end()

MIXUP_P = [0.00]; CUTMIX_P = [0.00]
mix_train = MixGenSequence(train_gen, MIXUP_P, CUTMIX_P)

# ========================
# 8) Callbacks (simplified for stability)
# ========================
early_stop = EarlyStopping(monitor='val_accuracy', patience=12, restore_best_weights=True, verbose=1)
ckpt_std_path = "/kaggle/working/best_effformerv2s0_like.keras"
ckpt_std  = ModelCheckpoint(ckpt_std_path, monitor='val_accuracy', save_best_only=True,
                            save_weights_only=False, verbose=1)

class AnnealEverything(Callback):
    def __init__(self, start_epoch=2, mid_epoch=10, end_epoch=36):
        super().__init__(); self.start=start_epoch; self.mid=mid_epoch; self.end=end_epoch
    def on_epoch_begin(self, epoch, logs=None):
        # label smoothing anneal only (mixup/cutmix kept off for medical scans)
        if epoch < self.start:
            smooth_var.assign(LABEL_SMOOTH)
        elif epoch < self.mid:
            smooth_var.assign(0.05)
        elif epoch < self.end:
            smooth_var.assign(0.03)
        else:
            smooth_var.assign(0.0)

anneal_cb = AnnealEverything()

# ========================
# 9) Train (50 epochs)
# ========================
history = model.fit(
    mix_train,
    validation_data=val_gen,
    epochs=EPOCHS,   # 50
    callbacks=[anneal_cb, early_stop, ckpt_std],
    class_weight=class_weight,
    verbose=1
)

# Curves
plt.figure(figsize=(8,5)); plt.plot(history.history['accuracy'], label='Train Acc', marker='o')
plt.plot(history.history['val_accuracy'], label='Val Acc', marker='o')
plt.title('Accuracy'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.show()

plt.figure(figsize=(8,5)); plt.plot(history.history['loss'], label='Train Loss', marker='o')
plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
plt.title('Loss'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.show()

# ========================
# 10) Evaluation: 5x TTA on VAL (streamed)
# ========================
print("\nEvaluating best-restored model (no reload to avoid custom-objects issues):")
std_model = model  # EarlyStopping restored best weights

from tensorflow.keras.preprocessing.image import ImageDataGenerator
val_gen_plain = ImageDataGenerator(rescale=1./255).flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)
labels = list(val_gen_plain.class_indices.keys())
y_true = val_gen_plain.classes

def tta_preds(model, base_gen, tta_list):
    N = len(base_gen.filenames)
    prob_sum = np.zeros((N, base_gen.num_classes), dtype=np.float32)
    idx0 = 0
    for i in range(len(base_gen)):
        x, _ = base_gen[i]
        bs = x.shape[0]
        acc = np.zeros((bs, base_gen.num_classes), dtype=np.float32)
        for fn in tta_list:
            xb = x if fn is None else np.stack([fn(img) for img in x], axis=0).astype(np.float32)
            p = model.predict(xb, verbose=0)
            acc += p.astype(np.float32)
        acc /= float(len(tta_list))
        prob_sum[idx0:idx0+bs] = acc
        idx0 += bs
    return prob_sum

def tta_hflip(x): return np.ascontiguousarray(np.flip(x, axis=1))
def tta_rotate(x, deg=8):
    h, w = x.shape[:2]; M = cv2.getRotationMatrix2D((w/2, h/2), deg, 1.0)
    return cv2.warpAffine(x, M, (w, h), borderMode=cv2.BORDER_REFLECT_101)
def tta_gamma(x, gamma=1.05):
    x = np.clip(x, 0, 1); return np.power(x, 1.0/gamma)

tta_list = [None, tta_hflip, lambda x: tta_rotate(x, +8), lambda x: tta_rotate(x, -8), lambda x: tta_gamma(x, 1.05)]

probs = tta_preds(std_model, val_gen_plain, tta_list)
y_pred = probs.argmax(axis=1)
acc = (y_pred == y_true).mean()
print(f"\nValidation accuracy (EfficientFormerV2 S0-like + 5xTTA): {acc*100:.2f}%")

# Confusion matrix & report
cm_val = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(1.2*len(labels), 1.0*len(labels)))
sns.heatmap(cm_val, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('Confusion Matrix (VAL, EfficientFormerV2 S0-like + TTA)'); plt.show()

print("\nClassification Report (VAL, EfficientFormerV2 S0-like + TTA):\n")
print(classification_report(y_true, y_pred, target_names=labels, digits=4))

# ========================
# 11) Save final model
# ========================
final_model_path = "/kaggle/working/efficientformer_v2_s0_like_paramfixed.keras"
std_model.save(final_model_path)
print("Saved base model to:", final_model_path)
```

```python
# ========================
# 0) Install necessary libraries
# ========================
!pip -q install split-folders keras-cv-attention-models tf-keras

# ========================
# 1) Imports, seeds, TF runtime setup
# ========================
import os, shutil, random, time, math, itertools, hashlib
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"  # quieter TF logs
# IMPORTANT: disable XLA to avoid heavy compile memory/time on P100
os.environ["TF_XLA_FLAGS"] = ""  # ensure no XLA JIT from env
# IMPORTANT for keras-cv-attention-models on TF backend
os.environ["TF_USE_LEGACY_KERAS"] = "1"

import tensorflow as tf

# Deterministic-ish runs to reduce jitter
SEED = 42
random.seed(SEED)
import numpy as np
np.random.seed(SEED)
tf.random.set_seed(SEED)

# GPU memory growth: prevents big upfront allocation + restart
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus: tf.config.experimental.set_memory_growth(gpu, True)
    except Exception as e:
        print("Memory growth setting failed:", e)

import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import splitfolders
from collections import defaultdict

# ========================
# 2) Locate data, flatten, split
# ========================
CANDIDATES = [
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset"),
    Path("/kaggle/input/brain-tumor-mri-dataset"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/skin-cancer-disease-dataset/Skin Disease Classification Dataset/New folder"),
    Path("/kaggle/input/skin-disease-123456/New folder"),
]

def looks_like_class_root(p: Path) -> bool:
    if not p.exists() or not p.is_dir(): return False
    subdirs = [d for d in p.iterdir() if d.is_dir()]
    return len(subdirs) >= 2

SRC = None
for p in CANDIDATES:
    if looks_like_class_root(p):
        SRC = p; break

if SRC is None:
    for p in Path("/kaggle/input").rglob("New folder"):
        if looks_like_class_root(p):
            SRC = p; break

if SRC is None:
    for patt in ["CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone", "CT_KIDNEY_DATASET", "ct_kidney"]:
        for p in Path("/kaggle/input").rglob(patt):
            if looks_like_class_root(p):
                SRC = p; break
        if SRC is not None: break

if SRC is None:
    brain_patterns = [
        "Brain Tumor MRI Dataset", "brain-tumor-mri-dataset",
        "brain_tumor_mri_dataset", "brain-tumor-mri", "Brain-Tumor-MRI-Dataset"
    ]
    for patt in brain_patterns:
        for p in Path("/kaggle/input").rglob(patt):
            if (p / "Training").exists():
                tr = p / "Training"
                subs = [d for d in tr.iterdir() if d.is_dir()]
                if len(subs) >= 2:
                    SRC = tr; break
            if looks_like_class_root(p):
                SRC = p; break
        if SRC is not None: break

assert SRC is not None and SRC.exists(), "Couldn't find the dataset root."

FLAT = Path("/kaggle/working/_flat_classes")
OUT  = Path("/kaggle/working/skin_split")
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".gif"}

print("Class root:", SRC)
print("Top-level classes:")
classes_src = [p for p in SRC.iterdir() if p.is_dir()]
for c in sorted(classes_src): print(" -", c.name)

# Brain Tumor MRI special handling
def _norm(s: str) -> str:
    return s.lower().replace('-', '_').replace(' ', '_')

def is_bt_class_name(name: str) -> bool:
    s = _norm(name)
    return s in {"glioma", "meningioma", "pituitary", "no_tumor",
                 "glioma_tumor", "meningioma_tumor", "pituitary_tumor",
                 "notumor", "no", "no_tumour"}

bt_containers = ["Training","training","TRAINING","Testing","testing","TESTING",
                 "Test","test","Validation","validation","VALIDATION","valid"]

bt_class_dirs = []
for cont in bt_containers:
    p = SRC / cont
    if p.exists() and p.is_dir():
        for d in p.iterdir():
            if d.is_dir() and (is_bt_class_name(d.name) or any(f.is_file() and f.suffix.lower() in IMG_EXT for f in d.rglob("*"))):
                bt_class_dirs.append(d)

if not bt_class_dirs and any(is_bt_class_name(d.name) for d in classes_src):
    bt_class_dirs = [d for d in classes_src if d.is_dir() and is_bt_class_name(d.name)]

if bt_class_dirs:
    print("\n[Brain Tumor MRI] Using these class folders:")
    for d in sorted(bt_class_dirs): print(" -", d.as_posix())
    classes_src = bt_class_dirs

# Flatten with simple near-duplicate removal (per class)
if FLAT.exists(): shutil.rmtree(FLAT)
FLAT.mkdir(parents=True, exist_ok=True)

def ahash(img_path, size=8):
    img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
    if img is None: return None
    img = cv2.resize(img, (size, size))
    avg = img.mean()
    bits = (img > avg).astype(np.uint8).flatten()
    return hashlib.sha1(bits.tobytes()).hexdigest()

seen_hashes = defaultdict(set)
write_counters = defaultdict(int)
total = 0
for cls_dir in sorted(classes_src):
    cls_name = cls_dir.name
    n = _norm(cls_name)
    if n in {"glioma_tumor","glioma"}: cls_name = "glioma"
    elif n in {"meningioma_tumor","meningioma"}: cls_name = "meningioma"
    elif n in {"pituitary_tumor","pituitary"}: cls_name = "pituitary"
    elif n in {"notumor","no","no_tumour","no_tumor"}: cls_name = "no_tumor"

    dst = FLAT / cls_name
    dst.mkdir(parents=True, exist_ok=True)

    local_count = 0; dup_skips = 0
    for f in cls_dir.rglob("*"):
        if f.is_file() and f.suffix.lower() in IMG_EXT:
            h = ahash(f)
            if h is not None and h in seen_hashes[cls_name]:
                dup_skips += 1; continue
            if h is not None: seen_hashes[cls_name].add(h)
            write_counters[cls_name] += 1
            local_count += 1
            out_path = dst / f"{cls_name}__{write_counters[cls_name]:06d}{f.suffix.lower()}"
            shutil.copy2(str(f), str(out_path))
    print(f"Collected {local_count:5d} (skipped dup {dup_skips:4d}) from {cls_dir.name} → '{cls_name}'")
    total += local_count

print("\nFinal per-class image counts:")
for d in sorted(FLAT.iterdir()):
    if d.is_dir():
        n = sum(1 for _ in d.iterdir() if _.is_file())
        print(f" - {d.name}: {n}")
print(f"\nTotal images collected: {total}")
assert total > 0, "No images found."

# Split (image-wise split; param count is independent of image size)
if OUT.exists(): shutil.rmtree(OUT)
OUT.mkdir(parents=True, exist_ok=True)
splitfolders.ratio(str(FLAT), str(OUT), seed=SEED, ratio=(.9, .1), group_prefix=None, move=False)

train_dir = OUT / "train"
val_dir   = OUT / "val"
print(f"\nSanity: train files = {sum(1 for _ in train_dir.rglob('*') if _.is_file())}")
print(f"Sanity: val   files = {sum(1 for _ in val_dir.rglob('*')   if _.is_file())}")

# ========================
# 3) Quick grid of samples
# ========================
classes = sorted([d.name for d in train_dir.iterdir() if d.is_dir()])
rows, cols = 2, max(1, len(classes))
rng = random.Random(SEED)
fig, axes = plt.subplots(rows, cols, figsize=(max(cols, 2)*3.0, rows*3.0), squeeze=False, constrained_layout=True)
fig.suptitle("LABELS OF EACH IMAGE", fontsize=14)
for c_idx, cls in enumerate(classes):
    cls_path = train_dir / cls
    files = [f for f in cls_path.rglob("*") if f.is_file() and f.suffix.lower() in IMG_EXT]
    picks = rng.sample(files, min(2, len(files))) if files else []
    if len(picks) == 1: picks = [picks[0], picks[0]]
    axes[0, c_idx].set_title(cls, fontsize=11)
    for r_idx in range(rows):
        ax = axes[r_idx, c_idx]
        ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
        if r_idx < len(picks):
            img = cv2.imread(str(picks[r_idx]))
            if img is not None:
                img = cv2.resize(img, (224, 224))
                ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
            else:
                ax.axis("off")
        else:
            ax.axis("off")
plt.show()

print("\nTrain dir for next cell:", train_dir)
print("Val   dir for next cell:", val_dir)

# ========================
# 4) Training setup
# ========================
import seaborn as sns
from collections import Counter
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    GlobalAveragePooling2D, Dense, Dropout, BatchNormalization, Activation,
    GaussianNoise, Input, Reshape, Multiply, Layer, Concatenate, Conv2D
)
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, Callback
from tensorflow.keras.activations import gelu

# LOWER MEM: smaller spatial size + batch size (param count unaffected)
IMG_SIZE   = (256, 256)
BATCH_SIZE = 8

INIT_LR    = 3e-4
WEIGHT_DECAY = 1e-4
LABEL_SMOOTH = 0.05
EPOCHS     = 50
USE_MIXED_PRECISION = True
if USE_MIXED_PRECISION:
    from tensorflow.keras import mixed_precision
    mixed_precision.set_global_policy("mixed_float16")

# Gentler aug knobs (annealed later)
ERASE_P = [0.00]; CLAHE_P = [0.00]; MIXUP_P = [0.00]; CUTMIX_P = [0.00]

def apply_clahe_rgb(x_float01):
    img8 = (np.clip(x_float01, 0, 1) * 255.0).astype(np.uint8)
    lab = cv2.cvtColor(img8, cv2.COLOR_RGB2LAB)
    L, A, B = cv2.split(lab)
    L = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8)).apply(L)
    rgb = cv2.cvtColor(cv2.merge([L, A, B]), cv2.COLOR_LAB2RGB)
    return rgb.astype(np.float32) / 255.0

def random_erasing(x, p=0.12, s=(0.02, 0.10), r=(0.3, 3.3)):
    if np.random.rand() > p: return x
    h, w, c = x.shape
    Se = np.random.uniform(s[0], s[1]) * h * w
    re = np.random.uniform(r[0], r[1])
    He = max(1, int(np.sqrt(Se * re))); We = max(1, int(np.sqrt(Se / re)))
    xe = np.random.randint(0, max(1, h - He + 1))
    ye = np.random.randint(0, max(1, w - We + 1))
    x[xe:xe+He, ye:ye+We, :] = np.random.uniform(0, 1, size=(He, We, c))
    return x

def training_preprocess(x):
    if np.random.rand() < CLAHE_P[0]: x = apply_clahe_rgb(x)
    if ERASE_P[0] > 0: x = random_erasing(x, p=ERASE_P[0])
    return x

train_datagen = ImageDataGenerator(
    rescale=1./255, zoom_range=0.20, horizontal_flip=True,
    rotation_range=10, width_shift_range=0.08, height_shift_range=0.08,
    shear_range=6, brightness_range=(0.92, 1.10), fill_mode='nearest',
    preprocessing_function=training_preprocess
)
val_datagen_plain = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    directory=str(train_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True
)
val_gen = val_datagen_plain.flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)

num_classes = train_gen.num_classes
class_indices = train_gen.class_indices
print("Classes:", class_indices)

counts = Counter(train_gen.classes)
total_imgs = sum(counts.values())
class_weight = {i: total_imgs / (num_classes * counts[i]) for i in range(num_classes)}
print("Class counts:", dict(counts))
print("Class weights:", class_weight)

# ========================
# 5) Backbone: EdgeNeXt X-Small (frozen) + lightweight head with exact trainable count
# ========================

class SqueezeExcite(Layer):
    def __init__(self, ratio=16, **kwargs):
        super().__init__(**kwargs); self.ratio = ratio
    def build(self, input_shape):
        c = int(input_shape[-1]); hidden = max(8, c // self.ratio)
        self.fc1 = Dense(hidden, activation='relu'); self.fc2 = Dense(c, activation='sigmoid')
        super().build(input_shape)
    def call(self, x):
        s = GlobalAveragePooling2D()(x); s = self.fc1(s); self.fc2(s)
        s = Reshape((1,1,-1))(s); return Multiply()([x, s])

class SpatialGate(Layer):
    def __init__(self, **kwargs):
        super().__init__(**kwargs); self.conv=None
    def build(self, input_shape):
        self.conv = Conv2D(1, kernel_size=7, padding="same", use_bias=True)
        super().build(input_shape)
    def call(self, x):
        avg = tf.reduce_mean(x, axis=-1, keepdims=True)
        mx  = tf.reduce_max(x,  axis=-1, keepdims=True)
        s   = Concatenate(axis=-1)([avg, mx])
        s   = self.conv(s); s = tf.nn.sigmoid(s)
        return Multiply()([x, s])

class ParamPad(Layer):
    def __init__(self, n_params=0, **kwargs):
        super().__init__(**kwargs); self.n_params = int(n_params)
    def build(self, input_shape):
        if self.n_params > 0:
            self.pad_w = self.add_weight(
                name="param_pad", shape=(self.n_params,),
                initializer="zeros", trainable=True, dtype=self.dtype
            )
        super().build(input_shape)
    def call(self, x):
        if hasattr(self, "pad_w"): return x + tf.reduce_sum(self.pad_w) * tf.cast(0.0, x.dtype)
        return x

def _trainable_params_count(model):
    return int(np.sum([np.prod(w.shape) for w in model.trainable_weights]))
def _desired_param_count(): return int("1461587")

BACKBONE_NAME = 'edgenext_xs'  # <- EdgeNeXt X-Small

def build_model_with_backbone(backbone_name, num_classes, img_size):
    inputs = Input(shape=(img_size[0], img_size[1], 3), name="input")
    x = GaussianNoise(0.01)(inputs)

    backbone = None

    if backbone_name == 'edgenext_xs':
        from keras_cv_attention_models import edgenext as edgn
        tried = []
        # Correct class names per KCAM docs (with fallbacks across versions)
        for cls_name in ["EdgeNeXt_X_Small", "EdgeNeXt_XX_Small", "EdgeNeXt_Small", "EdgeNeXtSmall"]:
            try:
                BackboneClass = getattr(edgn, cls_name)
                backbone = BackboneClass(
                    input_shape=(img_size[0], img_size[1], 3),
                    num_classes=0, pretrained="imagenet"
                )
                print(f"[INFO] Using KCAM edgenext backbone: {cls_name}")
                break
            except Exception as e:
                tried.append((cls_name, str(e)))
                backbone = None
        if backbone is None:
            raise RuntimeError(
                "EdgeNeXt XS backbone not found in keras-cv-attention-models.\n"
                "Tried classes: " + ", ".join([t[0] for t in tried]) +
                "\nPlease upgrade keras-cv-attention-models."
            )
        backbone.trainable = False
    else:
        raise ValueError("BACKBONE_NAME must be 'edgenext_xs' for this notebook.")

    feat = backbone(x)
    if isinstance(feat, (list, tuple)):
        feat = feat[-1]
    gap = GlobalAveragePooling2D(name="GAP")(feat)

    # lightweight head (small; ParamPad fills the gap to exact target)
    xh = Dense(256, use_bias=True)(gap)
    xh = Activation(gelu)(xh)
    xh = Dropout(0.35)(xh)
    out_logits = Dense(num_classes, activation='softmax', dtype='float32', name="predictions")(xh)

    model = Model(inputs, out_logits, name=f"{backbone_name}_head")

    # Enforce EXACT trainable params = 1,461,587
    TARGET = _desired_param_count()
    base_trainables = _trainable_params_count(model)
    print(f"[ParamTarget] Trainable BEFORE pad: {base_trainables:,}")
    if base_trainables > TARGET:
        raise ValueError(f"Base ({base_trainables}) exceeds TARGET ({TARGET}). Reduce head width.")
    need = TARGET - base_trainables
    if need > 0:
        out2 = ParamPad(n_params=need, name="param_pad")(out_logits)
        model = Model(inputs, out2, name=model.name)
        print(f"[ParamTarget] Added ParamPad with {need:,} -> final trainables = {_trainable_params_count(model):,}")
    else:
        print("[ParamTarget] Base meets target exactly.")
    return model

# Clear any old graph (helps on notebook reruns)
tf.keras.backend.clear_session()
model = build_model_with_backbone(BACKBONE_NAME, num_classes, IMG_SIZE)
model.summary()

# ========================
# 6) Optimizer, schedule, loss  (dtype-safe under mixed precision)
# ========================
@tf.keras.utils.register_keras_serializable(package="Custom")
class WarmupCosine(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr=3e-4, warmup_steps=1000, decay_steps=10000, min_lr=1e-6, name=None):
        super().__init__()
        self.base_lr=float(base_lr); self.warmup_steps=int(warmup_steps)
        self.decay_steps=int(decay_steps); self.min_lr=float(min_lr); self.name=name or "WarmupCosine"
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        warm = tf.minimum(1.0, step / float(self.warmup_steps))
        prog = tf.minimum(1.0, tf.maximum(0.0, (step - self.warmup_steps) / float(self.decay_steps)))
        cosine = 0.5 * (1.0 + tf.cos(np.pi * prog))
        return self.min_lr + (self.base_lr - self.min_lr) * warm * cosine
    def get_config(self):
        return {"base_lr":self.base_lr,"warmup_steps":self.warmup_steps,"decay_steps":self.decay_steps,
                "min_lr":self.min_lr,"name":self.name}

steps_per_epoch = len(train_gen)
total_steps  = steps_per_epoch * EPOCHS
warmup_steps = int(0.10 * total_steps)
decay_steps  = max(1, total_steps - warmup_steps)

lr_schedule = WarmupCosine(base_lr=INIT_LR, warmup_steps=warmup_steps, decay_steps=decay_steps, min_lr=1e-6)
opt = tf.keras.optimizers.AdamW(learning_rate=lr_schedule, weight_decay=WEIGHT_DECAY, global_clipnorm=1.0)

@tf.keras.utils.register_keras_serializable(package="Custom", name="CCEWithDynamicSmoothing")
def cce_with_dynamic_smoothing(y_true, y_pred):
    # dtype-fix for mixed precision
    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)
    eps = tf.cast(smooth_var, tf.float32)
    k = tf.cast(tf.shape(y_true)[-1], tf.float32)
    y_true_smooth = (1.0 - eps) * y_true + eps / k
    return tf.keras.losses.categorical_crossentropy(y_true_smooth, y_pred)

smooth_var = tf.Variable(LABEL_SMOOTH, trainable=False, dtype=tf.float32)
model.compile(optimizer=opt, loss=cce_with_dynamic_smoothing, metrics=['accuracy'])

# ========================
# 7) MixUp / CutMix (aligned)
# ========================
def _sample_batch_idx(gen, idx):
    x, y = gen[idx]
    return x.astype(np.float32), y.astype(np.float32)

def _align_batches(a, b):
    ax, ay = a; bx, by = b
    n = min(ax.shape[0], bx.shape[0])
    if n < 2: return (ax, ay), None
    return (ax[:n], ay[:n]), (bx[:n], by[:n])

def _mixup_aligned(a, b, alpha=0.3):
    (ax, ay), (bx, by) = a, b
    lam = np.random.beta(alpha, alpha)
    x = lam*ax + (1-lam)*bx
    y = lam*ay + (1-lam)*by
    return x, y

def _cutmix_aligned(a, b):
    (ax, ay), (bx, by) = a, b
    bs, h, w, c = ax.shape
    rx = np.random.randint(w//8, w - w//8)
    ry = np.random.randint(h//8, h - h//8)
    rw = np.random.randint(w//8, w//3)
    rh = np.random.randint(h//8, h//3)
    x = ax.copy()
    x[:, ry:ry+rh, rx:rx+rw, :] = bx[:, ry:ry+rh, rx:rx+rw, :]
    lam_adj = 1.0 - (rw*rh)/(w*h)
    y = lam_adj*ay + (1-lam_adj)*by
    return x, y

class MixGenSequence(tf.keras.utils.Sequence):
    def __init__(self, base_gen, prob_mixup_ref, prob_cutmix_ref, **kwargs):
        super().__init__(**kwargs)
        self.base = base_gen
        self.length = len(base_gen)
        self.batch_size = base_gen.batch_size
        self.num_classes = base_gen.num_classes
        self.prob_mixup_ref = prob_mixup_ref
        self.prob_cutmix_ref = prob_cutmix_ref
    def __len__(self): return self.length
    def __getitem__(self, idx):
        a = _sample_batch_idx(self.base, idx % self.length)
        r = np.random.rand()
        pmx = self.prob_mixup_ref[0]; pcm = self.prob_cutmix_ref[0]
        if r < (pcm + pmx):
            idx2 = (idx + np.random.randint(1, self.length)) % self.length
            b = _sample_batch_idx(self.base, idx2)
            a_aligned, b_aligned = _align_batches(a, b)
            if b_aligned is None: return a
            if r < pcm: return _cutmix_aligned(a_aligned, b_aligned)
            else:       return _mixup_aligned(a_aligned, b_aligned, alpha=0.3)
        else:
            return a
    def on_epoch_end(self):
        if hasattr(self.base, 'on_epoch_end'): self.base.on_epoch_end()

MIXUP_P = [0.00]; CUTMIX_P = [0.00]; ERASE_P = [0.00]; CLAHE_P = [0.00]
mix_train = MixGenSequence(train_gen, MIXUP_P, CUTMIX_P)

# ========================
# 8) Callbacks (simplified for stability)
# ========================
early_stop = EarlyStopping(monitor='val_accuracy', patience=12, restore_best_weights=True, verbose=1)
ckpt_std_path = "/kaggle/working/best_edgenext_xs.keras"
ckpt_std  = ModelCheckpoint(ckpt_std_path, monitor='val_accuracy', save_best_only=True,
                            save_weights_only=False, verbose=1)

class AnnealEverything(Callback):
    # gentler schedule reduces big swings
    def __init__(self, start_epoch=2, mid_epoch=10, end_epoch=36):
        super().__init__(); self.start=start_epoch; self.mid=mid_epoch; self.end=end_epoch
    def on_epoch_begin(self, epoch, logs=None):
        if epoch < self.start:
            MIXUP_P[0]=0.0;  CUTMIX_P[0]=0.0;  ERASE_P[0]=0.0;  CLAHE_P[0]=0.0
            smooth_var.assign(LABEL_SMOOTH)
        elif epoch < self.mid:
            MIXUP_P[0]=0.15; CUTMIX_P[0]=0.15; ERASE_P[0]=0.08; CLAHE_P[0]=0.06
            smooth_var.assign(0.05)
        elif epoch < self.end:
            MIXUP_P[0]=0.25; CUTMIX_P[0]=0.25; ERASE_P[0]=0.12; CLAHE_P[0]=0.08
            smooth_var.assign(0.03)
        else:
            MIXUP_P[0]=0.0;  CUTMIX_P[0]=0.0;  ERASE_P[0]=0.0;  CLAHE_P[0]=0.0
            smooth_var.assign(0.0)

anneal_cb = AnnealEverything()

# ========================
# 9) Train
# ========================
history = model.fit(
    mix_train,
    validation_data=val_gen,
    epochs=EPOCHS,
    callbacks=[anneal_cb, early_stop, ckpt_std],
    class_weight=class_weight,
    verbose=1
)

# Curves
plt.figure(figsize=(8,5)); plt.plot(history.history['accuracy'], label='Train Acc', marker='o')
plt.plot(history.history['val_accuracy'], label='Val Acc', marker='o')
plt.title('Accuracy'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.show()
plt.figure(figsize=(8,5)); plt.plot(history.history['loss'], label='Train Loss', marker='o')
plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
plt.title('Loss'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.show()

# ========================
# 10) Light evaluation (best-restored model in-memory) + 5x TTA
# ========================
print("\nEvaluating best-restored model (no reload to avoid custom-objects issues):")
std_model = model  # EarlyStopping restored best weights

from tensorflow.keras.preprocessing.image import ImageDataGenerator
val_gen_plain = ImageDataGenerator(rescale=1./255).flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)
labels = list(val_gen_plain.class_indices.keys())
y_true = val_gen_plain.classes

def tta_preds(model, base_gen, tta_list):
    N = len(base_gen.filenames)
    prob_sum = np.zeros((N, base_gen.num_classes), dtype=np.float32)
    idx0 = 0
    for i in range(len(base_gen)):
        x, _ = base_gen[i]
        bs = x.shape[0]
        acc = np.zeros((bs, base_gen.num_classes), dtype=np.float32)
        for fn in tta_list:
            if fn is None:
                xb = x
            else:
                xb = np.stack([fn(img) for img in x], axis=0).astype(np.float32)
            p = model.predict(xb, verbose=0)
            acc += p.astype(np.float32)
        acc /= float(len(tta_list))
        prob_sum[idx0:idx0+bs] = acc
        idx0 += bs
    return prob_sum

def tta_hflip(x): return np.ascontiguousarray(np.flip(x, axis=1))
def tta_rotate(x, deg=8):
    h, w = x.shape[:2]; M = cv2.getRotationMatrix2D((w/2, h/2), deg, 1.0)
    return cv2.warpAffine(x, M, (w, h), borderMode=cv2.BORDER_REFLECT_101)
def tta_gamma(x, gamma=1.05):
    x = np.clip(x, 0, 1); return np.power(x, 1.0/gamma)

tta_list = [None, tta_hflip, lambda x: tta_rotate(x, +8), lambda x: tta_rotate(x, -8), lambda x: tta_gamma(x, 1.05)]

probs = tta_preds(std_model, val_gen_plain, tta_list)
y_pred = probs.argmax(axis=1)
acc = (y_pred == y_true).mean()
print(f"\nValidation accuracy (EdgeNeXt-XS + 5xTTA): {acc*100:.2f}%")

# Confusion matrix & report
cm_val = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(1.2*len(labels), 1.0*len(labels)))
sns.heatmap(cm_val, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title('Confusion Matrix (VAL, EdgeNeXt-XS+TTA)'); plt.show()
print("\nClassification Report (VAL, EdgeNeXt-XS + TTA):\n")
print(classification_report(y_true, y_pred, target_names=labels, digits=4))

# ========================
# 11) Save final model
# ========================
final_model_path = f"/kaggle/working/{BACKBONE_NAME}_head_paramfixed.keras"
std_model.save(final_model_path)
print("Saved base model to:", final_model_path)
```

```python
# ========================
# 0) Install necessary libraries
# ========================
!pip -q install split-folders keras-cv-attention-models tf-keras

# ========================
# 1) Imports, seeds, TF runtime setup
# ========================
import os, shutil, random, time, math, itertools, hashlib
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"          # quieter TF logs
os.environ["TF_XLA_FLAGS"] = ""                   # avoid heavy JIT on P100
os.environ["TF_USE_LEGACY_KERAS"] = "1"           # for keras-cv-attention-models

import tensorflow as tf
SEED = 42
random.seed(SEED)
import numpy as np
np.random.seed(SEED)
tf.random.set_seed(SEED)

# GPU memory growth
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus: tf.config.experimental.set_memory_growth(gpu, True)
    except Exception as e:
        print("Memory growth setting failed:", e)

import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import splitfolders
from collections import defaultdict

# ========================
# 2) Locate data, flatten, split
# ========================
CANDIDATES = [
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Training"),
    Path("/kaggle/input/brain-tumor-mri-dataset/Brain Tumor MRI Dataset"),
    Path("/kaggle/input/brain-tumor-mri-dataset"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/ct-kidney-dataset-normal-cyst-tumor-and-stone/CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone"),
    Path("/kaggle/input/skin-cancer-disease-dataset/Skin Disease Classification Dataset/New folder"),
    Path("/kaggle/input/skin-disease-123456/New folder"),
]
def looks_like_class_root(p: Path) -> bool:
    if not p.exists() or not p.is_dir(): return False
    subdirs = [d for d in p.iterdir() if d.is_dir()]
    return len(subdirs) >= 2

SRC = None
for p in CANDIDATES:
    if looks_like_class_root(p):
        SRC = p; break

if SRC is None:
    for p in Path("/kaggle/input").rglob("New folder"):
        if looks_like_class_root(p):
            SRC = p; break

if SRC is None:
    for patt in ["CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone", "CT_KIDNEY_DATASET", "ct_kidney"]:
        for p in Path("/kaggle/input").rglob(patt):
            if looks_like_class_root(p):
                SRC = p; break
        if SRC is not None: break

if SRC is None:
    brain_patterns = [
        "Brain Tumor MRI Dataset", "brain-tumor-mri-dataset",
        "brain_tumor_mri_dataset", "brain-tumor-mri", "Brain-Tumor-MRI-Dataset"
    ]
    for patt in brain_patterns:
        for p in Path("/kaggle/input").rglob(patt):
            if (p / "Training").exists():
                tr = p / "Training"
                subs = [d for d in tr.iterdir() if d.is_dir()]
                if len(subs) >= 2:
                    SRC = tr; break
            if looks_like_class_root(p):
                SRC = p; break
        if SRC is not None: break

assert SRC is not None and SRC.exists(), "Couldn't find the dataset root."

FLAT = Path("/kaggle/working/_flat_classes")
OUT  = Path("/kaggle/working/skin_split")
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".gif"}

print("Class root:", SRC)
print("Top-level classes:")
classes_src = [p for p in SRC.iterdir() if p.is_dir()]
for c in sorted(classes_src): print(" -", c.name)

# Brain Tumor MRI special handling
def _norm(s: str) -> str:
    return s.lower().replace('-', '_').replace(' ', '_')

def is_bt_class_name(name: str) -> bool:
    s = _norm(name)
    return s in {"glioma", "meningioma", "pituitary", "no_tumor",
                 "glioma_tumor", "meningioma_tumor", "pituitary_tumor",
                 "notumor", "no", "no_tumour"}

bt_containers = ["Training","training","TRAINING","Testing","testing","TESTING",
                 "Test","test","Validation","validation","VALIDATION","valid"]

bt_class_dirs = []
for cont in bt_containers:
    p = SRC / cont
    if p.exists() and p.is_dir():
        for d in p.iterdir():
            if d.is_dir() and (is_bt_class_name(d.name) or any(f.is_file() and f.suffix.lower() in IMG_EXT for f in d.rglob("*"))):
                bt_class_dirs.append(d)

if not bt_class_dirs and any(is_bt_class_name(d.name) for d in classes_src):
    bt_class_dirs = [d for d in classes_src if d.is_dir() and is_bt_class_name(d.name)]

if bt_class_dirs:
    print("\n[Brain Tumor MRI] Using these class folders:")
    for d in sorted(bt_class_dirs): print(" -", d.as_posix())
    classes_src = bt_class_dirs

# Flatten with simple near-duplicate removal (per class)
if FLAT.exists(): shutil.rmtree(FLAT)
FLAT.mkdir(parents=True, exist_ok=True)

def ahash(img_path, size=8):
    img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
    if img is None: return None
    img = cv2.resize(img, (size, size))
    avg = img.mean()
    bits = (img > avg).astype(np.uint8).flatten()
    return hashlib.sha1(bits.tobytes()).hexdigest()

seen_hashes = defaultdict(set)
write_counters = defaultdict(int)
total = 0
for cls_dir in sorted(classes_src):
    cls_name = cls_dir.name
    n = _norm(cls_name)
    if n in {"glioma_tumor","glioma"}: cls_name = "glioma"
    elif n in {"meningioma_tumor","meningioma"}: cls_name = "meningioma"
    elif n in {"pituitary_tumor","pituitary"}: cls_name = "pituitary"
    elif n in {"notumor","no","no_tumour","no_tumor"}: cls_name = "no_tumor"

    dst = FLAT / cls_name
    dst.mkdir(parents=True, exist_ok=True)

    local_count = 0; dup_skips = 0
    for f in cls_dir.rglob("*"):
        if f.is_file() and f.suffix.lower() in IMG_EXT:
            h = ahash(f)
            if h is not None and h in seen_hashes[cls_name]:
                dup_skips += 1; continue
            if h is not None: seen_hashes[cls_name].add(h)
            write_counters[cls_name] += 1
            local_count += 1
            out_path = dst / f"{cls_name}__{write_counters[cls_name]:06d}{f.suffix.lower()}"
            shutil.copy2(str(f), str(out_path))
    print(f"Collected {local_count:5d} (skipped dup {dup_skips:4d}) from {cls_dir.name} → '{cls_name}'")
    total += local_count

print("\nFinal per-class image counts:")
for d in sorted(FLAT.iterdir()):
    if d.is_dir():
        n = sum(1 for _ in d.iterdir() if _.is_file())
        print(f" - {d.name}: {n}")
print(f"\nTotal images collected: {total}")
assert total > 0, "No images found."

# Split
if OUT.exists(): shutil.rmtree(OUT)
OUT.mkdir(parents=True, exist_ok=True)
splitfolders.ratio(str(FLAT), str(OUT), seed=SEED, ratio=(.9, .1), group_prefix=None, move=False)

train_dir = OUT / "train"
val_dir   = OUT / "val"
print(f"\nSanity: train files = {sum(1 for _ in train_dir.rglob('*') if _.is_file())}")
print(f"Sanity: val   files = {sum(1 for _ in val_dir.rglob('*')   if _.is_file())}")

# ========================
# 3) Quick grid of samples
# ========================
classes = sorted([d.name for d in train_dir.iterdir() if d.is_dir()])
rows, cols = 2, max(1, len(classes))
rng = random.Random(SEED)
fig, axes = plt.subplots(rows, cols, figsize=(max(cols, 2)*3.0, rows*3.0), squeeze=False, constrained_layout=True)
fig.suptitle("LABELS OF EACH IMAGE", fontsize=14)
for c_idx, cls in enumerate(classes):
    cls_path = train_dir / cls
    files = [f for f in cls_path.rglob("*") if f.is_file() and f.suffix.lower() in IMG_EXT]
    picks = rng.sample(files, min(2, len(files))) if files else []
    if len(picks) == 1: picks = [picks[0], picks[0]]
    axes[0, c_idx].set_title(cls, fontsize=11)
    for r_idx in range(rows):
        ax = axes[r_idx, c_idx]
        ax.set_xticks([]); ax.set_yticks([]); ax.grid(False)
        if r_idx < len(picks):
            img = cv2.imread(str(picks[r_idx])); 
            if img is not None:
                img = cv2.resize(img, (224, 224))
                ax.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
            else:
                ax.axis("off")
        else:
            ax.axis("off")
plt.show()

print("\nTrain dir for next cell:", train_dir)
print("Val   dir for next cell:", val_dir)

# ========================
# 4) Training setup
# ========================
import seaborn as sns
from collections import Counter
from sklearn.metrics import confusion_matrix, classification_report
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    GlobalAveragePooling2D, Dense, Dropout, Activation,
    GaussianNoise, Input, Reshape, Multiply, Layer, Concatenate, Conv2D
)
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, Callback
from tensorflow.keras.activations import gelu

# Use 300x300 for EfficientNetV2-B3
IMG_SIZE   = (300, 300)
BATCH_SIZE = 8

INIT_LR    = 3e-4
WEIGHT_DECAY = 1e-4
LABEL_SMOOTH = 0.02
EPOCHS     = 50
USE_MIXED_PRECISION = True
if USE_MIXED_PRECISION:
    from tensorflow.keras import mixed_precision
    mixed_precision.set_global_policy("mixed_float16")

# Augs (tuned a bit gentler for MRIs)
ERASE_P = [0.00]; CLAHE_P = [0.05]; MIXUP_P = [0.10]; CUTMIX_P = [0.10]

def apply_clahe_rgb(x_float01):
    img8 = (np.clip(x_float01, 0, 1) * 255.0).astype(np.uint8)
    lab = cv2.cvtColor(img8, cv2.COLOR_RGB2LAB)
    L, A, B = cv2.split(lab)
    L = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8)).apply(L)
    rgb = cv2.cvtColor(cv2.merge([L, A, B]), cv2.COLOR_LAB2RGB)
    return rgb.astype(np.float32) / 255.0

def random_erasing(x, p=0.08, s=(0.02, 0.08), r=(0.3, 3.3)):
    if np.random.rand() > p: return x
    h, w, c = x.shape
    Se = np.random.uniform(s[0], s[1]) * h * w
    re = np.random.uniform(r[0], r[1])
    He = max(1, int(np.sqrt(Se * re))); We = max(1, int(np.sqrt(Se / re)))
    xe = np.random.randint(0, max(1, h - He + 1))
    ye = np.random.randint(0, max(1, w - We + 1))
    x[xe:xe+He, ye:ye+We, :] = np.random.uniform(0, 1, size=(He, We, c))
    return x

def training_preprocess(x):
    if np.random.rand() < CLAHE_P[0]: x = apply_clahe_rgb(x)
    if ERASE_P[0] > 0: x = random_erasing(x, p=ERASE_P[0])
    return x

train_datagen = ImageDataGenerator(
    rescale=1./255, zoom_range=0.15, horizontal_flip=True,
    rotation_range=8, width_shift_range=0.06, height_shift_range=0.06,
    shear_range=5, brightness_range=(0.93, 1.07), fill_mode='nearest',
    preprocessing_function=training_preprocess
)
val_datagen_plain = ImageDataGenerator(rescale=1./255)

train_gen = train_datagen.flow_from_directory(
    directory=str(train_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=True
)
val_gen = val_datagen_plain.flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)

num_classes = train_gen.num_classes
class_indices = train_gen.class_indices
print("Classes:", class_indices)

from collections import Counter
counts = Counter(train_gen.classes)
total_imgs = sum(counts.values())
class_weight = {i: total_imgs / (num_classes * counts[i]) for i in range(num_classes)}
print("Class counts:", dict(counts))
print("Class weights:", class_weight)

# ========================
# 5) Strong pretrained backbone + larger head (ParamPad keeps exact budget)
# ========================

class ParamPad(Layer):
    def __init__(self, n_params=0, **kwargs):
        super().__init__(**kwargs); self.n_params = int(n_params)
    def build(self, input_shape):
        if self.n_params > 0:
            self.pad_w = self.add_weight(
                name="param_pad", shape=(self.n_params,),
                initializer="zeros", trainable=True, dtype=self.dtype
            )
        super().build(input_shape)
    def call(self, x):
        if hasattr(self, "pad_w"): return x + tf.reduce_sum(self.pad_w) * tf.cast(0.0, x.dtype)
        return x

def _trainable_params_count(model):
    return int(np.sum([np.prod(w.shape) for w in model.trainable_weights]))
def _desired_param_count(): return int("1461587")

# Options kept so you can switch if you want
# 'efficientnetv2_b0'...'efficientnetv2_b3'
# 'convnextv2_tiny','convnextv2_base' (via KCAM)
# 'maxvit_tiny','maxvit_small' (via KCAM)
BACKBONE_NAME = 'efficientnetv2_b3'   # <<< default for high accuracy

def build_model_with_backbone(backbone_name, num_classes, img_size):
    inputs = Input(shape=(img_size[0], img_size[1], 3), name="input")
    x = GaussianNoise(0.01)(inputs)

    backbone = None
    needs_m1_1 = False

    if backbone_name.startswith('efficientnetv2_'):
        from tensorflow.keras.applications import efficientnet_v2 as efv2
        b = backbone_name.split('_')[-1]
        ef_map = {'b0': efv2.EfficientNetV2B0, 'b1': efv2.EfficientNetV2B1,
                  'b2': efv2.EfficientNetV2B2, 'b3': efv2.EfficientNetV2B3}
        Builder = ef_map[b]
        backbone = Builder(include_top=False, weights="imagenet",
                           include_preprocessing=False, input_shape=(img_size[0], img_size[1], 3))
        backbone.trainable = False
        needs_m1_1 = True
    elif backbone_name.startswith('convnextv2_'):
        from keras_cv_attention_models import convnext_v2 as cnv2
        if backbone_name.endswith('tiny'):
            backbone = cnv2.ConvNeXtV2Tiny(input_shape=(img_size[0], img_size[1], 3), num_classes=0, pretrained="imagenet")
        else:
            backbone = cnv2.ConvNeXtV2Base(input_shape=(img_size[0], img_size[1], 3), num_classes=0, pretrained="imagenet")
        backbone.trainable = False
    elif backbone_name.startswith('maxvit_'):
        from keras_cv_attention_models import maxvit
        if backbone_name.endswith('tiny'):
            backbone = maxvit.MaxViTTiny(input_shape=(img_size[0], img_size[1], 3), num_classes=0, pretrained="imagenet")
        else:
            backbone = maxvit.MaxViTSmall(input_shape=(img_size[0], img_size[1], 3), num_classes=0, pretrained="imagenet")
        backbone.trainable = False
    else:
        raise ValueError("Choose a supported pretrained backbone.")

    if needs_m1_1:
        from tensorflow.keras.layers import Rescaling
        x = Rescaling(2.0, offset=-1.0, name="to_m1_1")(x)

    feat = backbone(x)
    if isinstance(feat, (list, tuple)):
        feat = feat[-1]

    gap = GlobalAveragePooling2D(name="GAP")(feat)

    # ---- Larger head for more learning capacity (still within exact budget) ----
    h = Dense(512, use_bias=True)(gap)         # was 256
    h = Activation(gelu)(h)
    h = Dropout(0.40)(h)
    h = Dense(256, use_bias=True)(h)
    h = Activation(gelu)(h)
    h = Dropout(0.30)(h)
    out_logits = Dense(num_classes, activation='softmax', dtype='float32', name="predictions")(h)

    model = Model(inputs, out_logits, name=f"{backbone_name}_head")

    # Enforce EXACT trainable params = 1,461,587
    TARGET = _desired_param_count()
    base_trainables = _trainable_params_count(model)
    print(f"[ParamTarget] Trainable BEFORE pad: {base_trainables:,}")
    if base_trainables > TARGET:
        raise ValueError(f"Base ({base_trainables}) exceeds TARGET ({TARGET}). Reduce head width.")
    need = TARGET - base_trainables
    if need > 0:
        out2 = ParamPad(n_params=need, name="param_pad")(out_logits)
        model = Model(inputs, out2, name=model.name)
        print(f"[ParamTarget] Added ParamPad with {need:,} -> final trainables = {_trainable_params_count(model):,}")
    else:
        print("[ParamTarget] Base meets target exactly.")
    return model

tf.keras.backend.clear_session()
model = build_model_with_backbone(BACKBONE_NAME, num_classes, IMG_SIZE)
model.summary()

# ========================
# 6) Optimizer, schedule, loss (dtype-safe)
# ========================
@tf.keras.utils.register_keras_serializable(package="Custom")
class WarmupCosine(tf.keras.optimizers.schedules.LearningRateSchedule):
    def __init__(self, base_lr=3e-4, warmup_steps=1000, decay_steps=10000, min_lr=1e-6, name=None):
        super().__init__()
        self.base_lr=float(base_lr); self.warmup_steps=int(warmup_steps)
        self.decay_steps=int(decay_steps); self.min_lr=float(min_lr); self.name=name or "WarmupCosine"
    def __call__(self, step):
        step = tf.cast(step, tf.float32)
        warm = tf.minimum(1.0, step / float(self.warmup_steps))
        prog = tf.minimum(1.0, tf.maximum(0.0, (step - self.warmup_steps) / float(self.decay_steps)))
        cosine = 0.5 * (1.0 + tf.cos(np.pi * prog))
        return self.min_lr + (self.base_lr - self.min_lr) * warm * cosine
    def get_config(self):
        return {"base_lr":self.base_lr,"warmup_steps":self.warmup_steps,"decay_steps":self.decay_steps,
                "min_lr":self.min_lr,"name":self.name}

steps_per_epoch = len(train_gen)
total_steps  = steps_per_epoch * EPOCHS
warmup_steps = int(0.10 * total_steps)
decay_steps  = max(1, total_steps - warmup_steps)

lr_schedule = WarmupCosine(base_lr=INIT_LR, warmup_steps=warmup_steps, decay_steps=decay_steps, min_lr=1e-6)
opt = tf.keras.optimizers.AdamW(learning_rate=lr_schedule, weight_decay=WEIGHT_DECAY, global_clipnorm=1.0)

@tf.keras.utils.register_keras_serializable(package="Custom", name="CCEWithDynamicSmoothing")
def cce_with_dynamic_smoothing(y_true, y_pred):
    # dtype-safe under mixed precision
    y_true = tf.cast(y_true, tf.float32)
    y_pred = tf.cast(y_pred, tf.float32)
    eps = tf.cast(smooth_var, tf.float32)
    k = tf.cast(tf.shape(y_true)[-1], tf.float32)
    y_true_smooth = (1.0 - eps) * y_true + eps / k
    return tf.keras.losses.categorical_crossentropy(y_true_smooth, y_pred)

smooth_var = tf.Variable(LABEL_SMOOTH, trainable=False, dtype=tf.float32)
model.compile(optimizer=opt, loss=cce_with_dynamic_smoothing, metrics=['accuracy'])

# ========================
# 7) MixUp / CutMix (aligned)
# ========================
def _sample_batch_idx(gen, idx):
    x, y = gen[idx]
    return x.astype(np.float32), y.astype(np.float32)

def _align_batches(a, b):
    ax, ay = a; bx, by = b
    n = min(ax.shape[0], bx.shape[0])
    if n < 2: return (ax, ay), None
    return (ax[:n], ay[:n]), (bx[:n], by[:n])

def _mixup_aligned(a, b, alpha=0.3):
    (ax, ay), (bx, by) = a, b
    lam = np.random.beta(alpha, alpha)
    x = lam*ax + (1-lam)*bx
    y = lam*ay + (1-lam)*by
    return x, y

def _cutmix_aligned(a, b):
    (ax, ay), (bx, by) = a, b
    bs, h, w, c = ax.shape
    rx = np.random.randint(w//8, w - w//8)
    ry = np.random.randint(h//8, h - h//8)
    rw = np.random.randint(w//8, w//3)
    rh = np.random.randint(h//8, h//3)
    x = ax.copy()
    x[:, ry:ry+rh, rx:rx+rw, :] = bx[:, ry:ry+rh, rx:rx+rw, :]
    lam_adj = 1.0 - (rw*rh)/(w*h)
    y = lam_adj*ay + (1-lam_adj)*by
    return x, y

class MixGenSequence(tf.keras.utils.Sequence):
    def __init__(self, base_gen, prob_mixup_ref, prob_cutmix_ref, **kwargs):
        super().__init__(**kwargs)
        self.base = base_gen
        self.length = len(base_gen)
        self.batch_size = base_gen.batch_size
        self.num_classes = base_gen.num_classes
        self.prob_mixup_ref = prob_mixup_ref
        self.prob_cutmix_ref = prob_cutmix_ref
    def __len__(self): return self.length
    def __getitem__(self, idx):
        a = _sample_batch_idx(self.base, idx % self.length)
        r = np.random.rand()
        pmx = self.prob_mixup_ref[0]; pcm = self.prob_cutmix_ref[0]
        if r < (pcm + pmx):
            idx2 = (idx + np.random.randint(1, self.length)) % self.length
            b = _sample_batch_idx(self.base, idx2)
            a_aligned, b_aligned = _align_batches(a, b)
            if b_aligned is None: return a
            if r < pcm: return _cutmix_aligned(a_aligned, b_aligned)
            else:       return _mixup_aligned(a_aligned, b_aligned, alpha=0.3)
        else:
            return a
    def on_epoch_end(self):
        if hasattr(self.base, 'on_epoch_end'): self.base.on_epoch_end()

mix_train = MixGenSequence(train_gen, MIXUP_P, CUTMIX_P)

# ========================
# 8) Callbacks
# ========================
early_stop = EarlyStopping(monitor='val_accuracy', patience=12, restore_best_weights=True, verbose=1)
ckpt_std_path = "/kaggle/working/best_backbone_model.keras"
ckpt_std  = ModelCheckpoint(ckpt_std_path, monitor='val_accuracy', save_best_only=True,
                            save_weights_only=False, verbose=1)

class AnnealEverything(Callback):
    def __init__(self, start_epoch=2, mid_epoch=12, end_epoch=40):
        super().__init__(); self.start=start_epoch; self.mid=mid_epoch; self.end=end_epoch
    def on_epoch_begin(self, epoch, logs=None):
        if epoch < self.start:
            MIXUP_P[0]=0.05;  CUTMIX_P[0]=0.05;  ERASE_P[0]=0.00; CLAHE_P[0]=0.04
            smooth_var.assign(LABEL_SMOOTH)
        elif epoch < self.mid:
            MIXUP_P[0]=0.12; CUTMIX_P[0]=0.12; ERASE_P[0]=0.06; CLAHE_P[0]=0.06
            smooth_var.assign(0.02)
        elif epoch < self.end:
            MIXUP_P[0]=0.18; CUTMIX_P[0]=0.18; ERASE_P[0]=0.08; CLAHE_P[0]=0.08
            smooth_var.assign(0.01)
        else:
            MIXUP_P[0]=0.00; CUTMIX_P[0]=0.00; ERASE_P[0]=0.00; CLAHE_P[0]=0.00
            smooth_var.assign(0.00)

anneal_cb = AnnealEverything()

# ========================
# 9) Train (50 epochs)
# ========================
history = model.fit(
    mix_train,
    validation_data=val_gen,
    epochs=EPOCHS,
    callbacks=[anneal_cb, early_stop, ckpt_std],
    class_weight=class_weight,
    verbose=1
)

# Curves
plt.figure(figsize=(8,5)); plt.plot(history.history['accuracy'], label='Train Acc', marker='o')
plt.plot(history.history['val_accuracy'], label='Val Acc', marker='o')
plt.title('Accuracy'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.show()
plt.figure(figsize=(8,5)); plt.plot(history.history['loss'], label='Train Loss', marker='o')
plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
plt.title('Loss'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.show()

# ========================
# 10) Evaluation + 5x TTA (streamed)
# ========================
print("\nEvaluating best-restored model (no reload to avoid custom-objects issues):")
std_model = model

val_gen_plain = val_datagen_plain.flow_from_directory(
    directory=str(val_dir), target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode='categorical', shuffle=False
)
labels = list(val_gen_plain.class_indices.keys())
y_true = val_gen_plain.classes

def tta_preds(model, base_gen, tta_list):
    N = len(base_gen.filenames)
    prob_sum = np.zeros((N, base_gen.num_classes), dtype=np.float32)
    idx0 = 0
    for i in range(len(base_gen)):
        x, _ = base_gen[i]
        bs = x.shape[0]
        acc = np.zeros((bs, base_gen.num_classes), dtype=np.float32)
        for fn in tta_list:
            xb = x if fn is None else np.stack([fn(img) for img in x], axis=0).astype(np.float32)
            p = model.predict(xb, verbose=0)
            acc += p.astype(np.float32)
        acc /= float(len(tta_list))
        prob_sum[idx0:idx0+bs] = acc
        idx0 += bs
    return prob_sum

def tta_hflip(x): return np.ascontiguousarray(np.flip(x, axis=1))
def tta_rotate(x, deg=8):
    h, w = x.shape[:2]; M = cv2.getRotationMatrix2D((w/2, h/2), deg, 1.0)
    return cv2.warpAffine(x, M, (w, h), borderMode=cv2.BORDER_REFLECT_101)
def tta_gamma(x, gamma=1.05):
    x = np.clip(x, 0, 1); return np.power(x, 1.0/gamma)

tta_list = [None, tta_hflip, lambda x: tta_rotate(x, +8), lambda x: tta_rotate(x, -8), lambda x: tta_gamma(x, 1.05)]

probs = tta_preds(std_model, val_gen_plain, tta_list)
y_pred = probs.argmax(axis=1)
acc = (y_pred == y_true).mean()
print(f"\nValidation accuracy ({BACKBONE_NAME} + larger head + 5xTTA): {acc*100:.2f}%")

cm_val = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(1.2*len(labels), 1.0*len(labels)))
sns.heatmap(cm_val, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted'); plt.ylabel('Actual'); plt.title(f'Confusion Matrix (VAL, {BACKBONE_NAME}+TTA)'); plt.show()
print("\nClassification Report (VAL):\n")
print(classification_report(y_true, y_pred, target_names=labels, digits=4))

# ========================
# 11) Save final model
# ========================
final_model_path = f"/kaggle/working/{BACKBONE_NAME}_head_paramfixed.keras"
std_model.save(final_model_path)
print("Saved base model to:", final_model_path)
```
